# Decoder Only Transformer

# **Steps of a Transformer**

---

## 🔹 Full Transformer Forward Pass

### **1. Embeddings**
Turn tokens (IDs) into vectors:

$$
H^{(0)} = \text{Embed}(\text{tokens}) + \text{PositionalEncodings}
$$

- `Embed(tokens)` = learned embedding lookup table (trainable).  
- `PositionalEncodings` = adds information about token order (fixed sinusoidal or learned).  

---

### **2. Repeated Transformer Blocks**

For each layer \( l = 1, …, L \):

1. **Normalize**  
$$
X = \text{LayerNorm}(H^{(l-1)})
$$

2. **Attention**  
$$
A = \text{MultiHeadAttention}(X)
$$  

Multi-head attention math:  

$$
Q = X W_Q, \quad K = X W_K, \quad V = X W_V
$$  

$$
\text{Attention}(Q,K,V) = \text{softmax}\!\left(\frac{QK^\top}{\sqrt{d}}\right)V
$$  

$$
\text{MHA}(X) = \text{Concat}(O_1, ..., O_h) W_O
$$

3. **Residual Connection**  
$$
H' = H^{(l-1)} + A
$$

4. **Normalize**  
$$
X = \text{LayerNorm}(H')
$$

5. **Feed-Forward Network (FFN)**  
$$
F = W_2 \, f(W_1 X + b_1) + b_2
$$  

6. **Residual Connection**  
$$
H^{(l)} = H' + F
$$

---

### **3. Final Hidden States**
After \(L\) layers:

$$
H^{(L)} \in \mathbb{R}^{T \times d}
$$

- Each row = final representation of one token.  
- \(T\) = number of tokens, \(d\) = hidden dimension.

---

### **4. Prediction Head**
Take the hidden state of the last token $h_T^{(L)}$:

$$
z = h_T^{(L)} W_{out} + b_{out}
$$

---

### **5. Softmax**
Convert logits \(z\) into probabilities for each vocab word:

$$
P(\text{next token}=v) = \frac{e^{z_v}}{\sum_j e^{z_j}}
$$

---

## 🔹 Plain English

1. Look up embeddings for each token, add position info.  
2. Send them through many transformer blocks (each block = normalize → attention → add back → normalize → feed-forward → add back).  
3. After the last block, you get the final “meaning vectors” for each token.  
4. Pick the last one, run it through a simple linear layer to turn it into scores for every word in the vocabulary.  
5. Apply softmax → probabilities for the next token.  

---

💡 **Key Notes**  
- The same equations are applied at each layer, but with different learned weights.  
- Attention = tokens “talk” to each other; FFN = tokens “think privately.”  
- Residuals ensure information flows through layers without vanishing.  
- During training, embeddings and all weights start random and are learned via backprop.

```python
# math-wise GPT, no fancy libs (just PyTorch)
import math, torch
import torch.nn as nn
import torch.nn.functional as F

# ---------------------------
# tiny tokenizer (character-level for demo)
# ---------------------------
class CharTokenizer:
    def __init__(self, text):
        chars = sorted(list(set(text)))
        self.stoi = {ch:i for i,ch in enumerate(chars)}
        self.itos = {i:ch for ch,i in self.stoi.items()}
        self.vocab_size = len(chars)

    def encode(self, s):  # str -> list[int]
        return [self.stoi[ch] for ch in s]
    def decode(self, ids):  # list[int] -> str
        return "".join(self.itos[i] for i in ids)

# ---------------------------
# masked multi-head self-attention
# A = softmax((Q K^T)/sqrt(d_head)) V
# ---------------------------
class MultiHeadSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, max_len):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.d_head = d_model // n_heads

        # learned projection matrices for Q, K, V, and output (W_Q, W_K, W_V, W_O)
        self.W_Q = nn.Linear(d_model, d_model, bias=False)
        self.W_K = nn.Linear(d_model, d_model, bias=False)
        self.W_V = nn.Linear(d_model, d_model, bias=False)
        self.W_O = nn.Linear(d_model, d_model, bias=False)

        # causal mask (upper triangle = -inf so j>i is masked)
        mask = torch.triu(torch.ones(max_len, max_len), diagonal=1).bool()
        self.register_buffer("causal_mask", mask)  # shape (T,T), not a parameter

    def forward(self, H):  # H: (B, T, d_model)
        B, T, D = H.shape
        # projections
        Q = self.W_Q(H)  # (B, T, D)
        K = self.W_K(H)  # (B, T, D)
        V = self.W_V(H)  # (B, T, D)

        # split into heads: (B, T, n_heads, d_head) -> (B, n_heads, T, d_head)
        def split_heads(x):
            return x.view(B, T, self.n_heads, self.d_head).transpose(1, 2)
        Q = split_heads(Q)
        K = split_heads(K)
        V = split_heads(V)

        # attention scores: (B, n_heads, T, T)
        scores = (Q @ K.transpose(-2, -1)) / math.sqrt(self.d_head)
        # apply causal mask so positions can only attend to <= current index
        scores = scores.masked_fill(self.causal_mask[:T, :T], float("-inf"))

        # attention weights
        A = F.softmax(scores, dim=-1)
        # weighted sum of values -> (B, n_heads, T, d_head)
        O = A @ V
        # concat heads -> (B, T, D)
        O = O.transpose(1, 2).contiguous().view(B, T, D)
        # final output projection
        return self.W_O(O)

# ---------------------------
# position-wise feed-forward network:
# FFN(x) = W2 f(W1 x + b1) + b2
# ---------------------------
class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.0):
        super().__init__()
        self.W1 = nn.Linear(d_model, d_ff)
        self.W2 = nn.Linear(d_ff, d_model)
        self.act = nn.GELU()
        self.drop = nn.Dropout(dropout)

    def forward(self, x):
        return self.drop(self.W2(self.act(self.W1(x))))

# ---------------------------
# one Transformer decoder block (Pre-LN):
# h1 = x + MHA(LN(x))
# h2 = h1 + FFN(LN(h1))
# ---------------------------
class DecoderBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, max_len, dropout=0.0):
        super().__init__()
        self.ln1 = nn.LayerNorm(d_model)              # LayerNorm γ,β are learned params
        self.mha = MultiHeadSelfAttention(d_model, n_heads, max_len)
        self.ln2 = nn.LayerNorm(d_model)
        self.ffn = FeedForward(d_model, d_ff, dropout)
        self.drop = nn.Dropout(dropout)

    def forward(self, x):
        # attention sublayer + residual
        a = self.mha(self.ln1(x))
        x = x + self.drop(a)
        # feed-forward sublayer + residual
        f = self.ffn(self.ln2(x))
        x = x + self.drop(f)
        return x

# ---------------------------
# GPT (decoder-only transformer):
# H0 = Embed(tokens) + PosEmbed
# repeat blocks
# logits = LN(H_L) @ W_out  (weight-tying with token embeddings is common)
# ---------------------------
class TinyGPT(nn.Module):
    def __init__(self, vocab_size, max_len=256, d_model=256, n_layers=4, n_heads=4, d_ff=1024, dropout=0.0, tie_weights=True):
        super().__init__()
        self.vocab_size = vocab_size
        self.max_len = max_len
        self.d_model = d_model

        # token & positional embeddings (learned parameters)
        self.token_embed = nn.Embedding(vocab_size, d_model)
        self.pos_embed   = nn.Embedding(max_len, d_model)

        # stack of decoder blocks
        self.blocks = nn.ModuleList([
            DecoderBlock(d_model, n_heads, d_ff, max_len, dropout)
            for _ in range(n_layers)
        ])
        self.ln_f = nn.LayerNorm(d_model)
        self.head = nn.Linear(d_model, vocab_size, bias=False)

        # (optional) weight tying: W_out = token_embed^T
        if tie_weights:
            self.head.weight = self.token_embed.weight

        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, (nn.Linear, nn.Embedding)):
            nn.init.normal_(m.weight, mean=0.0, std=0.02)
        if isinstance(m, nn.Linear) and m.bias is not None:
            nn.init.zeros_(m.bias)

    def forward(self, idx, targets=None):
        """
        idx: (B, T) int token IDs
        targets: (B, T) next-token IDs for loss (optional)
        returns: logits (B, T, V) and loss if targets provided
        """
        B, T = idx.shape
        assert T <= self.max_len, "sequence too long"

        # embeddings + positions  -> H^(0)
        tok = self.token_embed(idx)                              # (B, T, d)
        pos = self.pos_embed(torch.arange(T, device=idx.device)) # (T, d)
        H = tok + pos                                           # (B, T, d)

        # pass through L decoder blocks
        for blk in self.blocks:
            H = blk(H)

        # final norm and output projection -> logits
        H = self.ln_f(H)                                        # (B, T, d)
        logits = self.head(H)                                   # (B, T, V)

        loss = None
        if targets is not None:
            # classic next-token loss: shift logits and targets by 1
            # predict x_{t+1} from hidden at position t
            logits_flat = logits[:, :-1, :].reshape(-1, self.vocab_size)
            targets_flat = targets[:, 1:].reshape(-1)
            loss = F.cross_entropy(logits_flat, targets_flat)
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens=50, temperature=1.0, top_k=None):
        """
        Autoregressive generation (no KV cache for simplicity).
        idx: (B, T) current tokens
        """
        for _ in range(max_new_tokens):
            # crop to max_len from the right
            idx_cond = idx[:, -self.max_len:]
            logits, _ = self.forward(idx_cond)
            # take last time step's logits
            last_logits = logits[:, -1, :] / max(1e-6, temperature)

            if top_k is not None:
                # keep only top_k tokens
                v, ix = torch.topk(last_logits, top_k, dim=-1)
                out = torch.full_like(last_logits, float('-inf'))
                out.scatter_(dim=-1, index=ix, src=v)
                last_logits = out

            probs = F.softmax(last_logits, dim=-1)
            next_id = torch.multinomial(probs, num_samples=1)  # (B, 1)
            idx = torch.cat([idx, next_id], dim=1)
        return idx
```

## Demo

```python
# 1) Build a toy dataset and tokenizer
corpus = (
    "Once upon a time there was a cat.\n"
    "The cat liked to sleep.\n"
    "The cat liked to eat.\n"
    "The cat was very happy.\n"
) * 200
text = "my head hurts a lot. my head hurts a little. my head hurts!\n"
tok = CharTokenizer(corpus)
ids = torch.tensor([tok.encode(corpus)], dtype=torch.long)  # (1, T)

# 2) Make the model
device = "cuda" if torch.cuda.is_available() else "cpu"
model = TinyGPT(
    vocab_size=tok.vocab_size,
    max_len=128,
    d_model=256,
    n_layers=4,
    n_heads=4,
    d_ff=4*256,
    dropout=0.1,
    tie_weights=True
).to(device)

# 3) A super-mini "training" loop (just to show the loss goes down)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4)
model.train()
for step in range(200):
    idx = ids[:, :128].to(device)
    logits, loss = model(idx, targets=idx)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if step % 50 == 0:
        print(f"step {step:4d}  loss {loss.item():.3f}")

# 4) Generate
model.eval()
ctx = "the cat was "
ctx_ids = torch.tensor([tok.encode(ctx)], dtype=torch.long, device=device)
gen_ids = model.generate(ctx_ids, max_new_tokens=50, temperature=0.9, top_k=tok.vocab_size - 1)
print(tok.decode(gen_ids[0].tolist()))
```

## Numpy GPT Model

```python
# =========================
# Tiny GPT in pure JAX + Optax
# =========================
import math, functools, numpy as np
import jax
import jax.numpy as jnp
import optax

# -------------------------
# Tokenizer (character-level for simplicity)
# -------------------------
class CharTok:
    def __init__(self, text):
        vocab = sorted(set(text))
        self.stoi = {ch:i for i,ch in enumerate(vocab)}
        self.itos = {i:ch for ch,i in self.stoi.items()}
        self.vocab_size = len(vocab)
    def encode(self, s): return np.array([self.stoi[ch] for ch in s], dtype=np.int32)
    def decode(self, ids): return "".join(self.itos[int(i)] for i in ids)

# -------------------------
# Utility functions
# -------------------------
def gelu(x):
    return 0.5 * x * (1.0 + jnp.tanh(jnp.sqrt(2.0/jnp.pi) * (x + 0.044715 * x**3)))

def layernorm(x, gamma, beta, eps=1e-5):
    # normalize over last dim
    mu = jnp.mean(x, axis=-1, keepdims=True)
    var = jnp.mean((x - mu)**2, axis=-1, keepdims=True)
    y = (x - mu) / jnp.sqrt(var + eps)
    return y * gamma + beta

def causal_mask(T):
    # True where j > i (mask future), shape (1,1,T,T) to broadcast across batch/heads
    m = jnp.triu(jnp.ones((T, T), dtype=bool), k=1)
    return m[None, None, :, :]

# -------------------------
# Parameter initialization
# -------------------------
def normal_init(key, shape, std=0.02):
    return std * jax.random.normal(key, shape)

def init_params(rng, vocab_size, max_len, d_model, n_layers, n_heads, d_ff, tie_weights=True):
    keys = jax.random.split(rng, 3 + n_layers*8)
    k_tok, k_pos, k_out = keys[:3]
    idx = 3

    params = {}
    params['tok_embed'] = normal_init(k_tok,  (vocab_size, d_model))
    params['pos_embed'] = normal_init(k_pos,  (max_len,   d_model))

    params['blocks'] = []
    for _ in range(n_layers):
        kWQ, kWK, kWV, kWO = keys[idx:idx+4]; idx += 4
        kW1, kb1, kW2, kb2 = keys[idx:idx+4]; idx += 4

        block = {
            'ln1_g': jnp.ones((d_model,)),
            'ln1_b': jnp.zeros((d_model,)),
            'W_Q':   normal_init(kWQ, (d_model, d_model)),
            'W_K':   normal_init(kWK, (d_model, d_model)),
            'W_V':   normal_init(kWV, (d_model, d_model)),
            'W_O':   normal_init(kWO, (d_model, d_model)),
            'ln2_g': jnp.ones((d_model,)),
            'ln2_b': jnp.zeros((d_model,)),
            'W1':    normal_init(kW1, (d_model, d_ff)),
            'b1':    jnp.zeros((d_ff,)),
            'W2':    normal_init(kW2, (d_ff, d_model)),
            'b2':    jnp.zeros((d_model,))
        }
        params['blocks'].append(block)

    params['ln_f_g'] = jnp.ones((d_model,))
    params['ln_f_b'] = jnp.zeros((d_model,))
    W_out = normal_init(k_out, (d_model, vocab_size))
    if tie_weights:
        W_out = params['tok_embed'].T
    params['W_out'] = W_out
    return params

# -------------------------
# Multi-Head Self-Attention (masked)
# -------------------------
@functools.partial(jax.jit, static_argnames=('n_heads',))
def mha_forward(H, W_Q, W_K, W_V, W_O, n_heads, mask_4d):
    # H: (B, T, d), weights (d,d)
    B, T, d = H.shape
    d_head = d // n_heads

    Q = H @ W_Q  # (B,T,d)
    K = H @ W_K
    V = H @ W_V

    # reshape into heads: (B, h, T, d_head)
    def split_heads(x):
        x = x.reshape(B, T, n_heads, d_head)
        return jnp.transpose(x, (0, 2, 1, 3))
    Qh, Kh, Vh = split_heads(Q), split_heads(K), split_heads(V)

    # scores: (B, h, T, T)
    scores = jnp.einsum('bhtd,bhsd->bhts', Qh, Kh) / jnp.sqrt(d_head)

    # apply causal mask: set masked positions to very negative
    scores = jnp.where(mask_4d, jnp.full_like(scores, -1e9), scores)

    A = jax.nn.softmax(scores, axis=-1)        # (B,h,T,T)
    Oh = jnp.einsum('bhts,bhsd->bhtd', A, Vh)  # (B,h,T,d_head)

    # concat heads -> (B,T,d)
    O = jnp.transpose(Oh, (0, 2, 1, 3)).reshape(B, T, d)
    return O @ W_O  # output projection

# -------------------------
# Feed-Forward
# -------------------------
def ffn_forward(X, W1, b1, W2, b2):
    return (gelu(X @ W1 + b1)) @ W2 + b2

# -------------------------
# One decoder block (Pre-LN): x + MHA(LN(x)) ; x + FFN(LN(x))
# -------------------------
@functools.partial(jax.jit, static_argnames=('n_heads',))
def decoder_block_forward(H, block, n_heads, mask_4d):
    # LN -> MHA -> Residual
    X = layernorm(H, block['ln1_g'], block['ln1_b'])
    A = mha_forward(X, block['W_Q'], block['W_K'], block['W_V'], block['W_O'], n_heads, mask_4d)
    H = H + A
    # LN -> FFN -> Residual
    X = layernorm(H, block['ln2_g'], block['ln2_b'])
    F = ffn_forward(X, block['W1'], block['b1'], block['W2'], block['b2'])
    H = H + F
    return H

# -------------------------
# Model forward
# -------------------------
@functools.partial(jax.jit, static_argnames=('n_heads',))
def forward_logits(params, tokens, n_heads):
    """
    tokens: (B,T) int32
    returns: logits (B,T,V), final hidden (B,T,d)
    """
    B, T = tokens.shape
    d = params['tok_embed'].shape[1]
    V = params['W_out'].shape[1]
    # embeddings + positions
    tok = params['tok_embed'][tokens]                 # (B,T,d)
    pos = params['pos_embed'][jnp.arange(T)]          # (T,d)
    H = tok + pos                                     # (B,T,d)

    # causal mask for attention, broadcast over (B,h)
    m4d = causal_mask(T)

    # transformer blocks
    for block in params['blocks']:
        H = decoder_block_forward(H, block, n_heads, m4d)

    # final norm + output projection
    Hf = layernorm(H, params['ln_f_g'], params['ln_f_b'])  # (B,T,d)
    logits = jnp.einsum('btd,dv->btv', Hf, params['W_out'])  # (B,T,V)
    return logits, Hf

# -------------------------
# Loss (next-token cross-entropy, shift by 1)
# -------------------------
def next_token_loss(params, tokens, n_heads):
    logits, _ = forward_logits(params, tokens, n_heads)  # (B,T,V)
    # predict x_t from position t-1
    logits = logits[:, :-1, :]             # (B,T-1,V)
    targets = tokens[:, 1:]                # (B,T-1)

    logprobs = jax.nn.log_softmax(logits, axis=-1)
    nll = -jnp.take_along_axis(logprobs, targets[..., None], axis=-1).squeeze(-1)  # (B,T-1)
    return nll.mean()

# -------------------------
# Training step (Optax AdamW)
# -------------------------
@functools.partial(jax.jit, static_argnames=('n_heads', 'optimizer'))
def train_step(params, opt_state, batch_tokens, n_heads, optimizer):
    loss_fn = lambda p: next_token_loss(p, batch_tokens, n_heads)
    loss, grads = jax.value_and_grad(loss_fn)(params)
    updates, opt_state = optimizer.update(grads, opt_state, params)
    params = optax.apply_updates(params, updates)
    return params, opt_state, loss

# -------------------------
# Simple data batching
# -------------------------
def make_batches(ids, seq_len, batch_size, steps, rng):
    N = len(ids)
    for _ in range(steps):
        starts = jax.random.randint(rng, (batch_size,), 0, max(1, N-seq_len-1))
        batch = np.stack([ids[s:s+seq_len] for s in np.array(starts)], axis=0)  # (B,T)
        yield batch.astype(np.int32)

# -------------------------
# Autoregressive generation
# -------------------------
@functools.partial(jax.jit, static_argnames=('n_heads',))
def last_step_probs(params, tokens, n_heads, temperature=1.0):
    logits, _ = forward_logits(params, tokens, n_heads)   # (B,T,V)
    last = logits[:, -1, :] / temperature
    return jax.nn.softmax(last, axis=-1)                  # (B,V)

def generate(params, prefix_ids, n_heads, max_new_tokens, temperature=1.0, top_k=None, key=None, max_len=256):
    ids = list(prefix_ids)
    key = jax.random.PRNGKey(0) if key is None else key
    for _ in range(max_new_tokens):
        ctx = np.array(ids[-max_len:], dtype=np.int32)[None, :]  # (1,T)
        probs = np.array(last_step_probs(params, ctx, n_heads, temperature))[0]  # (V,)
        if top_k is not None:
            # Ensure top_k is within bounds of the vocabulary size
            actual_top_k = min(top_k, probs.shape[0] - 1)
            if actual_top_k < 1: # Handle the case where top_k is 0 or 1 and vocab is small
                 actual_top_k = 1
            top_idx = np.argpartition(-probs, actual_top_k)[:actual_top_k]
            mask = np.ones_like(probs, dtype=bool)
            mask[top_idx] = False
            probs = np.where(mask, 0.0, probs)
            probs = probs / probs.sum()
        key, sk = jax.random.split(key)
        nxt = int(jax.random.choice(sk, probs.shape[0], p=probs))
        ids.append(nxt)
    return ids

# =========================
# Demo run
# =========================
# Tiny repetitive corpus (learns quickly)
corpus = (
    "Once upon a time there was a cat.\n"
    "The cat liked to sleep.\n"
    "The cat liked to eat.\n"
    "The cat was very happy.\n"
) * 200
tok = CharTok(corpus)
ids = tok.encode(corpus)

# Model/config
max_len   = 128
d_model   = 192
n_layers  = 3
n_heads   = 3
d_ff      = 4 * d_model
vocab     = tok.vocab_size
tie_out   = True

rng = jax.random.PRNGKey(42)
params = init_params(rng, vocab, max_len, d_model, n_layers, n_heads, d_ff, tie_weights=tie_out)

# Optimizer
lr = 3e-4
weight_decay = 0.01
optimizer = optax.adamw(lr, weight_decay=weight_decay)
opt_state = optimizer.init(params) # Initialize optimizer state here

# Training loop (few steps just to see loss drop)
seq_len   = 128
batch_sz  = 32
steps     = 400
data_rng  = jax.random.PRNGKey(123)

for i, batch in enumerate(make_batches(ids, seq_len, batch_sz, steps, data_rng)):
    params, opt_state, loss = train_step(params, opt_state, batch, n_heads, optimizer)
    if i % 50 == 0:
        print(f"step {i:4d}  loss {float(loss):.4f}")

# Generate
prefix = "The cat "
gen_ids = generate(params, tok.encode(prefix), n_heads, max_new_tokens=100, temperature=0.9, top_k=40, key=jax.random.PRNGKey(7), max_len=max_len)
print(tok.decode(gen_ids))
```

##Torch Multi Token Prediction

```python
# ============================
# TinyGPT_MTP: multi-token heads
# ============================
import math, torch
import torch.nn as nn
import torch.nn.functional as F

# (Reuses your CharTokenizer, MultiHeadSelfAttention, FeedForward, DecoderBlock)

class TinyGPT_MTP(nn.Module):
    """
    Decoder-only GPT with Multi-Token Prediction (MTP) heads.

    n_out_heads: number of parallel output heads (k=1..n), where head-k predicts x_{t+k}
    lambda_weights: optional list/tuple/tensor of length n_out_heads with per-head loss weights
                    (default = uniform weights = 1.0 for each head)
    tie_first_head: if True, tie head-1 weights to token embeddings (classic weight tying).
                    Other heads remain independent.
    """
    def __init__(
        self,
        vocab_size,
        max_len=256,
        d_model=256,
        n_layers=4,
        n_heads=4,
        d_ff=1024,
        dropout=0.0,
        n_out_heads=3,
        lambda_weights=None,
        tie_first_head=True,
    ):
        super().__init__()
        assert n_out_heads >= 1
        self.vocab_size = vocab_size
        self.max_len = max_len
        self.d_model = d_model
        self.n_out_heads = n_out_heads

        # token & positional embeddings (learned)
        self.token_embed = nn.Embedding(vocab_size, d_model)
        self.pos_embed   = nn.Embedding(max_len, d_model)

        # transformer blocks
        self.blocks = nn.ModuleList([
            DecoderBlock(d_model, n_heads, d_ff, max_len, dropout)
            for _ in range(n_layers)
        ])
        self.ln_f = nn.LayerNorm(d_model)

        # multiple output heads (each: d_model -> vocab)
        self.heads = nn.ModuleList([
            nn.Linear(d_model, vocab_size, bias=False) for _ in range(n_out_heads)
        ])

        # optional weight tying for the *first* (1-ahead) head
        if tie_first_head:
            self.heads[0].weight = self.token_embed.weight

        # per-head loss weights
        if lambda_weights is None:
            lambda_weights = [1.0] * n_out_heads
        assert len(lambda_weights) == n_out_heads
        self.register_buffer(
            "_lambda_weights",
            torch.tensor(lambda_weights, dtype=torch.float32)
        )

        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, (nn.Linear, nn.Embedding)):
            nn.init.normal_(m.weight, mean=0.0, std=0.02)
        if isinstance(m, nn.Linear) and m.bias is not None:
            nn.init.zeros_(m.bias)

    def forward(self, idx, targets=None, return_all_logits=False):
        """
        idx:      (B, T) int token IDs
        targets:  (B, T) int token IDs (optional). If provided, compute summed MTP loss.
        return_all_logits: if True, return list of logits per head; else just head-1 logits.

        Returns:
          - logits (B, T, V) for head-1 if return_all_logits=False
            OR a list [logits_k] of length n_out_heads if True (each (B,T,V))
          - loss (scalar) if targets provided, else None
        """
        B, T = idx.shape
        assert T <= self.max_len, "sequence too long"

        # embeddings + positions
        tok = self.token_embed(idx)                               # (B, T, d)
        pos = self.pos_embed(torch.arange(T, device=idx.device))  # (T, d)
        H = tok + pos                                             # (B, T, d)

        # transformer blocks
        for blk in self.blocks:
            H = blk(H)

        # final norm
        H = self.ln_f(H)  # (B, T, d)

        # compute logits for each head
        logits_all = [head(H) for head in self.heads]  # list of (B, T, V)

        loss = None
        if targets is not None:
            # Sum cross-entropy over k=1..n_out_heads
            total = 0.0
            for k, logits_k in enumerate(logits_all, start=1):
                # head-k predicts x_{t+k} from position t => align:
                # logits_k[:, :-k, :] vs targets[:, k:]
                if T - k <= 0:
                    continue  # not enough context; skip if sequence is too short
                L = logits_k[:, :-k, :].reshape(-1, self.vocab_size)
                Y = targets[:, k:].reshape(-1)
                ce = F.cross_entropy(L, Y)
                total = total + self._lambda_weights[k-1] * ce
            loss = total

        if return_all_logits:
            return logits_all, loss
        else:
            # default: return head-1 logits (standard next-token)
            return logits_all[0], loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens=50, temperature=1.0, top_k=None, use_head_k=1):
        """
        Standard autoregressive generation. By default uses head-1 (next-token head).
        You can set use_head_k>1 to experiment, but typical decoding uses k=1.

        idx: (B, T) current tokens
        """
        assert 1 <= use_head_k <= self.n_out_heads
        head_ix = use_head_k - 1

        for _ in range(max_new_tokens):
            idx_cond = idx[:, -self.max_len:]
            # get logits for the chosen head only (avoid extra compute)
            tok = self.token_embed(idx_cond)
            T = idx_cond.size(1)
            pos = self.pos_embed(torch.arange(T, device=idx.device))
            H = tok + pos
            for blk in self.blocks:
                H = blk(H)
            H = self.ln_f(H)
            last_logits = self.heads[head_ix](H)[:, -1, :] / max(1e-6, temperature)

            if top_k is not None:
                v, ix = torch.topk(last_logits, top_k, dim=-1)
                out = torch.full_like(last_logits, float('-inf'))
                out.scatter_(dim=-1, index=ix, src=v)
                last_logits = out

            probs = F.softmax(last_logits, dim=-1)
            next_id = torch.multinomial(probs, num_samples=1)  # (B, 1)
            idx = torch.cat([idx, next_id], dim=1)
        return idx
```

## Math Heavy Implementation for Clarity

```python
# ============================================================
# TinyMathGPT — a from-scratch, math-first GPT (PyTorch)
# ------------------------------------------------------------
# Goal: make each step of the Transformer obvious:
#  - clear function/class names
#  - explicit shapes on inputs/outputs
#  - comments that tie code to the math equations
#
# Notation used below:
#   B = batch size
#   T = sequence length (# tokens)
#   V = vocabulary size
#   d = hidden dimension (aka d_model)
#   h = # attention heads
#   d_head = d // h
# ============================================================

import math
import torch
import torch.nn as nn
import torch.nn.functional as F


# ------------------------------------------------------------
# 0) A tiny character-level tokenizer for demos
# ------------------------------------------------------------
class CharTokenizer:
    """
    Converts strings to integer token IDs and back, using one token per character.
    """
    def __init__(self, text: str):
        chars = sorted(list(set(text)))
        self.stoi = {ch: i for i, ch in enumerate(chars)}
        self.itos = {i: ch for ch, i in self.stoi.items()}
        self.vocab_size = len(chars)

    def encode(self, s: str) -> list[int]:
        return [self.stoi[ch] for ch in s]

    def decode(self, ids: list[int]) -> str:
        return "".join(self.itos[i] for i in ids)


# ------------------------------------------------------------
# 1) Masked Multi-Head Self-Attention
# ------------------------------------------------------------
class MaskedMultiHeadSelfAttention(nn.Module):
    """
    Implements: softmax( (Q K^T) / sqrt(d_head) ) @ V  with a causal (look-back-only) mask.

    Input:
      H: (B, T, d)  -- hidden states from previous layer

    Trainable parameters (per layer):
      W_Q, W_K, W_V: (d, d)  -- project H->Q, H->K, H->V
      W_O:           (d, d)  -- final output projection (concat heads -> d, then project)

    Output:
      out: (B, T, d)
    """
    def __init__(self, d_model: int, num_heads: int, max_seq_len: int):
        super().__init__()
        assert d_model % num_heads == 0, "d_model must be divisible by num_heads"
        self.d_model = d_model
        self.num_heads = num_heads
        self.d_head = d_model // num_heads

        # Linear projections for Q, K, V, and output (no bias for clarity)
        self.W_Q = nn.Linear(d_model, d_model, bias=False)
        self.W_K = nn.Linear(d_model, d_model, bias=False)
        self.W_V = nn.Linear(d_model, d_model, bias=False)
        self.W_O = nn.Linear(d_model, d_model, bias=False)

        # Causal mask: disallow attention from position i to future positions j>i
        # Shape: (T, T) with True above main diagonal
        mask = torch.triu(torch.ones(max_seq_len, max_seq_len), diagonal=1).bool()
        self.register_buffer("causal_mask_TT", mask)  # non-parameter buffer

    def _split_heads(self, x: torch.Tensor) -> torch.Tensor:
        """
        x: (B, T, d) -> (B, h, T, d_head)
        We simply reshape and permute to view the 'head' axis separately.
        """
        B, T, d = x.shape
        x = x.view(B, T, self.num_heads, self.d_head)   # (B, T, h, d_head)
        x = x.transpose(1, 2)                           # (B, h, T, d_head)
        return x

    def _combine_heads(self, x: torch.Tensor) -> torch.Tensor:
        """
        x: (B, h, T, d_head) -> (B, T, d)
        """
        B, h, T, d_head = x.shape
        x = x.transpose(1, 2).contiguous()              # (B, T, h, d_head)
        x = x.view(B, T, h * d_head)                    # (B, T, d)
        return x

    def forward(self, H: torch.Tensor) -> torch.Tensor:
        """
        H: (B, T, d)
        returns: (B, T, d)
        """
        B, T, d = H.shape

        # 1) Linear projections to queries, keys, values
        #    Q = H W_Q ; K = H W_K ; V = H W_V
        Q = self.W_Q(H)  # (B, T, d)
        K = self.W_K(H)  # (B, T, d)
        V = self.W_V(H)  # (B, T, d)

        # 2) Split into heads: (B, T, d) -> (B, h, T, d_head)
        Q = self._split_heads(Q)
        K = self._split_heads(K)
        V = self._split_heads(V)

        # 3) Scaled dot-product attention per head
        #    scores = (Q @ K^T) / sqrt(d_head)  -> (B, h, T, T)
        scores = Q @ K.transpose(-2, -1) / math.sqrt(self.d_head)

        # 4) Apply causal mask so token i cannot see future tokens j>i
        scores = scores.masked_fill(self.causal_mask_TT[:T, :T], float("-inf"))

        # 5) Convert scores to attention weights with softmax along last dim (over "keys")
        weights = F.softmax(scores, dim=-1)  # (B, h, T, T)

        # 6) Weighted sum of values: (B, h, T, T) @ (B, h, T, d_head) -> (B, h, T, d_head)
        heads_out = weights @ V

        # 7) Re-combine heads and apply output projection W_O
        concat = self._combine_heads(heads_out)  # (B, T, d)
        out = self.W_O(concat)                   # (B, T, d)
        return out


# ------------------------------------------------------------
# 2) Position-wise Feed-Forward Network (FFN)
# ------------------------------------------------------------
class PositionwiseFeedForward(nn.Module):
    """
    Implements: FFN(x) = W2 * GELU(W1 * x + b1) + b2 (independently at each position)

    Input:
      X: (B, T, d)

    Trainable parameters (per layer):
      W1: (d, d_ff), b1: (d_ff,)
      W2: (d_ff, d), b2: (d,)

    Output:
      Y: (B, T, d)
    """
    def __init__(self, d_model: int, d_ff: int, dropout: float = 0.0):
        super().__init__()
        self.W1 = nn.Linear(d_model, d_ff)
        self.W2 = nn.Linear(d_ff, d_model)
        self.act = nn.GELU()
        self.drop = nn.Dropout(dropout)

    def forward(self, X: torch.Tensor) -> torch.Tensor:
        return self.drop(self.W2(self.act(self.W1(X))))


# ------------------------------------------------------------
# 3) One Decoder Block (Pre-LayerNorm variant)
# ------------------------------------------------------------
class DecoderBlock(nn.Module):
    """
    One Transformer decoder block (Pre-LN):
       A) X1 = LayerNorm(H)
          A  = SelfAttention(X1)
          H  = H + A                  # residual add

       B) X2 = LayerNorm(H)
          F  = FFN(X2)
          H  = H + F                  # residual add

    Input/Output:
      H: (B, T, d)
    """
    def __init__(self, d_model: int, num_heads: int, d_ff: int, max_seq_len: int, dropout: float = 0.0):
        super().__init__()
        self.norm_before_attn = nn.LayerNorm(d_model)
        self.self_attn = MaskedMultiHeadSelfAttention(d_model, num_heads, max_seq_len)
        self.norm_before_ffn = nn.LayerNorm(d_model)
        self.ffn = PositionwiseFeedForward(d_model, d_ff, dropout=dropout)
        self.drop = nn.Dropout(dropout)

    def forward(self, H: torch.Tensor) -> torch.Tensor:
        # --- Attention sub-layer ---
        X1 = self.norm_before_attn(H)  # (B, T, d)
        A  = self.self_attn(X1)        # (B, T, d)
        H  = H + self.drop(A)          # residual

        # --- FFN sub-layer ---
        X2 = self.norm_before_ffn(H)   # (B, T, d)
        F  = self.ffn(X2)              # (B, T, d)
        H  = H + self.drop(F)          # residual
        return H


# ------------------------------------------------------------
# 4) TinyMathGPT — full decoder-only Transformer
# ------------------------------------------------------------
class TinyMathGPT(nn.Module):
    """
    GPT-style (decoder-only) language model:

      H^(0) = TokenEmbedding(tokens) + PositionalEmbedding(positions)
      Repeat L times:
        H = DecoderBlock(H)    # Pre-LN: LN -> MHA -> +res -> LN -> FFN -> +res
      H^*   = LayerNorm(H)
      Logits = H^* @ W_out

    Training loss (next-token prediction):
      predict x_{t+1} from hidden at t (shifted cross-entropy)

    Inputs:
      idx:     (B, T) int token IDs
      targets: (B, T) int token IDs (optional; if provided, compute loss)

    Outputs:
      logits: (B, T, V)
      loss:   scalar (if targets provided)
    """
    def __init__(
        self,
        vocab_size: int,
        max_seq_len: int = 256,
        d_model: int = 256,
        num_layers: int = 4,
        num_heads: int = 4,
        d_ff: int = 1024,
        dropout: float = 0.0,
        tie_weights: bool = True
    ):
        super().__init__()
        self.vocab_size = vocab_size
        self.max_seq_len = max_seq_len
        self.d_model = d_model

        # (a) Token + Positional embeddings
        # TokenEmbedding: (V, d), PosEmbedding: (T_max, d)
        self.token_embed = nn.Embedding(vocab_size, d_model)
        self.pos_embed   = nn.Embedding(max_seq_len, d_model)

        # (b) Stack of decoder blocks
        self.blocks = nn.ModuleList([
            DecoderBlock(d_model, num_heads, d_ff, max_seq_len, dropout=dropout)
            for _ in range(num_layers)
        ])

        # (c) Final norm + output projection to vocab logits
        self.final_norm = nn.LayerNorm(d_model)
        self.output_head = nn.Linear(d_model, vocab_size, bias=False)

        # (optional) Weight tying: W_out = token_embed^T
        if tie_weights:
            self.output_head.weight = self.token_embed.weight

        # init weights (simple normal init)
        self.apply(self._init_weights)

    def _init_weights(self, m: nn.Module):
        if isinstance(m, (nn.Linear, nn.Embedding)):
            nn.init.normal_(m.weight, mean=0.0, std=0.02)
        if isinstance(m, nn.Linear) and m.bias is not None:
            nn.init.zeros_(m.bias)

    def forward(self, idx: torch.Tensor, targets: torch.Tensor | None = None):
        """
        Forward pass (training/eval).

        idx:     (B, T) int token IDs
        targets: (B, T) int token IDs (optional for loss)

        returns: (logits, loss)
          logits: (B, T, V)
          loss:   scalar or None
        """
        B, T = idx.shape
        assert T <= self.max_seq_len, f"sequence too long: T={T} > max_seq_len={self.max_seq_len}"

        # ----- Embeddings -----
        # Token embeddings: (B, T, d)
        tok_emb: torch.Tensor = self.token_embed(idx)
        # Positional embeddings: (T, d) -> broadcast to (B, T, d)
        pos_ids = torch.arange(T, device=idx.device)
        pos_emb: torch.Tensor = self.pos_embed(pos_ids)
        H = tok_emb + pos_emb  # (B, T, d) == H^(0)

        # ----- Transformer Blocks -----
        for block in self.blocks:
            H = block(H)  # (B, T, d)

        # ----- Final norm + Output head -----
        H_star = self.final_norm(H)           # (B, T, d)
        logits = self.output_head(H_star)     # (B, T, V)

        # ----- Next-token prediction loss (shifted by 1) -----
        loss = None
        if targets is not None:
            # predict x_{t+1} using hidden at t
            # align logits[:, :-1, :] with targets[:, 1:]
            logits_flat = logits[:, :-1, :].reshape(-1, self.vocab_size)  # ((B*(T-1)), V)
            targets_flat = targets[:, 1:].reshape(-1)                     # ((B*(T-1)),)
            loss = F.cross_entropy(logits_flat, targets_flat)

        return logits, loss

    @torch.no_grad()
    def generate(
        self,
        idx: torch.Tensor,
        max_new_tokens: int = 50,
        temperature: float = 1.0,
        top_k: int | None = None
    ) -> torch.Tensor:
        """
        Autoregressive generation: repeatedly predict the next token and append.

        idx: (B, T_start) int token IDs
        returns: (B, T_start + max_new_tokens) int token IDs
        """
        for _ in range(max_new_tokens):
            # Only feed the last max_seq_len tokens to respect positional table
            idx_cond = idx[:, -self.max_seq_len:]

            # Forward pass to get logits over vocab for each time step
            logits, _ = self.forward(idx_cond)  # logits: (B, T_cond, V)

            # Take the logits at the last time step
            next_token_logits = logits[:, -1, :] / max(1e-6, temperature)  # (B, V)

            # Optional top-k truncation (keep only top_k logits)
            if top_k is not None:
                v, ix = torch.topk(next_token_logits, top_k, dim=-1)
                filtered = torch.full_like(next_token_logits, float("-inf"))
                filtered.scatter_(dim=-1, index=ix, src=v)
                next_token_logits = filtered

            # Convert to probabilities and sample
            probs = F.softmax(next_token_logits, dim=-1)  # (B, V)
            next_id = torch.multinomial(probs, num_samples=1)  # (B, 1)

            # Append the new token
            idx = torch.cat([idx, next_id], dim=1)  # (B, T+1)
        return idx
```

# Multi-headed future token predictor (3 heads)

vanilla gpt 2 model
removed softmax prediction head. froze trunk. then added multiple multi future token prediction heads and trained only those.

```python
# ====== minimal installs ======
!pip -q install transformers datasets torch

# ====== imports ======
import math, torch, torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from dataclasses import dataclass
from typing import Dict, Tuple
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoConfig

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ====== data: small & simple (WikiText-2 raw) ======
raw = load_dataset("wikitext", "wikitext-2-raw-v1")
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # GPT-2 has no pad; use EOS

MAX_LEN = 256

def tokenize_fn(ex):
    # Fixed length => no ragged batches; attention_mask marks real tokens.
    return tokenizer(
        ex["text"],
        padding="max_length",
        truncation=True,
        max_length=MAX_LEN,
    )

tok_train = raw["train"].map(tokenize_fn, batched=True, remove_columns=raw["train"].column_names)
tok_val   = raw["validation"].map(tokenize_fn, batched=True, remove_columns=raw["validation"].column_names)


@dataclass
class CausalBatch:
    input_ids: torch.Tensor
    attention_mask: torch.Tensor

def collate(batch):
    return CausalBatch(
        input_ids=torch.tensor([b["input_ids"] for b in batch], dtype=torch.long),
        attention_mask=torch.tensor([b["attention_mask"] for b in batch], dtype=torch.long),
    )

# small slices for speed; increase for better signal
train_loader = DataLoader(tok_train.select(range(4000)), batch_size=8, shuffle=True,  collate_fn=collate)
val_loader   = DataLoader(tok_val.select(range(1000)),  batch_size=8, shuffle=False, collate_fn=collate)

# ====== model: load GPT-2, freeze trunk, add auxiliary heads (t+1, t+2, t+3) ======
base_name = "gpt2"
cfg = AutoConfig.from_pretrained(base_name, output_hidden_states=False)  # last_hidden_state is sufficient
model = AutoModelForCausalLM.from_pretrained(base_name, config=cfg).to(device)
vocab_size = model.config.vocab_size
hidden_size = model.transformer.wte.embedding_dim  # 768 for gpt2

# freeze all pretrained params
for p in model.parameters():
    p.requires_grad_(False)

class OffsetHeads(nn.Module):
    """
    Simple linear heads that read from the final hidden state at position t
    and predict tokens at offsets t+1, t+2, t+3.
    """
    def __init__(self, hidden_size: int, vocab_size: int, offsets=(1,2,3)):
        super().__init__()
        self.offsets = list(offsets)
        self.heads = nn.ModuleDict({str(k): nn.Linear(hidden_size, vocab_size) for k in self.offsets})

    def forward(self, H: torch.Tensor, attn_mask: torch.Tensor) -> Dict[int, Tuple[torch.Tensor, torch.Tensor]]:
        """
        H: (B, T, d) final hidden states from GPT-2 transformer
        attn_mask: (B, T)
        Returns dict: k -> (logits_k, mask_k),
          where logits_k shape is (B, T-k, V) aligned to predict x_{t+k} from H_t
        """
        B, T, D = H.shape
        out = {}
        for k in self.offsets:
            head = self.heads[str(k)]
            # valid source positions t: 0..T-k-1
            logits = head(H[:, :-k, :])                # (B, T-k, V)
            # mask requires both source pos and target pos to be valid:
            mask = (attn_mask[:, :-k] & attn_mask[:, k:])  # (B, T-k)
            out[k] = (logits, mask)
        return out

aux = OffsetHeads(hidden_size, vocab_size, offsets=(1,2,3)).to(device)
optim = torch.optim.AdamW(aux.parameters(), lr=5e-4)

# ====== helpers: loss/metrics for aux heads ======
def offset_loss_and_metrics(offset_logits_mask, targets):
    """
    offset_logits_mask: dict k -> (logits, mask)
      logits: (B, T-k, V)
      mask:   (B, T-k)
    targets: (B, T) token ids
    Returns total_loss, details[k] = {"loss", "acc@1", "n_tokens"}
    """
    total = 0.0
    details = {}
    for k, (logits, mask) in offset_logits_mask.items():
        tgt = targets[:, k:]                          # (B, T-k)
        B, Tk, V = logits.shape
        m = mask.bool().view(-1)
        if m.sum().item() == 0:
            details[k] = {"loss": 0.0, "acc@1": 0.0, "n_tokens": 0}
            continue
        logits_f = logits.reshape(-1, V)[m]           # (N, V)
        tgt_f    = tgt.reshape(-1)[m]                 # (N,)
        loss = F.cross_entropy(logits_f, tgt_f)
        with torch.no_grad():
            pred = logits_f.argmax(-1)
            acc = (pred == tgt_f).float().mean().item()
        total = total + loss
        details[k] = {"loss": float(loss.item()), "acc@1": acc, "n_tokens": int(m.sum().item())}
    return total, details

@torch.no_grad()
def eval_aux(aux, model, loader):
    model.eval(); aux.eval()
    agg = {1: {"loss_sum":0.0,"correct":0,"count":0},
           2: {"loss_sum":0.0,"correct":0,"count":0},
           3: {"loss_sum":0.0,"correct":0,"count":0}}
    for batch in loader:
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)
        if inp.size(1) < 4:  # needs at least 4 to compute up to k=3
            continue
        # get final hidden states without running lm_head
        H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,d)
        off = aux(H, attn)
        loss, detail = offset_loss_and_metrics(off, inp)
        for k, d in detail.items():
            n = d["n_tokens"]
            agg[k]["loss_sum"] += d["loss"] * n
            agg[k]["correct"]  += int(d["acc@1"] * n)
            agg[k]["count"]    += n
    out = {}
    for k in agg:
        cnt = max(1, agg[k]["count"])
        out[k] = {"loss": agg[k]["loss_sum"]/cnt,
                  "acc@1": agg[k]["correct"]/cnt,
                  "count": agg[k]["count"]}
    return out

# ====== baseline: vanilla GPT-2 for t+1, t+2, t+3 (teacher forcing) ======
@torch.no_grad()
@torch.no_grad()
def eval_vanilla(model, loader):
    """
    Baseline scores for offsets k=1,2,3 using the standard next-token head.
    For each k, we score target x_{t+k} using logits at positions (t + k - 1).
    Shapes are aligned so time length is always T - k.
    """
    model.eval()

    def summarize(logits_k, tgt_k, mask_k):
        m = mask_k.bool().view(-1)
        if m.sum().item() == 0:
            return 0.0, 0.0, 0
        logits_f = logits_k.reshape(-1, logits_k.size(-1))[m]
        tgt_f    = tgt_k.reshape(-1)[m]
        L = F.cross_entropy(logits_f, tgt_f)
        pred = logits_f.argmax(-1)
        acc = (pred == tgt_f).float().mean().item()
        return float(L.item()), acc, int(m.sum().item())

    agg = {1: {"loss_sum":0.0,"correct":0,"count":0},
           2: {"loss_sum":0.0,"correct":0,"count":0},
           3: {"loss_sum":0.0,"correct":0,"count":0}}

    for batch in loader:
        inp  = batch.input_ids.to(device)        # (B, T)
        attn = batch.attention_mask.to(device)   # (B, T)
        out  = model(input_ids=inp, attention_mask=attn)
        logits = out.logits                      # (B, T, V)
        B, T, V = logits.shape

        # handle each offset uniformly
        for k in (1, 2, 3):
            if T <= k:  # not enough time steps to score this offset
                continue
            # logits at positions (t + k - 1), truncated to length T - k
            # Using end index -1 keeps length aligned with tgt (T - k)
            start = k - 1
            logits_k = logits[:, start:-1, :]         # (B, T-k, V)
            tgt_k    = inp[:, k:]                     # (B, T-k)
            mask_k   = attn[:, start:-1] & attn[:, k:]# (B, T-k)

            L, A, N = summarize(logits_k, tgt_k, mask_k)
            agg[k]["loss_sum"] += L * N
            agg[k]["correct"]  += int(A * N)
            agg[k]["count"]    += N

    out = {}
    for k in (1, 2, 3):
        cnt = max(1, agg[k]["count"])
        out[k] = {
            "loss": agg[k]["loss_sum"]/cnt,
            "acc@1": agg[k]["correct"]/cnt,
            "count": agg[k]["count"],
        }
    return out

# ====== training loop: train ONLY the aux heads ======
def train_epoch(model, aux, loader, optim):
    model.eval()   # trunk frozen
    aux.train()
    running = 0.0
    steps = 0
    for steps, batch in enumerate(loader, 1):
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)

        with torch.no_grad():
            H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state

        off = aux(H, attn)
        loss, detail = offset_loss_and_metrics(off, inp)

        optim.zero_grad(set_to_none=True)
        loss.backward()
        optim.step()

        running += float(loss.item())
        if steps % 100 == 0:
            print(f"step {steps:4d} | loss={loss.item():.3f} | details={detail}")
    return running / max(1, steps)

# ====== run: baseline -> train -> compare ======
print(">>> Baseline (vanilla GPT-2)")
base = eval_vanilla(model, val_loader)
for k in sorted(base):
    print(f"vanilla k=+{k}: acc@1={base[k]['acc@1']:.3f}, loss={base[k]['loss']:.3f}, N={base[k]['count']}")

print("\n>>> Train aux heads (frozen trunk)")
EPOCHS = 2  # bump to 3–5 for better signal
for e in range(1, EPOCHS+1):
    tr_loss = train_epoch(model, aux, train_loader, optim)
    metrics = eval_aux(aux, model, val_loader)
    print(f"[epoch {e}] train_loss={tr_loss:.3f} | "
          f"aux +1 acc={metrics[1]['acc@1']:.3f} +2 acc={metrics[2]['acc@1']:.3f} +3 acc={metrics[3]['acc@1']:.3f}")

print("\n>>> Final comparison")
aux_metrics = eval_aux(aux, model, val_loader)
for k in (1,2,3):
    print(f"k=+{k} | AUX acc@1={aux_metrics[k]['acc@1']:.3f} (loss={aux_metrics[k]['loss']:.3f})"
          f"  vs  VANILLA acc@1={base[k]['acc@1']:.3f} (loss={base[k]['loss']:.3f})")
```

# Extended Experiments

## multi head prediction task variant with only 2 heads

keep only k+1 and k+2 heads. frozen trunk. all prediction heads initialized randomly and trained for current task.

calculate training loss and accuracy of predictions. compare with vanilla model.

calculate statistical significance of contrasting results with vanilla.

```python
# ====== minimal installs ======
!pip -q install transformers datasets torch

# ====== imports ======
import time, numpy as np
import math, torch, torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from dataclasses import dataclass
from typing import Dict, Tuple
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoConfig

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ====== data: small & simple (WikiText-2 raw) ======
raw = load_dataset("wikitext", "wikitext-2-raw-v1")
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # GPT-2 has no pad; use EOS

MAX_LEN = 256

def tokenize_fn(ex):
    # Fixed length => no ragged batches; attention_mask marks real tokens.
    return tokenizer(
        ex["text"],
        padding="max_length",
        truncation=True,
        max_length=MAX_LEN,
    )

tok_train = raw["train"].map(tokenize_fn, batched=True, remove_columns=raw["train"].column_names)
tok_val   = raw["validation"].map(tokenize_fn, batched=True, remove_columns=raw["validation"].column_names)

@dataclass
class CausalBatch:
    input_ids: torch.Tensor
    attention_mask: torch.Tensor

def collate(batch):
    return CausalBatch(
        input_ids=torch.tensor([b["input_ids"] for b in batch], dtype=torch.long),
        attention_mask=torch.tensor([b["attention_mask"] for b in batch], dtype=torch.long),
    )

# small slices for speed; increase for better signal
train_loader = DataLoader(tok_train.select(range(4000)), batch_size=8, shuffle=True,  collate_fn=collate)
val_loader   = DataLoader(tok_val.select(range(1000)),  batch_size=8, shuffle=False, collate_fn=collate)

# ====== model: load GPT-2, freeze trunk, add auxiliary heads (t+1, t+2) ======
base_name = "gpt2"
cfg = AutoConfig.from_pretrained(base_name, output_hidden_states=False)  # last_hidden_state is sufficient
model = AutoModelForCausalLM.from_pretrained(base_name, config=cfg).to(device)
vocab_size = model.config.vocab_size
hidden_size = model.transformer.wte.embedding_dim  # 768 for gpt2

# freeze all pretrained params
for p in model.parameters():
    p.requires_grad_(False)

OFFSETS = (1, 2)  # drop +3 everywhere

class OffsetHeads(nn.Module):
    """
    Simple linear heads that read from the final hidden state at position t
    and predict tokens at offsets t+1, t+2.
    """
    def __init__(self, hidden_size: int, vocab_size: int, offsets=OFFSETS):
        super().__init__()
        self.offsets = list(offsets)
        self.heads = nn.ModuleDict({str(k): nn.Linear(hidden_size, vocab_size) for k in self.offsets})

    def forward(self, H: torch.Tensor, attn_mask: torch.Tensor) -> Dict[int, Tuple[torch.Tensor, torch.Tensor]]:
        """
        H: (B, T, d) final hidden states from GPT-2 transformer
        attn_mask: (B, T)
        Returns dict: k -> (logits_k, mask_k),
          where logits_k shape is (B, T-k, V) aligned to predict x_{t+k} from H_t
        """
        B, T, D = H.shape
        out = {}
        for k in self.offsets:
            head = self.heads[str(k)]
            # valid source positions t: 0..T-k-1
            logits = head(H[:, :-k, :])                     # (B, T-k, V)
            # mask requires both source pos and target pos to be valid:
            mask = (attn_mask[:, :-k] & attn_mask[:, k:])   # (B, T-k)
            out[k] = (logits, mask)
        return out

aux = OffsetHeads(hidden_size, vocab_size, offsets=OFFSETS).to(device)
optim = torch.optim.AdamW(aux.parameters(), lr=5e-4)

# ====== helpers: alignment, metrics, significance ======
def _aligned_slices_for_k(logits, inp, attn, k):
    """
    Align logits (vanilla) at positions t+k-1 with targets at t+k.
    Returns: logits_k (B, T-k, V), tgt_k (B, T-k), mask_k (B, T-k)
    """
    B, T, V = logits.shape
    assert T > k, "sequence too short for this offset"
    start = k - 1
    logits_k = logits[:, start:-1, :]          # (B, T-k, V)
    tgt_k    = inp[:, k:]                      # (B, T-k)
    mask_k   = attn[:, start:-1] & attn[:, k:] # (B, T-k)
    return logits_k, tgt_k, mask_k

def _compute_topk_metrics(logits, targets, mask, topks=(1,3,5)):
    """
    logits: (B, T, V), targets: (B, T), mask: (B, T)
    Returns dict: {"loss":..., "hits@1":..., "hits@3":..., "hits@5":..., "count":...}
    """
    V = logits.size(-1)
    m = mask.bool().view(-1)
    if m.sum().item() == 0:
        return {"loss": 0.0, "count": 0, **{f"hits@{k}": 0.0 for k in topks}}
    logits_f = logits.reshape(-1, V)[m]
    tgt_f    = targets.reshape(-1)[m]
    loss     = F.cross_entropy(logits_f, tgt_f).item()
    out = {"loss": loss, "count": int(m.sum().item())}
    for k in topks:
        k = min(k, V)
        topk_idx = logits_f.topk(k, dim=-1).indices
        hits = (topk_idx == tgt_f.unsqueeze(-1)).any(dim=-1).float().mean().item()
        out[f"hits@{k}"] = hits
    return out

def _gather_correct_flags(logits, targets, mask):
    """
    Returns boolean vector (N,) of 'was top1 correct' for significance tests.
    """
    V = logits.size(-1)
    m = mask.bool().view(-1)
    if m.sum().item() == 0:
        return np.zeros((0,), dtype=bool)
    logits_f = logits.reshape(-1, V)[m]
    tgt_f    = targets.reshape(-1)[m]
    pred     = logits_f.argmax(-1)
    return (pred == tgt_f).detach().cpu().numpy().astype(bool)

def mcnemar_exact_p(b, c):
    """
    Exact two-sided McNemar via binomial test on discordant pairs (n=b+c).
    Uses log-space to avoid overflow for large n.
    b = count where method A correct and B wrong
    c = count where method B correct and A wrong
    """
    n = b + c
    if n == 0:
        return 1.0

    k_max = min(b, c)  # smaller side for the two-sided tail
    log_half = math.log(0.5)

    # log C(n, k) via lgamma
    def log_comb(n_, k_):
        return math.lgamma(n_ + 1) - math.lgamma(k_ + 1) - math.lgamma(n_ - k_ + 1)

    # accumulate in log-space: tail = sum_{k=0..k_max} C(n,k) * 0.5^n
    logs = [log_comb(n, k) + n * log_half for k in range(k_max + 1)]
    m = max(logs)
    tail = math.exp(m) * sum(math.exp(li - m) for li in logs)

    p_two_sided = min(1.0, 2.0 * tail)
    return p_two_sided


# ====== eval: AUX (frozen heads) ======
@torch.no_grad()
def eval_aux(aux, model, loader, topks=(1,3,5)):
    model.eval(); aux.eval()
    agg = {k: {"loss_sum":0.0, "count":0, **{f"hits@{t}":0.0 for t in topks}} for k in OFFSETS}
    flags = {k: [] for k in OFFSETS}

    for batch in loader:
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)
        if inp.size(1) <= max(OFFSETS):
            continue

        H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,d)
        out = aux(H, attn)  # dict k -> (logits, mask)

        for k in OFFSETS:
            logits_k, mask_k = out[k]
            tgt_k = inp[:, k:]  # (B, T-k)
            m = _compute_topk_metrics(logits_k, tgt_k, mask_k, topks=topks)
            N = m["count"]
            if N == 0:
                continue
            agg[k]["loss_sum"] += m["loss"] * N
            agg[k]["count"]    += N
            for t in topks:
                agg[k][f"hits@{t}"] += m[f"hits@{t}"] * N
            flags[k].append(_gather_correct_flags(logits_k, tgt_k, mask_k))

    res = {}
    for k in OFFSETS:
        N = max(1, agg[k]["count"])
        res[k] = {
            "loss": agg[k]["loss_sum"] / N,
            **{f"hits@{t}": agg[k][f"hits@{t}"] / N for t in topks},
            "count": agg[k]["count"]
        }
    flags = {k: (np.concatenate(flags[k]) if len(flags[k]) else np.zeros((0,), dtype=bool)) for k in OFFSETS}
    return res, flags

# ====== eval: VANILLA (teacher forced @ offsets) ======
@torch.no_grad()
def eval_vanilla(model, loader, topks=(1,3,5)):
    model.eval()
    agg = {k: {"loss_sum":0.0, "count":0, **{f"hits@{t}":0.0 for t in topks}} for k in OFFSETS}
    flags = {k: [] for k in OFFSETS}

    for batch in loader:
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)
        if inp.size(1) <= max(OFFSETS):
            continue
        out  = model(input_ids=inp, attention_mask=attn)
        logits = out.logits  # (B,T,V)

        for k in OFFSETS:
            logits_k, tgt_k, mask_k = _aligned_slices_for_k(logits, inp, attn, k)
            m = _compute_topk_metrics(logits_k, tgt_k, mask_k, topks=topks)
            N = m["count"]
            if N == 0:
                continue
            agg[k]["loss_sum"] += m["loss"] * N
            agg[k]["count"]    += N
            for t in topks:
                agg[k][f"hits@{t}"] += m[f"hits@{t}"] * N
            flags[k].append(_gather_correct_flags(logits_k, tgt_k, mask_k))

    res = {}
    for k in OFFSETS:
        N = max(1, agg[k]["count"])
        res[k] = {
            "loss": agg[k]["loss_sum"] / N,
            **{f"hits@{t}": agg[k][f"hits@{t}"] / N for t in topks},
            "count": agg[k]["count"]
        }
    flags = {k: (np.concatenate(flags[k]) if len(flags[k]) else np.zeros((0,), dtype=bool)) for k in OFFSETS}
    return res, flags

# ====== training loop: train ONLY the aux heads ======
history = {"time_s": [], "step": [], "train_loss": [], "val_aux_hits1_k1": [], "val_aux_hits1_k2": []}

def train_epoch(model, aux, loader, optim, val_loader=None, log_every=100):
    model.eval(); aux.train()
    t0 = time.perf_counter()
    running = 0.0
    steps_done = 0

    for step_idx, batch in enumerate(loader, 1):
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)

        with torch.no_grad():
            H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,d)

        off = aux(H, attn)  # dict: k -> (logits_k, mask_k)

        # collect per-offset losses
        per_losses = []
        for k in OFFSETS:
            logits_k, mask_k = off[k]          # (B,T-k,V), (B,T-k)
            tgt_k = inp[:, k:]                 # (B,T-k)
            V = logits_k.size(-1)
            m = mask_k.bool().view(-1)
            if m.any():
                logits_f = logits_k.reshape(-1, V)[m]
                tgt_f    = tgt_k.reshape(-1)[m]
                per_losses.append(F.cross_entropy(logits_f, tgt_f))

        # if no valid positions this batch, skip update
        if len(per_losses) == 0:
            # still do periodic logging snapshots
            if step_idx % log_every == 0 and val_loader is not None:
                wall = time.perf_counter() - t0
                aux_metrics_snap, _ = eval_aux(aux, model, val_loader, topks=(1,))
                h1 = aux_metrics_snap[1]["hits@1"]; h2 = aux_metrics_snap[2]["hits@1"]
                history["time_s"].append(wall)
                history["step"].append(steps_done)
                history["train_loss"].append(running / max(1, steps_done))
                history["val_aux_hits1_k1"].append(h1)
                history["val_aux_hits1_k2"].append(h2)
                print(f"step {step_idx:4d} | (skip batch) train_loss={running/max(1,steps_done):.3f} | aux hits@1: +1={h1:.3f} +2={h2:.3f}")
            continue

        total = torch.stack(per_losses).sum()   # Tensor
        optim.zero_grad(set_to_none=True)
        total.backward()
        optim.step()

        running += float(total.item())
        steps_done += 1

        if step_idx % log_every == 0:
            wall = time.perf_counter() - t0
            if val_loader is not None:
                aux_metrics_snap, _ = eval_aux(aux, model, val_loader, topks=(1,))
                h1 = aux_metrics_snap[1]["hits@1"]; h2 = aux_metrics_snap[2]["hits@1"]
            else:
                h1 = h2 = float('nan')
            history["time_s"].append(wall)
            history["step"].append(steps_done)
            history["train_loss"].append(running / steps_done)
            history["val_aux_hits1_k1"].append(h1)
            history["val_aux_hits1_k2"].append(h2)
            print(f"step {step_idx:4d} | train_loss={running/steps_done:.3f} | aux hits@1: +1={h1:.3f} +2={h2:.3f}")

    return running / max(1, steps_done)

# ====== run: baseline -> train -> compare ======
print(">>> Baseline (vanilla GPT-2)")
base_metrics, base_flags = eval_vanilla(model, val_loader, topks=(1,3,5))
for k in OFFSETS:
    bm = base_metrics[k]
    print(f"vanilla k=+{k}: hits@1={bm['hits@1']:.3f} hits@3={bm['hits@3']:.3f} hits@5={bm['hits@5']:.3f} loss={bm['loss']:.3f} N={bm['count']}")

print("\n>>> Train aux heads (frozen trunk)")
EPOCHS = 2
for e in range(1, EPOCHS+1):
    tr_loss = train_epoch(model, aux, train_loader, optim, val_loader=val_loader, log_every=100)
    aux_metrics, aux_flags = eval_aux(aux, model, val_loader, topks=(1,3,5))
    print(f"[epoch {e}] train_loss={tr_loss:.3f} | "
          f"aux +1 hits@1={aux_metrics[1]['hits@1']:.3f} +2 hits@1={aux_metrics[2]['hits@1']:.3f}")

print("\n>>> Final comparison (hits@1 / loss)")
aux_metrics, aux_flags = eval_aux(aux, model, val_loader, topks=(1,3,5))
for k in OFFSETS:
    print(f"k=+{k} | AUX hits@1={aux_metrics[k]['hits@1']:.3f} (loss={aux_metrics[k]['loss']:.3f})"
          f"  vs  VANILLA hits@1={base_metrics[k]['hits@1']:.3f} (loss={base_metrics[k]['loss']:.3f})")

print("\n>>> Significance (McNemar exact, two-sided) for top1 correctness")
for k in OFFSETS:
    v = base_flags[k]
    a = aux_flags[k]
    # align lengths (should match; just in case)
    n = min(len(v), len(a))
    v, a = v[:n], a[:n]
    # counts
    both_correct = np.sum((v == True) & (a == True))
    v_only = np.sum((v == True) & (a == False))
    a_only = np.sum((v == False) & (a == True))
    both_wrong = np.sum((v == False) & (a == False))
    p = mcnemar_exact_p(v_only, a_only)
    print(f"k=+{k}: table=[[{both_correct}, {v_only}], [{a_only}, {both_wrong}]]  p={p:.4g}")
```

Look at the context aND PREDICTED TOKENS IN THE TOP K TO VERIOFY HOW FAR OFF THE MODEL REALLY WAS.

```python
import torch

@torch.no_grad()
def show_stepwise_predictions(model, aux, loader, tokenizer, n_show=8, max_ctx_tokens=40, device=device):
    """
    For a few valid positions t, show:
      - CONTEXT up to t
      - GOLD t+1, GOLD t+2
      - VANILLA greedy autoregressive predictions: pred(t+1), then pred(t+2) after appending pred(t+1)
      - AUX heads top-5 for +1 and +2 from the same source hidden state at t.
    """
    model.eval(); aux.eval()
    shown = 0

    for batch in loader:
        inp  = batch.input_ids.to(device)           # (B,T)
        attn = batch.attention_mask.to(device)      # (B,T)
        B, T = inp.shape
        if T <= 2:
            continue

        # 1) Vanilla logits for alignment (for reference if needed)
        out_v = model(input_ids=inp, attention_mask=attn)
        logits_all = out_v.logits  # (B,T,V)

        # 2) Hidden states for aux heads
        H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,d)
        off = aux(H, attn)  # dict: {1: (logits_+1, mask_+1), 2: (logits_+2, mask_+2)}
        logits_aux1, mask_aux1 = off[1]  # (B,T-1,V), (B,T-1)
        logits_aux2, mask_aux2 = off[2]  # (B,T-2,V), (B,T-2)

        # collect valid source positions t where both +1 and +2 are valid
        # (i.e., t <= T-3 so t+1 and t+2 exist; masks also true)
        valid_mask = torch.zeros((B, T-2), dtype=torch.bool, device=device)
        valid_mask = mask_aux2  # already ensures both positions are valid for +2; +1 is implied for t by construction

        idxs = valid_mask.nonzero(as_tuple=False)  # [(b, t_src)] t_src in [0..T-3]
        if idxs.numel() == 0:
            continue

        # randomize to see variety
        perm = torch.randperm(idxs.size(0), device=device)
        idxs = idxs[perm]

        for b, t_src in idxs:
            b = int(b.item()); t_src = int(t_src.item())

            # --- build human-readable CONTEXT up to t_src ---
            # clip left to last max_ctx_tokens (avoid long contexts)
            left = max(0, t_src - max_ctx_tokens)
            ctx_ids = inp[b, left:t_src]  # tokens strictly before position t_src
            # NOTE: the source hidden state is at position t_src, which corresponds to token at index t_src
            # In GPT-style, to predict next token, we "see" tokens up to and including position t_src
            ctx_with_src = inp[b, left:t_src+1]
            # drop pads for printing
            ctx_print = tokenizer.decode([x for x in ctx_with_src.tolist() if x != tokenizer.pad_token_id])

            # --- GOLD tokens at t+1 and t+2 ---
            gold1_id = int(inp[b, t_src+1].item())
            gold2_id = int(inp[b, t_src+2].item())
            gold1 = tokenizer.decode([gold1_id]) or repr(gold1_id)
            gold2 = tokenizer.decode([gold2_id]) or repr(gold2_id)

            # --- VANILLA greedy autoregressive predictions from exactly this context ---
            #   Step A: predict +1 from context up to t_src
            #   Step B: append pred(+1) and predict +2
            ctx_for_v = ctx_with_src[ctx_with_src != tokenizer.pad_token_id].unsqueeze(0)  # (1, L)
            # predict +1
            out_step1 = model(input_ids=ctx_for_v)
            last_logits1 = out_step1.logits[:, -1, :]  # (1,V)
            pred1_id = int(torch.argmax(last_logits1, dim=-1).item())

            # predict +2 after appending pred1
            ctx_step2 = torch.cat([ctx_for_v, torch.tensor([[pred1_id]], device=device, dtype=ctx_for_v.dtype)], dim=1)
            out_step2 = model(input_ids=ctx_step2)
            last_logits2 = out_step2.logits[:, -1, :]
            pred2_id = int(torch.argmax(last_logits2, dim=-1).item())

            van1 = tokenizer.decode([pred1_id]) or repr(pred1_id)
            van2 = tokenizer.decode([pred2_id]) or repr(pred2_id)

            # --- AUX top-5 for +1 and +2 from same H at t_src ---
            la1 = logits_aux1[b, t_src, :]  # (V,)
            pa1 = torch.softmax(la1, dim=-1)
            top5_p1, top5_i1 = torch.topk(pa1, k=5)
            aux1_top5 = [(tokenizer.decode([i.item()]) or repr(i.item()), float(p.item())) for p,i in zip(top5_p1, top5_i1)]
            aux1_top1 = aux1_top5[0][0]

            la2 = logits_aux2[b, t_src, :]
            pa2 = torch.softmax(la2, dim=-1)
            top5_p2, top5_i2 = torch.topk(pa2, k=5)
            aux2_top5 = [(tokenizer.decode([i.item()]) or repr(i.item()), float(p.item())) for p,i in zip(top5_p2, top5_i2)]
            aux2_top1 = aux2_top5[0][0]

            # flags: whether gold in top5
            gold_in_aux1 = any(int(i.item()) == gold1_id for i in top5_i1)
            gold_in_aux2 = any(int(i.item()) == gold2_id for i in top5_i2)

            print("—"*96)
            print(f"CONTEXT [...t={t_src}] : {ctx_print}")
            print(f"GOLD:  t+1={gold1!r}, t+2={gold2!r}")
            print(f"VANIL: t+1={van1!r}, t+2={van2!r}   (greedy autoregressive)")
            print(f"AUX+1 top5: {[(tok, f'{p:.3f}') for tok,p in aux1_top5]}  gold_in_top5={gold_in_aux1}")
            print(f"AUX+2 top5: {[(tok, f'{p:.3f}') for tok,p in aux2_top5]}  gold_in_top5={gold_in_aux2}")

            shown += 1
            if shown >= n_show:
                return

# Run it:
show_stepwise_predictions(model, aux, val_loader, tokenizer, n_show=8)
```

```python
@torch.no_grad()
def aux_gold_rank_histogram(model, aux, loader, tokenizer, offset=2, max_rank=50, device=device):
    model.eval(); aux.eval()
    ranks = []

    for batch in loader:
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)
        if inp.size(1) <= offset:
            continue

        H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state
        logits_a, mask_a = aux(H, attn)[offset]     # (B,T-k,V),(B,T-k)
        tgt_a = inp[:, offset:]

        B, Tk, V = logits_a.shape
        m = mask_a.bool()
        if not m.any():
            continue

        la = logits_a[m]                           # (N,V)
        gold = tgt_a[m]                            # (N,)
        # rank of gold (1 = best)
        order = torch.argsort(la, dim=-1, descending=True)   # (N,V)
        # find index where gold appears
        eq = (order == gold.unsqueeze(-1))                    # (N,V)
        pos = torch.argmax(eq.int(), dim=-1)                  # (N,)
        ranks.extend((pos + 1).cpu().tolist())

    import matplotlib.pyplot as plt
    import numpy as np
    ranks = np.array(ranks)
    if len(ranks) == 0:
        print("No valid ranks collected.")
        return
    plt.figure()
    bins = np.arange(1, max_rank+2)  # 1..max_rank
    plt.hist(np.clip(ranks, 1, max_rank), bins=bins, edgecolor='k', align='left')
    plt.xlabel("Gold rank under AUX logits (k=+2)")
    plt.ylabel("Count")
    plt.title("Where does the gold token rank under aux (+2)?")
    plt.xticks(range(1, max_rank+1, 5))
    plt.show()

aux_gold_rank_histogram(model, aux, val_loader, tokenizer, offset=2, max_rank=30)
```

## multi head future token predictor with vanilla pretrained head and new future token head

trunk is froze as before. k+1 weights are initialized from gpt2 pretrained head. added a future k+2 head and trained both heads together for future token prediction task. K+2 is initialized randomly as normal.

```python
# ====== minimal installs ======
!pip -q install transformers datasets torch

# ====== imports ======
import time, numpy as np
import math, torch, torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from dataclasses import dataclass
from typing import Dict, Tuple
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoConfig

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ====== data: small & simple (WikiText-2 raw) ======
raw = load_dataset("wikitext", "wikitext-2-raw-v1")
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # GPT-2 has no pad; use EOS

MAX_LEN = 256

def tokenize_fn(ex):
    # Fixed length => no ragged batches; attention_mask marks real tokens.
    return tokenizer(
        ex["text"],
        padding="max_length",
        truncation=True,
        max_length=MAX_LEN,
    )

tok_train = raw["train"].map(tokenize_fn, batched=True, remove_columns=raw["train"].column_names)
tok_val   = raw["validation"].map(tokenize_fn, batched=True, remove_columns=raw["validation"].column_names)

@dataclass
class CausalBatch:
    input_ids: torch.Tensor
    attention_mask: torch.Tensor

def collate(batch):
    return CausalBatch(
        input_ids=torch.tensor([b["input_ids"] for b in batch], dtype=torch.long),
        attention_mask=torch.tensor([b["attention_mask"] for b in batch], dtype=torch.long),
    )

# small slices for speed; increase for better signal
train_loader = DataLoader(tok_train.select(range(4000)), batch_size=8, shuffle=True,  collate_fn=collate)
val_loader   = DataLoader(tok_val.select(range(1000)),  batch_size=8, shuffle=False, collate_fn=collate)

# ====== model: load GPT-2, freeze trunk, add auxiliary heads (t+1, t+2) ======
base_name = "gpt2"
cfg = AutoConfig.from_pretrained(base_name, output_hidden_states=False)  # last_hidden_state is sufficient
model = AutoModelForCausalLM.from_pretrained(base_name, config=cfg).to(device)
vocab_size = model.config.vocab_size
hidden_size = model.transformer.wte.embedding_dim  # 768 for gpt2

# freeze all pretrained params
for p in model.parameters():
    p.requires_grad_(False)

OFFSETS = (1, 2)  # drop +3 everywhere

# --- CHANGE 1: make aux heads bias-free so we can copy lm_head exactly ---
class OffsetHeads(nn.Module):
    """
    Simple linear heads that read from the final hidden state at position t
    and predict tokens at offsets t+1, t+2.
    """
    def __init__(self, hidden_size: int, vocab_size: int, offsets=OFFSETS):
        super().__init__()
        self.offsets = list(offsets)
        # bias=False to match GPT-2's lm_head
        self.heads = nn.ModuleDict({str(k): nn.Linear(hidden_size, vocab_size, bias = False) for k in self.offsets})

    def forward(self, H: torch.Tensor, attn_mask: torch.Tensor) -> Dict[int, Tuple[torch.Tensor, torch.Tensor]]:
        """
        H: (B, T, d) final hidden states from GPT-2 transformer
        attn_mask: (B, T)
        Returns dict: k -> (logits_k, mask_k),
          where logits_k shape is (B, T-k, V) aligned to predict x_{t+k} from H_t
        """
        B, T, D = H.shape
        out = {}
        for k in self.offsets:
            head = self.heads[str(k)]
            # valid source positions t: 0..T-k-1
            logits = head(H[:, :-k, :])                     # (B, T-k, V)
            # mask requires both source pos and target pos to be valid:
            mask = (attn_mask[:, :-k] & attn_mask[:, k:])   # (B, T-k)
            out[k] = (logits, mask)
        return out

aux = OffsetHeads(hidden_size, vocab_size, offsets=OFFSETS).to(device)

# --- CHANGE 2: clone the pretrained GPT-2 head weights into our k=+1 aux head ---
# This keeps the GPT-2 trunk and original tied head frozen; we only train our independent aux heads.
with torch.no_grad():
    if "1" in aux.heads:
        aux.heads["1"].weight.copy_(model.lm_head.weight)  # copy weights; now k+1 aux starts from GPT-2's head
    else:
        raise ValueError("OFFSETS must include 1 to initialize k+1 head from pretrained weights.")

# optimizer stays the same: it will update ONLY aux heads (k=+1 cloned, k=+2 random) while trunk remains frozen
optim = torch.optim.AdamW(aux.parameters(), lr=5e-4)

# ====== helpers: alignment, metrics, significance ======
def _aligned_slices_for_k(logits, inp, attn, k):
    """
    Align logits (vanilla) at positions t+k-1 with targets at t+k.
    Returns: logits_k (B, T-k, V), tgt_k (B, T-k), mask_k (B, T-k)
    """
    B, T, V = logits.shape
    assert T > k, "sequence too short for this offset"
    start = k - 1
    logits_k = logits[:, start:-1, :]          # (B, T-k, V)
    tgt_k    = inp[:, k:]                      # (B, T-k)
    mask_k   = attn[:, start:-1] & attn[:, k:] # (B, T-k)
    return logits_k, tgt_k, mask_k

def _compute_topk_metrics(logits, targets, mask, topks=(1,3,5)):
    """
    logits: (B, T, V), targets: (B, T), mask: (B, T)
    Returns dict: {"loss":..., "hits@1":..., "hits@3":..., "hits@5":..., "count":...}
    """
    V = logits.size(-1)
    m = mask.bool().view(-1)
    if m.sum().item() == 0:
        return {"loss": 0.0, "count": 0, **{f"hits@{k}": 0.0 for k in topks}}
    logits_f = logits.reshape(-1, V)[m]
    tgt_f    = targets.reshape(-1)[m]
    loss     = F.cross_entropy(logits_f, tgt_f).item()
    out = {"loss": loss, "count": int(m.sum().item())}
    for k in topks:
        k = min(k, V)
        topk_idx = logits_f.topk(k, dim=-1).indices
        hits = (topk_idx == tgt_f.unsqueeze(-1)).any(dim=-1).float().mean().item()
        out[f"hits@{k}"] = hits
    return out

def _gather_correct_flags(logits, targets, mask):
    """
    Returns boolean vector (N,) of 'was top1 correct' for significance tests.
    """
    V = logits.size(-1)
    m = mask.bool().view(-1)
    if m.sum().item() == 0:
        return np.zeros((0,), dtype=bool)
    logits_f = logits.reshape(-1, V)[m]
    tgt_f    = targets.reshape(-1)[m]
    pred     = logits_f.argmax(-1)
    return (pred == tgt_f).detach().cpu().numpy().astype(bool)

def mcnemar_exact_p(b, c):
    """
    Exact two-sided McNemar via binomial test on discordant pairs (n=b+c).
    Uses log-space to avoid overflow for large n.
    b = count where method A correct and B wrong
    c = count where method B correct and A wrong
    """
    n = b + c
    if n == 0:
        return 1.0

    k_max = min(b, c)  # smaller side for the two-sided tail
    log_half = math.log(0.5)

    # log C(n, k) via lgamma
    def log_comb(n_, k_):
        return math.lgamma(n_ + 1) - math.lgamma(k_ + 1) - math.lgamma(n_ - k_ + 1)

    # accumulate in log-space: tail = sum_{k=0..k_max} C(n,k) * 0.5^n
    logs = [log_comb(n, k) + n * log_half for k in range(k_max + 1)]
    m = max(logs)
    tail = math.exp(m) * sum(math.exp(li - m) for li in logs)

    p_two_sided = min(1.0, 2.0 * tail)
    return p_two_sided


# ====== eval: AUX (frozen heads) ======
@torch.no_grad()
def eval_aux(aux, model, loader, topks=(1,3,5)):
    model.eval(); aux.eval()
    agg = {k: {"loss_sum":0.0, "count":0, **{f"hits@{t}":0.0 for t in topks}} for k in OFFSETS}
    flags = {k: [] for k in OFFSETS}

    for batch in loader:
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)
        if inp.size(1) <= max(OFFSETS):
            continue

        H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,d)
        out = aux(H, attn)  # dict k -> (logits, mask)

        for k in OFFSETS:
            logits_k, mask_k = out[k]
            tgt_k = inp[:, k:]  # (B, T-k)
            m = _compute_topk_metrics(logits_k, tgt_k, mask_k, topks=topks)
            N = m["count"]
            if N == 0:
                continue
            agg[k]["loss_sum"] += m["loss"] * N
            agg[k]["count"]    += N
            for t in topks:
                agg[k][f"hits@{t}"] += m[f"hits@{t}"] * N
            flags[k].append(_gather_correct_flags(logits_k, tgt_k, mask_k))

    res = {}
    for k in OFFSETS:
        N = max(1, agg[k]["count"])
        res[k] = {
            "loss": agg[k]["loss_sum"] / N,
            **{f"hits@{t}": agg[k][f"hits@{t}"] / N for t in topks},
            "count": agg[k]["count"]
        }
    flags = {k: (np.concatenate(flags[k]) if len(flags[k]) else np.zeros((0,), dtype=bool)) for k in OFFSETS}
    return res, flags

# ====== eval: VANILLA (teacher forced @ offsets) ======
@torch.no_grad()
def eval_vanilla(model, loader, topks=(1,3,5)):
    model.eval()
    agg = {k: {"loss_sum":0.0, "count":0, **{f"hits@{t}":0.0 for t in topks}} for k in OFFSETS}
    flags = {k: [] for k in OFFSETS}

    for batch in loader:
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)
        if inp.size(1) <= max(OFFSETS):
            continue
        out  = model(input_ids=inp, attention_mask=attn)
        logits = out.logits  # (B,T,V)

        for k in OFFSETS:
            logits_k, tgt_k, mask_k = _aligned_slices_for_k(logits, inp, attn, k)
            m = _compute_topk_metrics(logits_k, tgt_k, mask_k, topks=topks)
            N = m["count"]
            if N == 0:
                continue
            agg[k]["loss_sum"] += m["loss"] * N
            agg[k]["count"]    += N
            for t in topks:
                agg[k][f"hits@{t}"] += m[f"hits@{t}"] * N
            flags[k].append(_gather_correct_flags(logits_k, tgt_k, mask_k))

    res = {}
    for k in OFFSETS:
        N = max(1, agg[k]["count"])
        res[k] = {
            "loss": agg[k]["loss_sum"] / N,
            **{f"hits@{t}": agg[k][f"hits@{t}"] / N for t in topks},
            "count": agg[k]["count"]
        }
    flags = {k: (np.concatenate(flags[k]) if len(flags[k]) else np.zeros((0,), dtype=bool)) for k in OFFSETS}
    return res, flags

# ====== training loop: train ONLY the aux heads ======
history = {"time_s": [], "step": [], "train_loss": [], "val_aux_hits1_k1": [], "val_aux_hits1_k2": []}

def train_epoch(model, aux, loader, optim, val_loader=None, log_every=100):
    model.eval(); aux.train()
    t0 = time.perf_counter()
    running = 0.0
    steps_done = 0

    for step_idx, batch in enumerate(loader, 1):
        inp  = batch.input_ids.to(device)
        attn = batch.attention_mask.to(device)

        with torch.no_grad():
            H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,d)

        off = aux(H, attn)  # dict: k -> (logits_k, mask_k)

        # collect per-offset losses
        per_losses = []
        for k in OFFSETS:
            logits_k, mask_k = off[k]          # (B,T-k,V), (B,T-k)
            tgt_k = inp[:, k:]                 # (B,T-k)
            V = logits_k.size(-1)
            m = mask_k.bool().view(-1)
            if m.any():
                logits_f = logits_k.reshape(-1, V)[m]
                tgt_f    = tgt_k.reshape(-1)[m]
                per_losses.append(F.cross_entropy(logits_f, tgt_f))

        # if no valid positions this batch, skip update
        if len(per_losses) == 0:
            # still do periodic logging snapshots
            if step_idx % log_every == 0 and val_loader is not None:
                wall = time.perf_counter() - t0
                aux_metrics_snap, _ = eval_aux(aux, model, val_loader, topks=(1,))
                h1 = aux_metrics_snap[1]["hits@1"]; h2 = aux_metrics_snap[2]["hits@1"]
                history["time_s"].append(wall)
                history["step"].append(steps_done)
                history["train_loss"].append(running / max(1, steps_done))
                history["val_aux_hits1_k1"].append(h1)
                history["val_aux_hits1_k2"].append(h2)
                print(f"step {step_idx:4d} | (skip batch) train_loss={running/max(1,steps_done):.3f} | aux hits@1: +1={h1:.3f} +2={h2:.3f}")
            continue

        total = torch.stack(per_losses).sum()   # Tensor
        optim.zero_grad(set_to_none=True)
        total.backward()
        optim.step()

        running += float(total.item())
        steps_done += 1

        if step_idx % log_every == 0:
            wall = time.perf_counter() - t0
            if val_loader is not None:
                aux_metrics_snap, _ = eval_aux(aux, model, val_loader, topks=(1,))
                h1 = aux_metrics_snap[1]["hits@1"]; h2 = aux_metrics_snap[2]["hits@1"]
            else:
                h1 = h2 = float('nan')
            history["time_s"].append(wall)
            history["step"].append(steps_done)
            history["train_loss"].append(running / steps_done)
            history["val_aux_hits1_k1"].append(h1)
            history["val_aux_hits1_k2"].append(h2)
            print(f"step {step_idx:4d} | train_loss={running/steps_done:.3f} | aux hits@1: +1={h1:.3f} +2={h2:.3f}")

    return running / max(1, steps_done)

# ====== run: baseline -> train -> compare ======
print(">>> Baseline (vanilla GPT-2)")
base_metrics, base_flags = eval_vanilla(model, val_loader, topks=(1,3,5))
for k in OFFSETS:
    bm = base_metrics[k]
    print(f"vanilla k=+{k}: hits@1={bm['hits@1']:.3f} hits@3={bm['hits@3']:.3f} hits@5={bm['hits@5']:.3f} loss={bm['loss']:.3f} N={bm['count']}")

print("\n>>> Train aux heads (frozen trunk)")
EPOCHS = 2
for e in range(1, EPOCHS+1):
    tr_loss = train_epoch(model, aux, train_loader, optim, val_loader=val_loader, log_every=100)
    aux_metrics, aux_flags = eval_aux(aux, model, val_loader, topks=(1,3,5))
    print(f"[epoch {e}] train_loss={tr_loss:.3f} | "
          f"aux +1 hits@1={aux_metrics[1]['hits@1']:.3f} +2 hits@1={aux_metrics[2]['hits@1']:.3f}")

print("\n>>> Final comparison (hits@1 / loss)")
aux_metrics, aux_flags = eval_aux(aux, model, val_loader, topks=(1,3,5))
for k in OFFSETS:
    print(f"k=+{k} | AUX hits@1={aux_metrics[k]['hits@1']:.3f} (loss={aux_metrics[k]['loss']:.3f})"
          f"  vs  VANILLA hits@1={base_metrics[k]['hits@1']:.3f} (loss={base_metrics[k]['loss']:.3f})")

print("\n>>> Significance (McNemar exact, two-sided) for top1 correctness")
for k in OFFSETS:
    v = base_flags[k]
    a = aux_flags[k]
    # align lengths (should match; just in case)
    n = min(len(v), len(a))
    v, a = v[:n], a[:n]
    # counts
    both_correct = np.sum((v == True) & (a == True))
    v_only = np.sum((v == True) & (a == False))
    a_only = np.sum((v == False) & (a == True))
    both_wrong = np.sum((v == False) & (a == False))
    p = mcnemar_exact_p(v_only, a_only)
    print(f"k=+{k}: table=[[{both_correct}, {v_only}], [{a_only}, {both_wrong}]]  p={p:.4g}")
```

## Testing with base GPT2 and an additional "Probe" head predicting on token t+2

```python
# =======================
# Minimal installs
# =======================
!pip -q install transformers datasets torch

# =======================
# Imports
# =======================
import math, time, numpy as np
import torch, torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from dataclasses import dataclass
from typing import Dict, Tuple
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoConfig

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# =======================
# Data: WikiText-2 (raw)
# =======================
raw = load_dataset("wikitext", "wikitext-2-raw-v1")
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # GPT-2 has no pad; use EOS as pad

MAX_LEN = 256

def tokenize_fn(ex):
    # Fixed length => attention_mask tells which positions are real tokens.
    return tokenizer(
        ex["text"],
        padding="max_length",
        truncation=True,
        max_length=MAX_LEN,
    )

tok_train = raw["train"].map(tokenize_fn, batched=True, remove_columns=raw["train"].column_names)
tok_val   = raw["validation"].map(tokenize_fn, batched=True, remove_columns=raw["validation"].column_names)

@dataclass
class CausalBatch:
    input_ids: torch.Tensor
    attention_mask: torch.Tensor

def collate(batch):
    return CausalBatch(
        input_ids=torch.tensor([b["input_ids"] for b in batch], dtype=torch.long),
        attention_mask=torch.tensor([b["attention_mask"] for b in batch], dtype=torch.long),
    )

# Small subset & modest batch for speed; scale up later for stronger signal
train_loader = DataLoader(tok_train.select(range(4000)), batch_size=8, shuffle=True,  collate_fn=collate)
val_loader   = DataLoader(tok_val.select(range(1000)),  batch_size=8, shuffle=False, collate_fn=collate)

# =======================
# Base model: GPT-2 (frozen)
# =======================
base_name = "gpt2"
cfg = AutoConfig.from_pretrained(base_name, output_hidden_states=False)
model = AutoModelForCausalLM.from_pretrained(base_name, config=cfg).to(device)
vocab_size = model.config.vocab_size
hidden_size = model.transformer.wte.embedding_dim  # 768 for gpt2

# Freeze EVERYTHING in GPT-2
for p in model.parameters():
    p.requires_grad_(False)
model.eval()  # stay in eval: we never train this trunk

# ======================================================
# 1) Probe head: identical to GPT-2 lm_head (Linear)
#    Optionally init from lm_head for a true "identical" start
# ======================================================
class IdenticalLMHeadProbe(nn.Module):
    def __init__(self, hidden_size: int, vocab_size: int, init_from: nn.Linear | None = None):
        super().__init__()
        self.proj = nn.Linear(hidden_size, vocab_size, bias=True)
        if init_from is not None:
            with torch.no_grad():
                self.proj.weight.copy_(init_from.weight)
                if init_from.bias is not None and self.proj.bias is not None:
                    self.proj.bias.copy_(init_from.bias)

    def forward(self, H: torch.Tensor, attn_mask: torch.Tensor):
        """
        Train/eval on ALL valid positions in a batch:
          input H: (B, T, D) final hidden states from frozen GPT-2
          returns logits_k2: (B, T-2, V) aligned so that time idx t maps to x_{t+2}
                  mask_k2:   (B, T-2) valid positions (both t and t+2 are real tokens)
        """
        B, T, D = H.shape
        if T <= 2:
            return H.new_zeros(B, 0, self.proj.out_features), attn_mask[:, :0]
        src_H   = H[:, :-2, :]                           # h_t for t in [0..T-3]
        logits  = self.proj(src_H)                       # (B, T-2, V)
        mask_k2 = (attn_mask[:, :-2] & attn_mask[:,  2:]).to(torch.bool)
        return logits, mask_k2

probe_t2 = IdenticalLMHeadProbe(hidden_size, vocab_size, init_from=model.lm_head).to(device)
optim = torch.optim.AdamW(probe_t2.parameters(), lr=5e-4)

# ======================================================
# 2) TRAINING: probe learns GOLD x_{t+2} from h_t (frozen trunk)
#    (vectorized over all valid t in each batch)
# ======================================================
def train_probe_t2_epoch_gold(model, probe, loader, optim, log_every=100):
    model.eval(); probe.train()
    running, steps = 0.0, 0
    t0 = time.perf_counter()
    for step, batch in enumerate(loader, 1):
        inp  = batch.input_ids.to(device)           # (B,T)
        attn = batch.attention_mask.to(device)      # (B,T)
        with torch.no_grad():
            H = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,D)

        logits_k2, mask_k2 = probe(H, attn)         # (B,T-2,V), (B,T-2)
        if logits_k2.size(1) == 0:
            continue
        tgt_k2 = inp[:, 2:]                         # gold x_{t+2}, (B,T-2)

        m = mask_k2.view(-1)
        if not m.any():
            continue
        V = logits_k2.size(-1)
        loss = F.cross_entropy(logits_k2.reshape(-1, V)[m], tgt_k2.reshape(-1)[m])

        optim.zero_grad(set_to_none=True)
        loss.backward()
        optim.step()

        running += float(loss.item()); steps += 1
        if step % log_every == 0:
            print(f"step {step:4d} | probe(loss)={running/steps:.3f} | {time.perf_counter()-t0:.1f}s")
    return running / max(1, steps)

# ======================================================
# 3) EVAL (QUANT): compare Base(AR t+2) vs GOLD and Probe(t+2) vs GOLD
#    Also measure Base↔Probe KL to quantify how similar their distributions are.
#    For each sequence we sample one (or few) valid t to keep this light.
# ======================================================
@torch.no_grad()
def eval_ar_vs_probe_metrics(model, probe, loader, tokenizer, samples_per_seq=1, ks=(1,3,5)):
    model.eval(); probe.eval()
    agg = {
        "probe": {"loss_sum":0.0, "count":0, **{f"hits@{k}":0.0 for k in ks}},
        "base":  {"loss_sum":0.0, "count":0, **{f"hits@{k}":0.0 for k in ks}},
        "kl_bp_sum": 0.0,  # KL(base || probe)
        "kl_pb_sum": 0.0,  # KL(probe || base)
        "kl_count": 0
    }
    for batch in loader:
        inp  = batch.input_ids.to(device)     # (B,T)
        attn = batch.attention_mask.to(device)
        B, T = inp.shape
        # Precompute full hidden states for probe training-style path (for h_t)
        H_full = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,D)

        for b in range(B):
            # valid t where (t, t+1, t+2) exist (not pads)
            valid = ((attn[b, :-2] > 0) & (attn[b, 1:-1] > 0) & (attn[b, 2:] > 0)).nonzero(as_tuple=False).flatten()
            if len(valid) == 0:
                continue
            # sample a few t per sequence
            idxs = valid.cpu().tolist()
            np.random.shuffle(idxs)
            for i, t in enumerate(idxs[:samples_per_seq]):
                # ===== Base AR: step1 (t+1), append, step2 (t+2)
                ctx_ids = inp[b, :t+1]
                ctx_att = attn[b, :t+1]
                out1 = model(input_ids=ctx_ids.unsqueeze(0), attention_mask=ctx_att.unsqueeze(0))
                logits_t1 = out1.logits[:, -1, :]                         # (1,V)
                next_id   = int(logits_t1.argmax(-1).item())              # greedy; swap to sampling if you want
                ctx2_ids = torch.cat([ctx_ids, torch.tensor([next_id], device=ctx_ids.device, dtype=ctx_ids.dtype)], dim=0)
                ctx2_att = torch.cat([ctx_att, torch.tensor([1], device=ctx_att.device, dtype=ctx_att.dtype)], dim=0)
                out2 = model(input_ids=ctx2_ids.unsqueeze(0), attention_mask=ctx2_att.unsqueeze(0))
                logits_t2_base = out2.logits[:, -1, :].squeeze(0)         # (V,)

                # ===== Probe: from same h_t predict t+2
                # We already have H_full on full sequence, but h_t with causal mask equals h_t from ctx-only
                h_t = H_full[b, t, :].unsqueeze(0).unsqueeze(0)           # (1,1,D)
                # Reuse probe projection cleanly: it expects (B,T,D) + mask, so feed a tiny length-3 shim to produce 1 slot
                # Simpler: call the linear directly on h_t
                logits_t2_probe = probe.proj(h_t).squeeze(0).squeeze(0)   # (V,)

                # ===== GOLD @ t+2
                gold_t2 = int(inp[b, t+2].item())

                # ===== Metrics vs GOLD
                for label, logits in (("base", logits_t2_base), ("probe", logits_t2_probe)):
                    V = logits.size(0)
                    loss = F.cross_entropy(logits.view(1, V), torch.tensor([gold_t2], device=logits.device)).item()
                    agg[label]["loss_sum"] += loss
                    agg[label]["count"]    += 1
                    probs = logits.softmax(-1)
                    for k in ks:
                        topk = probs.topk(min(k, V), dim=-1).indices
                        hit = float((topk == gold_t2).any().item())
                        agg[label][f"hits@{k}"] += hit

                # ===== Distribution similarity (optional but useful)
                p_base  = logits_t2_base.log_softmax(-1)
                p_probe = logits_t2_probe.log_softmax(-1)
                # KL(base || probe) and KL(probe || base)
                kl_bp = F.kl_div(p_probe, p_base.exp(), reduction="batchmean", log_target=False).item()
                kl_pb = F.kl_div(p_base,  p_probe.exp(), reduction="batchmean", log_target=False).item()
                agg["kl_bp_sum"] += kl_bp
                agg["kl_pb_sum"] += kl_pb
                agg["kl_count"]  += 1

    def finalize(side):
        N = max(1, agg[side]["count"])
        return {
            "loss": agg[side]["loss_sum"]/N,
            "count": agg[side]["count"],
            **{f"hits@{k}": agg[side][f"hits@{k}"]/N for k in ks}
        }

    results = {
        "probe_vs_gold": finalize("probe"),
        "baseAR_vs_gold": finalize("base"),
        "kl_base_probe": {
            "KL(base||probe)": agg["kl_bp_sum"] / max(1, agg["kl_count"]),
            "KL(probe||base)": agg["kl_pb_sum"] / max(1, agg["kl_count"]),
            "pairs": agg["kl_count"],
        }
    }
    return results

# =======================
# Pretty helpers (simple & readable)
# =======================
def _pretty(id_, tokenizer):
    s = tokenizer.decode([int(id_)])  # decode ONE id at a time
    # make leading space and newline visible in the printout
    if s.startswith(" "): s = "␠" + s[1:]
    s = s.replace("\n", "⏎")
    return s if s != "" else "<∅>"

def topk_pretty(logits_1d, tokenizer, k=5):
    ids = logits_1d.topk(k, dim=-1).indices.tolist()
    return "[" + ", ".join(_pretty(i, tokenizer) for i in ids) + "]"

# =======================
# QUAL: AR base vs probe (readable, randomized t)
# =======================
@torch.no_grad()
def show_qual_ar_vs_probe(model, probe, loader, tokenizer, n_examples=5, min_t=10, ctx_chars=140):
    """
    Base: AR predict t+1 (top-5), append top-1, AR predict t+2 (top-5).
    Probe: predict t+2 directly from the same h_t.
    Prints readable context, gold tokens, and top-5 lists.
    """
    model.eval(); probe.eval()
    shown = 0

    for batch in loader:
        if shown >= n_examples: break
        inp  = batch.input_ids.to(device)      # (B,T)
        attn = batch.attention_mask.to(device) # (B,T)
        H    = model.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,D)
        B, T = inp.shape

        for b in range(B):
            if shown >= n_examples: break
            # valid t with room for t+1 and t+2, and not too early
            valid = ((attn[b, :-2] > 0) & (attn[b, 1:-1] > 0) & (attn[b, 2:] > 0)).nonzero(as_tuple=False).flatten()
            valid = valid[valid >= min_t]
            if len(valid) == 0:
                continue

            # pick a random source index t for this sequence
            t = int(valid[torch.randint(0, len(valid), ()).item()].item())

            # ----- Base AR step 1: predict t+1 from context -----
            ctx_ids = inp[b, :t+1]                           # x_0..x_t
            out1 = model(input_ids=ctx_ids.unsqueeze(0))
            logits_t1 = out1.logits[:, -1, :].squeeze(0)     # (V,)
            next_id   = int(logits_t1.argmax(-1).item())

            # ----- Base AR step 2: append predicted t+1; predict t+2 -----
            ctx2_ids = torch.cat([ctx_ids, torch.tensor([next_id], device=ctx_ids.device, dtype=ctx_ids.dtype)], dim=0)
            out2 = model(input_ids=ctx2_ids.unsqueeze(0))
            logits_t2_base = out2.logits[:, -1, :].squeeze(0)  # (V,)

            # ----- Probe: predict t+2 from same h_t -----
            h_t = H[b, t, :].unsqueeze(0).unsqueeze(0)         # (1,1,D)
            logits_t2_probe = probe.proj(h_t).squeeze(0).squeeze(0)

            # ----- Gold tokens (decoded, readable)
            gold_t1 = _pretty(inp[b, t+1].item(), tokenizer)
            gold_t2 = _pretty(inp[b, t+2].item(), tokenizer)

            # ----- Top-5 (decoded, readable)
            base_top5_t1  = topk_pretty(logits_t1, tokenizer, k=5)
            base_appended = _pretty(next_id, tokenizer)
            base_top5_t2  = topk_pretty(logits_t2_base, tokenizer, k=5)
            probe_top5_t2 = topk_pretty(logits_t2_probe, tokenizer, k=5)

            # ----- Readable context: last N characters before t
            ctx_text = tokenizer.decode(inp[b, :t+1].tolist()).replace("\n", "⏎")
            ctx_snip = ctx_text[-ctx_chars:]

            print("—"*96)
            print(f"CONTEXT (last {ctx_chars} chars before t={t}): …{ctx_snip}")
            print(f"GOLD  t+1: {gold_t1}")
            print(f"BASE  t+1 top5: {base_top5_t1}")
            print(f"(appended {base_appended!r} → AR step for t+2)")
            print(f"GOLD  t+2: {gold_t2}")
            print(f"BASE  t+2 top5: {base_top5_t2}")
            print(f"PROBE t+2 top5 (from same h_t): {probe_top5_t2}")

            shown += 1
```

## Full multi token predictor with back propogating gradients

```python
import torch, torch.nn as nn
import torch.nn.functional as F

class GPT2TwoHead(nn.Module):
    """
    Head1: standard LM head for t+1 (reuse GPT-2's lm_head)
    Head2: auxiliary head for t+2, trained jointly
    """
    def __init__(self, base):
        super().__init__()
        self.base = base  # AutoModelForCausalLM
        d = base.transformer.wte.embedding_dim
        V = base.config.vocab_size
        self.head_t2 = nn.Linear(d, V, bias=True)
        # (optional) init t+2 head from lm_head for faster start
        with torch.no_grad():
            self.head_t2.weight.copy_(base.lm_head.weight)
            if self.head_t2.bias is not None and base.lm_head.bias is not None:
                self.head_t2.bias.copy_(base.lm_head.bias)

    def forward(self, input_ids, attention_mask=None):
        # Hidden states from trunk
        H = self.base.transformer(input_ids=input_ids, attention_mask=attention_mask).last_hidden_state  # (B,T,D)
        # t+1 logits from original lm_head
        logits_t1 = self.base.lm_head(H)  # (B,T,V)
        # t+2 logits from aux head (aligned so position t maps to target x_{t+2})
        if H.size(1) > 2:
            logits_t2 = self.head_t2(H[:, :-2, :])         # (B, T-2, V)
        else:
            logits_t2 = H.new_zeros(H.size(0), 0, self.base.config.vocab_size)
        return logits_t1, logits_t2
```

```python
def ce_t1(logits_t1, input_ids, attention_mask):
    B, T, V = logits_t1.shape
    if T <= 1:
        # graph-connected zero
        return logits_t1.sum() * 0.0
    logits = logits_t1[:, :-1, :]
    targets = input_ids[:, 1:]
    mask = attention_mask[:, 1:].bool()
    m = mask.view(-1)
    if m.sum() == 0:
        return logits.sum() * 0.0
    return F.cross_entropy(logits.reshape(-1, V)[m], targets.reshape(-1)[m])

def ce_t2(logits_t2, input_ids, attention_mask):
    if logits_t2.size(1) == 0:
        return logits_t2.sum() * 0.0
    targets = input_ids[:, 2:]
    mask = (attention_mask[:, :-2].bool() & attention_mask[:, 2:].bool())
    m = mask.view(-1)
    if m.sum() == 0:
        return logits_t2.sum() * 0.0
    V = logits_t2.size(-1)
    return F.cross_entropy(logits_t2.reshape(-1, V)[m], targets.reshape(-1)[m])

# build model
model_two = GPT2TwoHead(model).to(device)  # `model` is your existing pretrained GPT-2 loaded earlier

# UNFREEZE the base *before* making/re-making the optimizer
for p in model_two.base.parameters():
    p.requires_grad_(True)

# optimizer (tune lr based on how much of trunk you unfreeze)
lr = 5e-5 if any(p.requires_grad for p in model_two.base.parameters()) else 5e-4
optim2 = torch.optim.AdamW(filter(lambda p: p.requires_grad, model_two.parameters()), lr=lr, weight_decay=0.01)

# joint training
def train_two_head_epochs(model_two, train_loader, epochs=2, lam1=1.0, lam2=0.3, grad_clip=1.0, log_every=100):
    model_two.train()
    for e in range(1, epochs+1):
        running, steps = 0.0, 0
        for step, batch in enumerate(train_loader, 1):
            inp  = batch.input_ids.to(device)
            attn = batch.attention_mask.to(device)
            logits_t1, logits_t2 = model_two(inp, attention_mask=attn)
            loss1 = ce_t1(logits_t1, inp, attn)
            loss2 = ce_t2(logits_t2, inp, attn)
            loss = lam1*loss1 + lam2*loss2
            if torch.isnan(loss) or torch.isinf(loss):
                print(f"[WARN] skipping step due to NaN/Inf loss (t+1={loss1.item():.3f}, t+2={loss2.item():.3f})")
                optim2.zero_grad(set_to_none=True)
                continue
            optim2.zero_grad(set_to_none=True)
            loss.backward()
            if grad_clip is not None:
                torch.nn.utils.clip_grad_norm_(model_two.parameters(), grad_clip)
            optim2.step()
            running += float(loss.item()); steps += 1
            if step % log_every == 0:
                print(f"[ep {e} | step {step}] loss={running/steps:.3f} (t+1 {loss1.item():.3f} | t+2 {loss2.item():.3f})")
        print(f"[epoch {e}] avg loss={running/max(1,steps):.3f}")
```

```python
@torch.no_grad()
def eval_two_head(model_two, loader, ks=(1,3,5), debug=False):
    """
    Evaluate two-head GPT-2:
      - t+1: compute teacher-forced perplexity
      - t+2: compute hits@k using auxiliary head
    Works with both dict and CausalBatch dataloaders.
    """
    model_two.eval()
    print("eval mode:", model_two.training, model_two.base.training)
    ce_sum, tok_count = 0.0, 0                # for t+1 ppl
    hits = {k: 0.0 for k in ks}; t2_count = 0 # for t+2 hits@k

    for batch in loader:
        # Support dict or dataclass
        if isinstance(batch, dict):
            inp  = batch["input_ids"].to(device)
            attn = batch["attention_mask"].to(device)
        else:
            inp  = batch.input_ids.to(device)
            attn = batch.attention_mask.to(device)

        logits_t1, logits_t2 = model_two(inp, attention_mask=attn)

        # ----- t+1 teacher-forced perplexity -----
        if logits_t1.size(1) > 1:
            logits = logits_t1[:, :-1, :]        # (B,T-1,V)
            targets = inp[:, 1:]
            mask = attn[:, 1:].bool()
            m = mask.view(-1)
            if m.any():
                V = logits.size(-1)
                ce_sum += F.cross_entropy(
                    logits.reshape(-1, V)[m],
                    targets.reshape(-1)[m],
                    reduction="sum"
                ).item()
                tok_count += int(m.sum().item())

        # ----- t+2 hits@k (aux head) -----
        if logits_t2.size(1) > 0:
            targets2 = inp[:, 2:]
            mask2 = (attn[:, :-2].bool() & attn[:, 2:].bool())
            m2 = mask2.view(-1)
            if m2.any():
                V2 = logits_t2.size(-1)
                logits_f = logits_t2.reshape(-1, V2)[m2]
                targets_f = targets2.reshape(-1)[m2]
                t2_count += targets_f.size(0)
                for k in ks:
                    topk_idx = logits_f.topk(min(k, V2), dim=-1).indices
                    hits[k] += (topk_idx == targets_f.unsqueeze(-1)).any(-1).float().sum().item()

    ppl_t1 = float(math.exp(ce_sum / tok_count)) if tok_count > 0 else float("nan")
    out = {"t+1_ppl": ppl_t1, "t+2_count": t2_count, **{f"hits@{k}": hits[k]/max(1,t2_count) for k in ks}}
    if debug: out.update({"t+1_tok_count": tok_count})
    return out
```

```python
# train a bit; start small to sanity-check
print(">>> Joint training (two-headed)")
train_two_head_epochs(model_two, train_loader, epochs=2, lam1=1.0, lam2=1.0, grad_clip=1.0, log_every=100)

print("\n>>> Eval on validation")
metrics = eval_two_head(model_two, val_loader, ks=(1,3,5))
metrics
```

```python
@torch.no_grad()
def sanity_one_batch_ce(model_two, loader):
    model_two.eval()
    batch = next(iter(loader))
    inp  = (batch["input_ids"] if isinstance(batch, dict) else batch.input_ids).to(device)
    attn = (batch["attention_mask"] if isinstance(batch, dict) else batch.attention_mask).to(device)

    logits_t1, _ = model_two(inp, attention_mask=attn)

    # Eval-style CE (sum / count)
    logits = logits_t1[:, :-1, :]
    targets = inp[:, 1:]
    mask = attn[:, 1:].bool()
    V = logits.size(-1)
    m = mask.view(-1)
    ce_sum_eval = F.cross_entropy(logits.reshape(-1, V)[m], targets.reshape(-1)[m], reduction="sum").item()
    tok_count   = int(m.sum().item())
    ce_mean_eval = ce_sum_eval / max(1, tok_count)

    # Train helper CE (mean over valid tokens)
    ce_mean_train = ce_t1(logits_t1, inp, attn).item()

    print(f"CE mean (eval-style):  {ce_mean_eval:.6f}")
    print(f"CE mean (train helper): {ce_mean_train:.6f}")
    print(f"Tokens counted: {tok_count}")
```

```python
# =======================
# QUAL: AR base vs HEAD2 (readable, randomized t)
# =======================
@torch.no_grad()
def show_qual_two_head(model_two, loader, tokenizer, n_examples=5, min_t=10, ctx_chars=140):
    """
    Base: AR predict t+1 (top-5), append top-1, AR predict t+2 (top-5).
    HEAD2: predict t+2 directly from the same h_t.
    Prints readable context, gold tokens, and top-5 lists.
    """
    model_two.eval()
    shown = 0

    for batch in loader:
        if shown >= n_examples: break
        # if you use DataCollatorWithPadding:
        inp  = (batch["input_ids"] if isinstance(batch, dict) else batch.input_ids).to(device)      # (B,T)
        attn = (batch["attention_mask"] if isinstance(batch, dict) else batch.attention_mask).to(device) # (B,T)

        # hidden states from the current trunk
        H = model_two.base.transformer(input_ids=inp, attention_mask=attn).last_hidden_state  # (B,T,D)
        B, T = inp.shape

        for b in range(B):
            if shown >= n_examples: break
            # valid t with room for t+1 and t+2, and not too early
            valid = ((attn[b, :-2] > 0) & (attn[b, 1:-1] > 0) & (attn[b, 2:] > 0)).nonzero(as_tuple=False).flatten()
            valid = valid[valid >= min_t]
            if len(valid) == 0:
                continue

            # pick a random source index t for this sequence
            t = int(valid[torch.randint(0, len(valid), ()).item()].item())

            # ----- Base AR step 1: predict t+1 from context -----
            ctx_ids = inp[b, :t+1]                           # x_0..x_t
            out1 = model_two.base(input_ids=ctx_ids.unsqueeze(0))
            logits_t1 = out1.logits[:, -1, :].squeeze(0)     # (V,)
            next_id   = int(logits_t1.argmax(-1).item())

            # ----- Base AR step 2: append predicted t+1; predict t+2 -----
            ctx2_ids = torch.cat([ctx_ids, torch.tensor([next_id], device=ctx_ids.device, dtype=ctx_ids.dtype)], dim=0)
            out2 = model_two.base(input_ids=ctx2_ids.unsqueeze(0))
            logits_t2_base = out2.logits[:, -1, :].squeeze(0)  # (V,)

            # ----- HEAD2: predict t+2 from same h_t -----
            h_t = H[b, t, :].unsqueeze(0)                     # (1,D)
            logits_t2_head = model_two.head_t2(h_t).squeeze(0) # (V,)

            # ----- Gold tokens (decoded, readable)
            gold_t1 = _pretty(inp[b, t+1].item(), tokenizer)
            gold_t2 = _pretty(inp[b, t+2].item(), tokenizer)

            # ----- Top-5 (decoded, readable)
            base_top5_t1  = topk_pretty(logits_t1, tokenizer, k=5)
            base_appended = _pretty(next_id, tokenizer)
            base_top5_t2  = topk_pretty(logits_t2_base, tokenizer, k=5)
            head2_top5_t2 = topk_pretty(logits_t2_head, tokenizer, k=5)

            # ----- Readable context: last N characters before t (include token at t for clarity)
            ctx_text = tokenizer.decode(inp[b, :t+1].tolist()).replace("\n", "⏎")
            ctx_snip = ctx_text[-ctx_chars:]

            print("—"*96)
            print(f"CONTEXT (last {ctx_chars} chars before t={t}): …{ctx_snip}")
            print(f"GOLD  t+1: {gold_t1}")
            print(f"BASE  t+1 top5: {base_top5_t1}")
            print(f"(appended {base_appended!r} → AR step for t+2)")
            print(f"GOLD  t+2: {gold_t2}")
            print(f"BASE  t+2 top5: {base_top5_t2}")
            print(f"HEAD2 t+2 top5 (from same h_t): {head2_top5_t2}")

            shown += 1
```

### Generate Teacher answers

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
from textwrap import dedent
import math

# pick a teacher checkpoint you can actually load on your Colab GPU
TEACHER_CKPT = "mistralai/Mistral-7B-Instruct-v0.2"  # example; change if you need smaller

device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer_t = AutoTokenizer.from_pretrained(TEACHER_CKPT)
# Mistral doesn't have a pad token by default, use eos_token
tokenizer_t.pad_token = tokenizer_t.eos_token
model_t = AutoModelForCausalLM.from_pretrained(
    TEACHER_CKPT,
    torch_dtype=torch.float16,
    device_map="auto"
)

def build_teacher_prompt(user_prompt: str) -> str:
    # simple chat-style wrap. different models want different formats;
    # adjust to match your teacher model's chat template if it has one.
    return f"<s>[INST] {user_prompt.strip()} [/INST]"

def generate_teacher_batch(batch_prompts, max_new_tokens=256):
    # batch_prompts: list[str]
    formatted = [build_teacher_prompt(p) for p in batch_prompts]

    enc = tokenizer_t(
        formatted,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=512
    ).to(device)

    with torch.no_grad():
        gen_ids = model_t.generate(
            **enc,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    # decode full sequence then strip the prompt part
    outs = []
    for i, full_ids in enumerate(gen_ids):
        full_text = tokenizer_t.decode(full_ids, skip_special_tokens=True)
        # very crude "answer = full minus prompt". adjust per model formatting.
        prompt_text = formatted[i].replace("<s>","").strip()
        if full_text.startswith(prompt_text):
            ans = full_text[len(prompt_text):].strip()
        else:
            ans = full_text.strip()
        outs.append(ans)
    return outs

# loop over all prompts in batches, call teacher, collect pairs
BATCH_SIZE = 8  # keep this small so you don't OOM
MAX_EXAMPLES = 2000  # you don't need all 15k right now; take a slice to start

distill_rows = []
for start in range(0, min(len(clean_prompts), MAX_EXAMPLES), BATCH_SIZE):
    batch_prompts = clean_prompts[start:start+BATCH_SIZE]
    batch_answers = generate_teacher_batch(batch_prompts, max_new_tokens=256)

    for p, a in zip(batch_prompts, batch_answers):
        distill_rows.append({"prompt": p, "answer": a})

    if (start // BATCH_SIZE) % 20 == 0:
        print(f"done {start + len(batch_prompts)} examples")

print("total pairs collected:", len(distill_rows))

# save JSONL so you can fine-tune your student
out_path = "/content/distill_pairs.jsonl"
with open(out_path, "w", encoding="utf-8") as f:
    for row in distill_rows:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

print("wrote", out_path)
```

### Build train data set from teacher answers

```python
import json

data_path = "/content/distill_pairs.jsonl"

pairs = []
with open(data_path, "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)
        # basic cleaning / safety
        q = obj["prompt"].strip()
        a = obj["answer"].strip()
        if len(q) > 0 and len(a) > 0:
            pairs.append((q, a))

print("num pairs:", len(pairs))
print(pairs[0][0])
print(pairs[0][1])
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer

student_tokenizer = tokenizer  # use your GPT-2 tokenizer here

# you can set padding side for causal LM to "left" or "right", but keep consistent
student_tokenizer.padding_side = "left"
if student_tokenizer.pad_token is None:
    # GPT-2 famously doesn't have a pad token; reuse eos as pad
    student_tokenizer.pad_token = student_tokenizer.eos_token


class DistillChatDataset(Dataset):
    def __init__(self, qa_pairs, tokenizer, max_len=512):
        self.qa_pairs = qa_pairs
        self.tok = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.qa_pairs)

    def __getitem__(self, idx):
        user_q, teacher_a = self.qa_pairs[idx]

        # build conversation text
        # keep formatting simple/consistent, this becomes your model's "chat style"
        prompt_text = f"User: {user_q}\nAssistant:"
        full_text   = f"{prompt_text} {teacher_a}"

        # tokenize full sequence
        enc = self.tok(
            full_text,
            truncation=True,
            max_length=self.max_len,
            return_tensors=None,
            add_special_tokens=False,
        )

        input_ids = enc["input_ids"]
        attn_mask = enc["attention_mask"]

        # we now need labels, where:
        #  - positions BEFORE the start of the assistant answer are -100
        #  - positions DURING the assistant answer are the same as input_ids
        #
        # find where the assistant answer starts, in token space
        # easiest hack: tokenize prompt_text alone, measure its length
        enc_prompt = self.tok(
            prompt_text,
            truncation=True,
            max_length=self.max_len,
            return_tensors=None,
            add_special_tokens=False,
        )
        prompt_len = len(enc_prompt["input_ids"])

        labels = [-100] * len(input_ids)
        for j in range(prompt_len, len(input_ids)):
            labels[j] = input_ids[j]

        return {
            "input_ids": torch.tensor(input_ids, dtype=torch.long),
            "attention_mask": torch.tensor(attn_mask, dtype=torch.long),
            "labels": torch.tensor(labels, dtype=torch.long),
        }


train_dataset = DistillChatDataset(pairs, student_tokenizer, max_len=512)
```

```python
def collate_batch(batch, pad_id, label_pad_id=-100):
    # batch is a list of dicts from __getitem__
    max_len = max(len(item["input_ids"]) for item in batch)

    input_ids_batch = []
    attn_batch = []
    labels_batch = []

    for item in batch:
        seq_len = len(item["input_ids"])

        # pad input_ids and attention_mask on the left (to match tokenizer.padding_side)
        pad_len = max_len - seq_len

        input_ids_pad = [pad_id] * pad_len + item["input_ids"].tolist()
        attn_pad      = [0]     * pad_len + item["attention_mask"].tolist()

        labels_seq    = item["labels"].tolist()
        labels_pad    = [label_pad_id] * pad_len + labels_seq

        input_ids_batch.append(input_ids_pad)
        attn_batch.append(attn_pad)
        labels_batch.append(labels_pad)

    return {
        "input_ids": torch.tensor(input_ids_batch, dtype=torch.long),
        "attention_mask": torch.tensor(attn_batch, dtype=torch.long),
        "labels": torch.tensor(labels_batch, dtype=torch.long),
    }


pad_id = student_tokenizer.pad_token_id
train_loader = DataLoader(
    train_dataset,
    batch_size=4,          # tune for VRAM
    shuffle=True,
    collate_fn=lambda b: collate_batch(b, pad_id),
)
```

```python
from torch.utils.data import Dataset, DataLoader
from transformers import DataCollatorWithPadding
import torch
from sklearn.model_selection import train_test_split

class DialogueCausalDataset(Dataset):
    def __init__(self, rows, tokenizer, max_len=512):
        self.samples = []

        for r in rows:
            user_q = r["prompt"]
            ans_a  = r["answer"]

            full_text = (
                "USER:\n" + user_q.strip() +
                "\n\nASSISTANT:\n" + ans_a.strip()
            )

            enc = tokenizer(
                full_text,
                truncation=True,
                max_length=max_len,
                padding=False,          # no padding here; we'll pad in collator
                return_tensors="pt",
            )

            self.samples.append({
                "input_ids": enc["input_ids"][0],
                "attention_mask": enc["attention_mask"][0],
            })

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        return {
            "input_ids": self.samples[idx]["input_ids"],
            "attention_mask": self.samples[idx]["attention_mask"],
        }

def build_loader_from_rows(rows, tokenizer, batch_size=8, max_len=512, shuffle=False):
    dataset  = DialogueCausalDataset(rows, tokenizer, max_len=max_len)
    collator = DataCollatorWithPadding(
        tokenizer,
        padding=True,
        return_tensors="pt"
    )

    loader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        collate_fn=collator,
    )
    return loader

# === train/val split happens here ===
# assumes you already have `distill_rows` in memory:
# distill_rows = [{"prompt": "...", "answer": "..."}, ...]

train_rows, val_rows = train_test_split(
    distill_rows,
    test_size=0.1,
    random_state=42,
)

print(f"train size: {len(train_rows)}")
print(f"val size:   {len(val_rows)}")

# build loaders
distill_train_loader = build_loader_from_rows(
    train_rows,
    tokenizer,
    batch_size=8,
    max_len=512,
    shuffle=True,      # shuffle for training
)

distill_val_loader = build_loader_from_rows(
    val_rows,
    tokenizer,
    batch_size=8,
    max_len=512,
    shuffle=False,     # no shuffle for eval
)
```

```python
import torch
import torch.nn.functional as F
from torch.nn.utils import clip_grad_norm_

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ---- loss helpers (same logic you already used) ----
def ce_t1(logits_t1, input_ids, attention_mask):
    B, T, V = logits_t1.shape
    if T <= 1:
        return logits_t1.sum() * 0.0  # graph-safe zero
    # predict x_{t+1} at position t
    logits = logits_t1[:, :-1, :]        # (B, T-1, V)
    targets = input_ids[:, 1:]           # (B, T-1)
    mask = attention_mask[:, 1:].bool()  # (B, T-1)
    m = mask.reshape(-1)
    if m.sum() == 0:
        return logits.sum() * 0.0
    return F.cross_entropy(
        logits.reshape(-1, V)[m],
        targets.reshape(-1)[m],
    )

def ce_t2(logits_t2, input_ids, attention_mask):
    # logits_t2 is aligned so position t tries to guess x_{t+2}
    # which means logits_t2 has shape (B, T-2, V)
    if logits_t2.size(1) == 0:
        return logits_t2.sum() * 0.0
    targets = input_ids[:, 2:]  # gold x_{t+2}
    mask2 = (attention_mask[:, :-2].bool() & attention_mask[:, 2:].bool())
    m2 = mask2.reshape(-1)
    if m2.sum() == 0:
        return logits_t2.sum() * 0.0
    V = logits_t2.size(-1)
    return F.cross_entropy(
        logits_t2.reshape(-1, V)[m2],
        targets.reshape(-1)[m2],
    )

# ---- training loop over the distilled data only ----
def train_distill_two_head(
    model_two,
    loader,
    epochs=3,
    lam1=1.0,
    lam2=1.0,
    lr=5e-5,
    grad_clip=1.0,
    log_every=50,
):
    model_two.to(device)
    model_two.train()

    # IMPORTANT: now we assume trunk is unfrozen, so include all params
    optim = torch.optim.AdamW(model_two.parameters(), lr=lr, weight_decay=0.01)

    for e in range(1, epochs+1):
        running_loss = 0.0
        steps = 0
        for step, batch in enumerate(loader, start=1):
            inp  = batch["input_ids"].to(device)
            attn = batch["attention_mask"].to(device)

            logits_t1, logits_t2 = model_two(inp, attention_mask=attn)

            loss1 = ce_t1(logits_t1, inp, attn)
            loss2 = ce_t2(logits_t2, inp, attn)
            loss  = lam1*loss1 + lam2*loss2

            optim.zero_grad(set_to_none=True)
            loss.backward()
            if grad_clip is not None:
                clip_grad_norm_(model_two.parameters(), grad_clip)
            optim.step()

            running_loss += loss.item()
            steps += 1

            if step % log_every == 0:
                print(f"[ep {e} | step {step}] "
                      f"loss={running_loss/steps:.3f} "
                      f"(t+1 {loss1.item():.3f} | t+2 {loss2.item():.3f})")

        print(f"[epoch {e}] avg_loss={running_loss/max(1,steps):.4f}")
```

```python
import math
import torch
import torch.nn.functional as F

@torch.no_grad()
def eval_two_head(model_two, loader, ks=(1,3,5), debug=False):
    """
    Quantitative eval for the 2-head model:
      - Head1 (t+1): teacher-forced cross-entropy → perplexity
      - Head2 (t+2): hits@k for the 'next+1' token task

    Works with:
      - dict batches (batch["input_ids"])
      - CausalBatch-style batches (batch.input_ids)
    """
    model_two.eval()

    ce_sum = 0.0
    tok_count = 0

    hits = {k: 0.0 for k in ks}
    t2_count = 0

    for batch in loader:
        # --- unify batch access ---
        if isinstance(batch, dict):
            inp  = batch["input_ids"].to(device)        # (B,T)
            attn = batch["attention_mask"].to(device)   # (B,T)
        else:
            inp  = batch.input_ids.to(device)           # (B,T)
            attn = batch.attention_mask.to(device)      # (B,T)

        logits_t1, logits_t2 = model_two(inp, attention_mask=attn)
        # logits_t1: (B,T,V)  predicts token t+1 at position t
        # logits_t2: (B,T-2,V) predicts token t+2 at position t

        # ====== Head1 perplexity ======
        # shift so position t predicts gold at t+1
        if logits_t1.size(1) > 1:
            logits   = logits_t1[:, :-1, :]      # (B,T-1,V)
            targets  = inp[:, 1:]               # (B,T-1)
            mask     = attn[:, 1:].bool()       # (B,T-1)

            flat_mask = mask.view(-1)
            if flat_mask.any():
                V = logits.size(-1)
                ce_sum += F.cross_entropy(
                    logits.reshape(-1, V)[flat_mask],
                    targets.reshape(-1)[flat_mask],
                    reduction="sum"
                ).item()
                tok_count += int(flat_mask.sum().item())

        # ====== Head2 hits@k ======
        if logits_t2.size(1) > 0:
            gold_t2   = inp[:, 2:]  # gold token at t+2
            validmask = (attn[:, :-2].bool() & attn[:, 2:].bool())  # (B,T-2)

            flat_valid = validmask.view(-1)
            if flat_valid.any():
                V2 = logits_t2.size(-1)
                logits_flat  = logits_t2.reshape(-1, V2)[flat_valid]   # (N,V2)
                targets_flat = gold_t2.reshape(-1)[flat_valid]         # (N,)
                t2_count += targets_flat.size(0)

                for k in ks:
                    topk_idx = logits_flat.topk(min(k, V2), dim=-1).indices  # (N,k)
                    correct_any = (topk_idx == targets_flat.unsqueeze(-1)).any(-1)
                    hits[k] += correct_any.float().sum().item()

    ppl = float(math.exp(ce_sum / tok_count)) if tok_count > 0 else float("nan")
    result = {
        "t+1_perplexity": ppl,
        "t+1_token_count": tok_count,
        "t+2_count": t2_count,
    }
    for k in ks:
        result[f"hits@{k}"] = hits[k] / max(1, t2_count)

    if debug:
        print("DEBUG ce_sum:", ce_sum, "tok_count:", tok_count)

    return result
```

```python
import torch

def tok_pretty(tok_id, tokenizer):
    s = tokenizer.convert_ids_to_tokens([int(tok_id)])[0]
    if s == "\n":
        return "⏎"
    if s == " ":
        return "␠"
    if s.startswith(" "):
        return "␠" + s[1:]
    return s

def topk_pretty_vec(logits_1d, tokenizer, k=5):
    top_ids = logits_1d.topk(k, dim=-1).indices.tolist()
    toks = [tok_pretty(t, tokenizer) for t in top_ids]
    return "[" + ", ".join(toks) + "]"

@torch.no_grad()
def show_qual_two_head(model_two, loader, tokenizer, n_examples=5, min_t=5, ctx_chars=200):
    """
    For random index t:
      - show text context up to t
      - gold t+1, gold t+2
      - head1 guess for t+1 (top5)
      - AR rollout guess for t+2 (top5)
      - head2 direct guess for t+2 (top5)
    """
    model_two.eval()
    shown = 0

    for batch in loader:
        if shown >= n_examples:
            break

        # --- unify batch access ---
        if isinstance(batch, dict):
            inp  = batch["input_ids"].to(device)        # (B,T)
            attn = batch["attention_mask"].to(device)   # (B,T)
        else:
            inp  = batch.input_ids.to(device)
            attn = batch.attention_mask.to(device)

        # hidden states from trunk for head2
        H = model_two.base.transformer(
            input_ids=inp,
            attention_mask=attn
        ).last_hidden_state  # (B,T,D)

        B, T = inp.shape

        for b in range(B):
            if shown >= n_examples:
                break

            # valid time steps: t, t+1, t+2 must be real tokens
            valid_idx = (
                (attn[b, :-2] > 0) &
                (attn[b, 1:-1] > 0) &
                (attn[b, 2:] > 0)
            ).nonzero(as_tuple=False).flatten()

            # avoid first few tokens so context isn't nonsense
            valid_idx = valid_idx[valid_idx >= min_t]
            if len(valid_idx) == 0:
                continue

            # pick random t
            t = int(valid_idx[torch.randint(0, len(valid_idx), ()).item()].item())

            # ===== head1 next-token prediction =====
            ctx_ids = inp[b, :t+1]  # tokens [0..t]
            out1 = model_two.base(input_ids=ctx_ids.unsqueeze(0))
            logits_t1_last = out1.logits[:, -1, :].squeeze(0)  # (V,)
            next_id = int(logits_t1_last.argmax(-1).item())

            # ===== AR rollout for t+2 =====
            ctx2_ids = torch.cat(
                [ctx_ids, torch.tensor([next_id], device=ctx_ids.device, dtype=ctx_ids.dtype)],
                dim=0
            )
            out2 = model_two.base(input_ids=ctx2_ids.unsqueeze(0))
            logits_t2_ar = out2.logits[:, -1, :].squeeze(0)  # (V,)

            # ===== head2 direct t+2 prediction from h_t =====
            h_t = H[b, t, :].unsqueeze(0)                 # (1,D)
            logits_t2_head2 = model_two.head_t2(h_t).squeeze(0)  # (V,)

            # ===== gold tokens =====
            gold_t1 = tok_pretty(inp[b, t+1].item(), tokenizer)
            gold_t2 = tok_pretty(inp[b, t+2].item(), tokenizer)

            # ===== readable context snippet =====
            raw_ctx_text = tokenizer.decode(inp[b, :t+1].tolist())
            visible_ctx = raw_ctx_text.replace("\n", "⏎")
            ctx_snip = visible_ctx[-ctx_chars:]

            # ===== print block =====
            print("—"*96)
            print(f"CONTEXT (…last {ctx_chars} chars before t={t}): {ctx_snip}")
            print(f"GOLD t+1: {gold_t1}")
            print(f"HEAD1 t+1 top5: {topk_pretty_vec(logits_t1_last, tokenizer, k=5)}")
            print(f"(we appended head1 argmax '{tok_pretty(next_id, tokenizer)}' and rolled AR to get t+2)")
            print(f"GOLD t+2: {gold_t2}")
            print(f"AR rollout t+2 top5: {topk_pretty_vec(logits_t2_ar, tokenizer, k=5)}")
            print(f"HEAD2 direct t+2 top5: {topk_pretty_vec(logits_t2_head2, tokenizer, k=5)}")

            shown += 1
```

```python
import torch
import torch.nn.functional as F

@torch.no_grad()
def generate_dual_stream(
    model_two,
    tokenizer,
    user_message,
    steps=40,
    temperature=0.7,
    top_k=40,
):
    """
    We create TWO continuations from the same prompt:

    stream_head1:
        append only head1's chosen next token each step
        (this is "normal" autoregressive generation)

    stream_combo:
        append head1 token AND head2 token each step
        so we kind of 'time warp' 2 tokens into the response at once

    We'll compare both at the end.
    """

    model_two.eval()

    # 1. Build chat-style prefix
    prefix = (
        "USER:\n" + user_message.strip() +
        "\n\nASSISTANT:\n"
    )
    enc = tokenizer(prefix, return_tensors="pt")
    base_ids = enc["input_ids"].to(device)        # (1, T0)
    base_att = enc["attention_mask"].to(device)   # (1, T0)

    # We'll clone that state for both streams
    ids_h1   = base_ids.clone()
    attn_h1  = base_att.clone()

    ids_combo  = base_ids.clone()
    attn_combo = base_att.clone()

    # sampling helper
    def sample_from_logits(logits_row):
        """Take a single (V,) logits row, apply temp+top_k, return sampled token_id int."""
        l = logits_row.clone()
        if temperature and temperature > 0.0:
            l = l / temperature
        if top_k and top_k > 0:
            topk_vals, topk_idx = torch.topk(l, k=top_k, dim=-1)
            filtered = torch.full_like(l, float("-inf"))
            filtered.scatter_(0, topk_idx, topk_vals)
            l = filtered
        probs = F.softmax(l, dim=-1)
        tok = torch.multinomial(probs, num_samples=1)
        return int(tok.item())

    # we will capture trace for debugging
    trace_rows = []

    for step in range(steps):

        # ===== STREAM 1: HEAD1-ONLY ==================================
        # forward pass
        logits_t1_h1, logits_t2_h1 = model_two(ids_h1, attention_mask=attn_h1)
        # head1 next-token is from last position
        next_logits_h1 = logits_t1_h1[0, -1, :]  # (V,)
        tok_h1 = sample_from_logits(next_logits_h1)

        # head2 proposed token, aligned to t+2
        # recall logits_t2 has shape (1, T-2, V). last index j predicts j+2.
        # we just grab the last available row to represent head2's strongest guess.
        if logits_t2_h1.size(1) > 0:
            head2_logits_last = logits_t2_h1[0, -1, :]  # (V,)
            tok_h2 = sample_from_logits(head2_logits_last)
        else:
            tok_h2 = None

        # append tok_h1 to the head1 stream
        new_tok_h1 = torch.tensor([[tok_h1]], device=ids_h1.device, dtype=ids_h1.dtype)
        ids_h1  = torch.cat([ids_h1, new_tok_h1], dim=1)
        attn_h1 = torch.cat([attn_h1, torch.ones_like(new_tok_h1)], dim=1)

        # ===== STREAM 2: COMBO (HEAD1 + HEAD2) ========================
        # forward pass for *its own* current ids
        logits_t1_combo, logits_t2_combo = model_two(ids_combo, attention_mask=attn_combo)

        # head1 pick for combo stream
        next_logits_combo_h1 = logits_t1_combo[0, -1, :]
        tok_h1_combo = sample_from_logits(next_logits_combo_h1)

        # head2 pick for combo stream
        if logits_t2_combo.size(1) > 0:
            head2_logits_last_combo = logits_t2_combo[0, -1, :]
            tok_h2_combo = sample_from_logits(head2_logits_last_combo)
        else:
            tok_h2_combo = None

        # now actually APPEND BOTH to combo stream
        # always append tok_h1_combo
        chunk = [tok_h1_combo]
        # append head2 suggestion if it exists
        if tok_h2_combo is not None:
            chunk.append(tok_h2_combo)

        new_chunk = torch.tensor([chunk], device=ids_combo.device, dtype=ids_combo.dtype)
        # attention mask chunk of 1s same length
        new_mask_chunk = torch.ones_like(new_chunk)

        ids_combo  = torch.cat([ids_combo, new_chunk], dim=1)
        attn_combo = torch.cat([attn_combo, new_mask_chunk], dim=1)

        # record debug info so we can inspect alignment drift
        trace_rows.append({
            "step": step,
            "h1_tok"      : tokenizer.decode([tok_h1], skip_special_tokens=False),
            "h2_tok"      : (tokenizer.decode([tok_h2], skip_special_tokens=False) if tok_h2 is not None else None),
            "combo_h1_tok": tokenizer.decode([tok_h1_combo], skip_special_tokens=False),
            "combo_h2_tok": (tokenizer.decode([tok_h2_combo], skip_special_tokens=False) if tok_h2_combo is not None else None),
        })

        # crude stopping heuristic for both streams:
        # if either stream starts outputting a new "USER:" marker, bail
        decoded_h1 = tokenizer.decode(
            ids_h1[0, base_ids.size(1):].tolist(),
            skip_special_tokens=False
        )
        decoded_combo = tokenizer.decode(
            ids_combo[0, base_ids.size(1):].tolist(),
            skip_special_tokens=False
        )
        if "USER:" in decoded_h1 and len(decoded_h1) > 20:
            break
        if "USER:" in decoded_combo and len(decoded_combo) > 20:
            break

    # final decodes
    final_h1 = tokenizer.decode(
        ids_h1[0, base_ids.size(1):].tolist(),
        skip_special_tokens=False
    )
    final_combo = tokenizer.decode(
        ids_combo[0, base_ids.size(1):].tolist(),
        skip_special_tokens=False
    )

    # trim if we hallucinated a new USER: prompt
    if "USER:" in final_h1:
        final_h1 = final_h1.split("USER:")[0]
    if "USER:" in final_combo:
        final_combo = final_combo.split("USER:")[0]

    # print results
    print("============== TRACE (per step) ==============")
    for row in trace_rows:
        print(
            f"[step {row['step']}] "
            f"h1='{row['h1_tok']}' h2='{row['h2_tok']}' | "
            f"combo_h1='{row['combo_h1_tok']}' combo_h2='{row['combo_h2_tok']}'"
        )

    print("\n============== FINAL NORMAL (HEAD1 ONLY) ==============")
    print(final_h1.strip())

    print("\n============== FINAL COMBO (HEAD1+HEAD2 INTERLEAVED) ==============")
    print(final_combo.strip())

    return final_h1.strip(), final_combo.strip()


# EXAMPLE USAGE:
normal_out, combo_out = generate_dual_stream(
    model_two,
    tokenizer,
    user_message="Explain how to safely change a flat tire on the highway.",
    steps=60,
    temperature=0.7,
    top_k=40,
)
```

# Deep Training for 2nd Head Using HuggingFace Fine Web Data Set

## Dual-Head GPT-2 on FineWeb (t+2 Look-Ahead Head)

In this section, we start building a **multi-token GPT-style model** by adding a second output head to a pretrained GPT-2 and training it to predict tokens **two steps into the future**.

### Model setup

- We load a pretrained **GPT-2** model (decoder-only transformer) and its tokenizer from Hugging Face.
- GPT-2 normally has a **single output head** (`lm_head`) that predicts the **next token** \(x_{t+1}\) from the final hidden state at position \(t\).
- We:
  - **Freeze** the entire GPT-2 trunk (all transformer layers + original `lm_head`) so we do not disturb its existing next-token behavior.
  - Add a **new linear head** (`second_head`) on top of the final hidden states.
  - Train this new head to predict **\(x_{t+2}\)** (the token two positions ahead), i.e. to approximate \(P(x_{t+2} \mid x_{\le t})\).
- During this first phase, **only the new head learns**. This lets us see how much “look-ahead” signal is already present in GPT-2’s frozen representations.

### Dataset: FineWeb and our subset

We train on **FineWeb**, a large-scale, high-quality web text corpus released by Hugging Face:

- **Dataset:** `HuggingFaceFW/fineweb`
- **Full corpus:** trillions of tokens (18.5T tokens) of filtered, deduplicated web text designed specifically for LLM pretraining.
- We do not download the whole dataset; instead, we use it in **streaming mode** so examples are loaded on the fly.

For practicality and reproducibility, we use a specific **subset configuration**:

- **Subset configuration:** `"sample-10BT"`
  - This is a ~10B-token sample of FineWeb.
  - It is large enough to expose rich long-range language patterns, but still manageable for experimental training on a single GPU.
- Each raw text document is:
  1. **Tokenized** with the GPT-2 tokenizer.
  2. Packed into a rolling buffer of token IDs.
  3. Sliced into fixed-length **blocks** of size `BLOCK_SIZE` (e.g., 256 tokens).

From each block of length `BLOCK_SIZE + 2`:

- The **inputs** are the first `BLOCK_SIZE` tokens: positions \(t, t+1, \dots, t + \text{BLOCK\_SIZE} - 1\).
- The **labels** for the new head are the same block shifted by 2: positions \(t+2, \dots, t + \text{BLOCK\_SIZE} + 1\).

This means that for every token position in the input, the new head learns to predict the token **two steps ahead**. Over time, this should teach the second head to exploit whatever future-relevant structure is already encoded in GPT-2’s hidden states, without changing the base model yet.


```python
# ============================================================
# Dual-head GPT-2 on FineWeb: training a t+2 look-ahead head
# ============================================================
# This cell assumes a fresh Colab runtime with a GPU (A100 if possible).
# It:
#  1. Loads GPT-2 and its tokenizer.
#  2. Streams the FineWeb dataset (sample-10BT subset).
#  3. Builds an iterable PyTorch dataset of fixed-length token chunks.
#  4. Adds a new linear head that predicts token t+2.
#  5. Freezes the GPT-2 trunk + original head; trains only the new head.
#  6. Prints clean training progress (loss + accuracy).
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import math
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader

from datasets import load_dataset
from transformers import AutoTokenizer, GPT2LMHeadModel, get_cosine_schedule_with_warmup

from tqdm.auto import tqdm

# -------------------------------
# 1. Basic config & device setup
# -------------------------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True  # speedup on Ampere
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16  # or torch.float16, bf16 is safer numerically

MODEL_NAME = "gpt2"          # can try "gpt2-medium" later
BLOCK_SIZE = 512             # number of input tokens per training example
BATCH_SIZE = 32              # adjust based on GPU memory, A100 should be fine here (40gb VRAM)
LR = 5e-4                    # learning rate for the new head, a bit conservative for a single linear head
WARMUP_STEPS = 2_000         # LR warmup steps
TRAINING_STEPS = 100_000      # adjust based on patience / time
LOG_EVERY = 100              # how often to log progress

# -------------------------------
# 2. Load tokenizer & GPT-2 model
# -------------------------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
# GPT-2 has no pad token by default; we add one for batching convenience.
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = GPT2LMHeadModel.from_pretrained(MODEL_NAME)
model = model.to(DEVICE).to(dtype)
model.eval()  # trunk is frozen, so no need for dropout during our head training

# Freeze all GPT-2 parameters (trunk + original lm_head)
for param in model.parameters():
    param.requires_grad = False

hidden_size = model.lm_head.in_features
vocab_size = model.config.vocab_size

# -------------------------------------
# 3. Add the new t+2 head (random init)
# -------------------------------------
# This head will learn to map hidden states h_t to logits for x_{t+2}.
second_head = nn.Linear(hidden_size, vocab_size, device=DEVICE, dtype=dtype)

# Only this head will be trained at first.
optimizer = torch.optim.AdamW(second_head.parameters(), lr=LR)

# We'll define the scheduler once we know total steps (we'll use TRAINING_STEPS).
scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=WARMUP_STEPS,
    num_training_steps=TRAINING_STEPS,
)

# --------------------------------------------------------
# 4. FineWeb streaming dataset → token chunks for training
# --------------------------------------------------------
# FineWeb is huge. We stream a sample subset (sample-10BT) and
# convert it into fixed-length sequences of GPT-2 tokens.
#
# Each example we produce is:
#   input_ids: [BLOCK_SIZE] tokens       (positions t .. t+BLOCK_SIZE-1)
#   labels:    [BLOCK_SIZE] tokens       (positions t+2 .. t+BLOCK_SIZE+1)
#
# So the head learns P(x_{t+2} | x_{<=t}).

print("Loading FineWeb (sample-10BT) in streaming mode...")
hf_stream = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",          # ~10B-token subset; good for experiments
    split="train",
    streaming=True,
)

class FineWebTokenDataset(IterableDataset):
    """
    Iterable PyTorch dataset that:
      - Reads raw text examples from a streaming HF dataset.
      - Tokenizes them with GPT-2 tokenizer.
      - Maintains a rolling buffer of tokens.
      - Yields fixed-length (BLOCK_SIZE) input sequences and t+2 labels.
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        buffer = []
        # We stream documents one by one and pack their tokens.
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            # Convert a piece of text into GPT-2 token IDs.
            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False
            )
            if not ids:
                continue

            buffer.extend(ids)

            # While we have enough tokens to make a full (BLOCK_SIZE + 2) window:
            #   - we use the first BLOCK_SIZE as inputs
            #   - and the next BLOCK_SIZE as "shifted by 2" labels
            #
            # That is, for positions 0..BLOCK_SIZE-1:
            #   label[i] = token[i + 2]
            #
            # This gives P(x_{t+2} | x_{<=t}).
            while len(buffer) >= self.block_size + 2:
                # Take a slice of length BLOCK_SIZE+2
                chunk = buffer[: self.block_size + 2]
                # Drop BLOCK_SIZE tokens from the front to advance the window
                buffer = buffer[self.block_size :]

                input_ids = chunk[: self.block_size]          # t .. t+BLOCK_SIZE-1
                labels = chunk[2 : self.block_size + 2]       # t+2 .. t+BLOCK_SIZE+1

                yield {
                    "input_ids": torch.tensor(input_ids, dtype=torch.long),
                    "labels": torch.tensor(labels, dtype=torch.long),
                }

train_iterable = FineWebTokenDataset(hf_stream, tokenizer, BLOCK_SIZE)

# DataLoader will automatically batch the tensors from our iterable dataset.
train_loader = DataLoader(
    train_iterable,
    batch_size=BATCH_SIZE,
    num_workers=4,      # A100 + Colab High-RAM should like 4 workers
    pin_memory=True,    # faster host→device copies
)

# --------------------------------------------------------
# 5. Training loop: train only the new t+2 head
# --------------------------------------------------------
# We:
#   - Compute hidden states from the frozen GPT-2 transformer.
#   - Apply second_head to get logits for x_{t+2}.
#   - Compute cross-entropy loss.
#   - Backprop through second_head only.
#   - Log loss and accuracy every LOG_EVERY steps.
# --------------------------------------------------------

torch.cuda.empty_cache()

step = 0
running_loss = 0.0
running_correct = 0
running_tokens = 0

progress_bar = tqdm(total=TRAINING_STEPS, desc="Training (t+2 head)", unit="step")

while step < TRAINING_STEPS:
    for batch in train_loader:
        step += 1
        if step > TRAINING_STEPS:
            break

        input_ids = batch["input_ids"].to(DEVICE)    # [B, T]
        labels = batch["labels"].to(DEVICE)          # [B, T]

        # 1. Get hidden states from the frozen GPT-2 trunk.
        #    We do this without gradients to save memory & compute.
        with torch.no_grad():
            outputs = model.transformer(input_ids)
            hidden_states = outputs.last_hidden_state.to(dtype)  # [B, T, hidden_size]

        # 2. Apply the new head to get logits for each position.
        logits = second_head(hidden_states)  # [B, T, vocab_size]

        # 3. Reshape for cross-entropy:
        #    - Flatten token dimension.
        #    - Predict labels for every token in the sequence.
        B, T, V = logits.shape
        loss = F.cross_entropy(
            logits.view(B * T, V),
            labels.view(B * T)
        )

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        optimizer.step()
        scheduler.step()

        # 4. Compute accuracy for monitoring (top-1).
        with torch.no_grad():
            preds = logits.argmax(dim=-1)  # [B, T]
            correct = (preds == labels).sum().item()
            total = labels.numel()

        running_loss += loss.item()
        running_correct += correct
        running_tokens += total

        # 5. Logging
        if step % LOG_EVERY == 0:
            avg_loss = running_loss / LOG_EVERY
            avg_acc = running_correct / running_tokens

            lr = scheduler.get_last_lr()[0]
            progress_bar.set_postfix({
                "loss": f"{avg_loss:.4f}",
                "acc": f"{avg_acc:.4f}",
                "lr": f"{lr:.2e}"
            })
            progress_bar.update(LOG_EVERY)

            running_loss = 0.0
            running_correct = 0
            running_tokens = 0

progress_bar.close()
print("Training finished.")
```

## Stage 2: Joint Training of GPT-2 with Dual Heads

In Stage 1, we took a pretrained **GPT-2** model and added a new output head:

- **Head₁ (original `lm_head`)**: predicts the **next token** \(x_{t+1}\).
- **Head₂ (new `second_head`)**: predicts the token **two steps ahead** \(x_{t+2}\).

We:
- **Froze** the entire GPT-2 trunk and the original head.
- Trained only Head₂ on **FineWeb** (`HuggingFaceFW/fineweb`, `sample-10BT` subset).
- Used a sliding-window dataset where each example provided:
  - `input_ids[t]` → current token
  - `labels[t]`   → token at \(t+2\)

After **100k steps**, the new head reached a stable performance (top-1 accuracy around ~0.11 on \(t+2\) prediction), which is substantially above random and suggests that GPT-2’s frozen representations do contain some look-ahead signal.

Now, in **Stage 2**, we:

- **Unfreeze** the GPT-2 trunk and original head.
- Train **both heads jointly** on the same FineWeb stream.
- Use a combined loss:
  \[
    \mathcal{L} = \mathcal{L}_{t+1} + \lambda \,\mathcal{L}_{t+2}
  \]
  where:
  - \(\mathcal{L}_{t+1}\) is the standard next-token loss for Head₁,
  - \(\mathcal{L}_{t+2}\) is the future-token loss for Head₂,
  - \(\lambda\) controls how strongly we optimize for the t+2 objective.

The goal is to let the model **reshape its internal representations** so that both heads can cooperate:
- Head₁ remains a strong next-token predictor.
- Head₂ becomes a more accurate look-ahead head, now supported by a trunk that is actually trained for multi-token prediction.

We reuse the same FineWeb subset because it is extremely large, minimizing overfitting concerns even when we fine-tune the full model.


## Joint Fine-Tuning Code Overview

The following code cell implements **Stage 2** of our training:

1. **Unfreezing parameters & optimizer groups**
   - We re-enable gradients for:
     - The GPT-2 **transformer trunk**
     - The original **`lm_head`** (Head₁)
     - The new **`second_head`** (Head₂)
   - We create an optimizer with two learning-rate groups:
     - A **small learning rate** for the transformer trunk (to avoid damaging the pretrained backbone).
     - A **larger learning rate** for both heads (since they are small and need to adapt more quickly).

2. **New dataset with t+1 and t+2 labels**
   - We define `FineWebMultiTokenDataset`, an iterable dataset that:
     - Streams **FineWeb / sample-10BT** in text form.
     - Tokenizes text into GPT-2 tokens.
     - Packs tokens into a rolling buffer.
     - Produces fixed-length sequences of length `BLOCK_SIZE + 2` and slices them into:
       - `input_ids`  : tokens at positions \(t .. t+T-1\)
       - `labels_t1` : tokens at positions \(t+1 .. t+T\)  (for Head₁ / t+1)
       - `labels_t2` : tokens at positions \(t+2 .. t+T+1\) (for Head₂ / t+2)

3. **Forward pass and losses**
   - For each batch:
     - We pass `input_ids` through the GPT-2 transformer to get hidden states.
     - Head₁ (`lm_head`) predicts logits for **t+1**.
     - Head₂ (`second_head`) predicts logits for **t+2**.
   - We compute:
     - Cross-entropy loss for Head₁ vs `labels_t1`
     - Cross-entropy loss for Head₂ vs `labels_t2`
   - Total loss:
     \[
       \mathcal{L}_\text{total} = \mathcal{L}_{t+1} + \lambda \,\mathcal{L}_{t+2}
     \]

4. **Training loop and logging**
   - We train for a configurable number of `TRAINING_STEPS_STAGE2`.
   - Every `LOG_EVERY` steps, we log:
     - `loss_t1`, `loss_t2`
     - `acc_t1`, `acc_t2` (top-1 token accuracy)
     - current learning rate
   - This lets us **watch both heads’ performance evolve**, and see whether they stabilize together as the trunk adapts to the multi-token objective.

This gives us a clean, professional fine-tuning setup that:
- Uses **full backpropagation** through the model.
- Maintains a clear split between next-token and future-token behavior.
- Is still relatively safe from overfitting due to the size of the FineWeb corpus.

```python
# ============================================================
# Stage 2: Joint fine-tuning of GPT-2 trunk + dual heads
# ============================================================
# We now:
#   - Unfreeze the GPT-2 transformer and original lm_head.
#   - Train both Head₁ (t+1) and Head₂ (t+2) jointly.
#   - Use a combined loss: L = L_{t+1} + lambda * L_{t+2}.
#   - Reuse the FineWeb sample-10BT stream with a new dataset
#     that provides both t+1 and t+2 labels.
#   - Periodically save checkpoints (model + optimizer + step)
#     so we can recover from Colab timeouts or resume training.
# ============================================================

import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader

from datasets import load_dataset
from tqdm.auto import tqdm
from transformers import get_cosine_schedule_with_warmup

# (Colab) Mount Google Drive to store checkpoints persistently
from google.colab import drive
drive.mount("/content/drive")

# Directory where checkpoints will be saved
CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
os.makedirs(CKPT_DIR, exist_ok=True)
CKPT_BASENAME = "stage2_gpt2_dual_head"

# -------------------------------
# 1. Stage 2 training config
# -------------------------------
LR_TRUNK = 1e-5           # small LR for the pretrained transformer trunk
LR_HEADS = 3e-4           # larger LR for both heads
LAMBDA_T2 = 0.5           # weight for the t+2 loss
TRAINING_STEPS_STAGE2 = 10_000
LOG_EVERY = 100
CHECKPOINT_EVERY = 2_000  # save checkpoint every N steps

# (Optional) Make tokenizer stop complaining about long docs.
tokenizer.model_max_length = BLOCK_SIZE + 2

# -------------------------------
# 2. Unfreeze model parameters
# -------------------------------
# We want gradients through:
#   - model.transformer (the trunk)
#   - model.lm_head     (Head₁: t+1)
#   - second_head       (Head₂: t+2)

for param in model.transformer.parameters():
    param.requires_grad = True

for param in model.lm_head.parameters():
    param.requires_grad = True

for param in second_head.parameters():
    param.requires_grad = True

# Build parameter groups for the optimizer.
# GPT-2 ties input embeddings and lm_head weights, so the same
# underlying Parameter can appear under multiple names
# (e.g., "transformer.wte.weight" and "lm_head.weight").
# We must ensure each Parameter object is only in ONE group.

trunk_params = []
head1_params = []
head2_params = list(second_head.parameters())

seen = set()  # track Parameter objects we've already assigned

for name, param in model.named_parameters():
    if not param.requires_grad:
        continue
    if id(param) in seen:
        # This Parameter is already assigned to a group; skip duplicates.
        continue
    if name.startswith("transformer."):
        trunk_params.append(param)
        seen.add(id(param))
    elif name.startswith("lm_head."):
        head1_params.append(param)
        seen.add(id(param))
    # Note: second_head is a separate module, so its params are handled
    # via head2_params and do not appear in model.named_parameters().

optimizer = torch.optim.AdamW(
    [
        {"params": trunk_params, "lr": LR_TRUNK},
        {"params": head1_params, "lr": LR_HEADS},
        {"params": head2_params, "lr": LR_HEADS},
    ],
    weight_decay=0.01,
)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=2_000,
    num_training_steps=TRAINING_STEPS_STAGE2,
)

# --------------------------------------------------------
# 3. FineWeb dataset with both t+1 and t+2 labels
# --------------------------------------------------------
print("Loading FineWeb (sample-10BT) in streaming mode for Stage 2...")
hf_stream_stage2 = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

class FineWebMultiTokenDataset(IterableDataset):
    """
    Iterable dataset yielding:
      - input_ids  : tokens at positions t .. t+T-1
      - labels_t1  : tokens at positions t+1 .. t+T   (for t+1 head)
      - labels_t2  : tokens at positions t+2 .. t+T+1 (for t+2 head)
    using a rolling buffer over the FineWeb token stream.
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        buffer = []
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            # Tokenize the document
            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if not ids:
                continue

            buffer.extend(ids)

            # While we have enough tokens for BLOCK_SIZE + 2:
            # chunk = [c0, c1, ..., c_{T+1}]
            # input_ids = [c0..c_{T-1}]
            # labels_t1 = [c1..c_{T}]
            # labels_t2 = [c2..c_{T+1}]
            while len(buffer) >= self.block_size + 2:
                chunk = buffer[: self.block_size + 2]
                buffer = buffer[self.block_size :]

                input_ids = chunk[: self.block_size]            # [T]
                labels_t1 = chunk[1 : self.block_size + 1]      # [T]
                labels_t2 = chunk[2 : self.block_size + 2]      # [T]

                yield {
                    "input_ids": torch.tensor(input_ids, dtype=torch.long),
                    "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                    "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
                }

train_iterable_stage2 = FineWebMultiTokenDataset(hf_stream_stage2, tokenizer, BLOCK_SIZE)

train_loader_stage2 = DataLoader(
    train_iterable_stage2,
    batch_size=BATCH_SIZE,
    num_workers=4,
    pin_memory=True,
)

# --------------------------------------------------------
# 4. Stage 2 training loop: joint t+1 and t+2 + checkpoints
# --------------------------------------------------------

torch.cuda.empty_cache()

step = 0
running_loss_t1 = 0.0
running_loss_t2 = 0.0
running_correct_t1 = 0
running_correct_t2 = 0
running_tokens = 0

progress_bar = tqdm(total=TRAINING_STEPS_STAGE2, desc="Stage 2: joint training", unit="step")

model.train()  # enable training mode for the now-unfrozen trunk

while step < TRAINING_STEPS_STAGE2:
    for batch in train_loader_stage2:
        step += 1
        if step > TRAINING_STEPS_STAGE2:
            break

        input_ids = batch["input_ids"].to(DEVICE)    # [B, T]
        labels_t1 = batch["labels_t1"].to(DEVICE)    # [B, T]
        labels_t2 = batch["labels_t2"].to(DEVICE)    # [B, T]

        # Forward pass through the transformer (with gradients this time).
        outputs = model.transformer(input_ids)
        hidden_states = outputs.last_hidden_state     # [B, T, H]

        # Head₁: next-token logits (t+1)
        logits_t1 = model.lm_head(hidden_states)      # [B, T, V]
        # Head₂: future-token logits (t+2)
        logits_t2 = second_head(hidden_states)        # [B, T, V]

        B, T, V = logits_t1.shape

        # Compute losses.
        # We convert logits to float32 for numerical stability in cross-entropy.
        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )

        loss = loss_t1 + LAMBDA_T2 * loss_t2

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        # Optional: gradient clipping to be extra safe
        torch.nn.utils.clip_grad_norm_(trunk_params + head1_params + head2_params, max_norm=1.0)
        optimizer.step()
        scheduler.step()

        # Compute accuracies for monitoring.
        with torch.no_grad():
            preds_t1 = logits_t1.argmax(dim=-1)  # [B, T]
            preds_t2 = logits_t2.argmax(dim=-1)  # [B, T]

            correct_t1 = (preds_t1 == labels_t1).sum().item()
            correct_t2 = (preds_t2 == labels_t2).sum().item()
            total_tokens = labels_t1.numel()

        running_loss_t1 += loss_t1.item()
        running_loss_t2 += loss_t2.item()
        running_correct_t1 += correct_t1
        running_correct_t2 += correct_t2
        running_tokens += total_tokens

        # Logging + checkpointing
        if step % LOG_EVERY == 0:
            avg_loss_t1 = running_loss_t1 / LOG_EVERY
            avg_loss_t2 = running_loss_t2 / LOG_EVERY
            avg_acc_t1 = running_correct_t1 / running_tokens
            avg_acc_t2 = running_correct_t2 / running_tokens

            lr = scheduler.get_last_lr()[0]
            progress_bar.set_postfix({
                "L_t1": f"{avg_loss_t1:.3f}",
                "L_t2": f"{avg_loss_t2:.3f}",
                "acc_t1": f"{avg_acc_t1:.3f}",
                "acc_t2": f"{avg_acc_t2:.3f}",
                "lr": f"{lr:.2e}",
            })
            progress_bar.update(LOG_EVERY)

            running_loss_t1 = 0.0
            running_loss_t2 = 0.0
            running_correct_t1 = 0
            running_correct_t2 = 0
            running_tokens = 0

            # ---- Periodic checkpointing ----
            if step % CHECKPOINT_EVERY == 0:
                ckpt_path = f"{CKPT_DIR}/{CKPT_BASENAME}_step_{step}.pt"
                torch.save(
                    {
                        "step": step,
                        "model_state": model.state_dict(),
                        "second_head_state": second_head.state_dict(),
                        "optimizer_state": optimizer.state_dict(),
                        "scheduler_state": scheduler.state_dict(),
                        "config": {
                            "BLOCK_SIZE": BLOCK_SIZE,
                            "BATCH_SIZE": BATCH_SIZE,
                            "LR_TRUNK": LR_TRUNK,
                            "LR_HEADS": LR_HEADS,
                            "LAMBDA_T2": LAMBDA_T2,
                            "TRAINING_STEPS_STAGE2": TRAINING_STEPS_STAGE2,
                        },
                    },
                    ckpt_path,
                )
                print(f"\n[Checkpoint] Saved to {ckpt_path}\n")

progress_bar.close()

# Final checkpoint after training completes
final_ckpt_path = f"{CKPT_DIR}/{CKPT_BASENAME}_final.pt"
torch.save(
    {
        "step": step,
        "model_state": model.state_dict(),
        "second_head_state": second_head.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
    },
    final_ckpt_path,
)
print(f"Stage 2 joint training finished. Final checkpoint saved to {final_ckpt_path}")
```

## From-Scratch Checkpoint Restore and Model Setup

This cell is designed to be **self-contained** so that we can start from a
fresh Colab runtime and fully restore our dual-head GPT-2 setup from a
saved checkpoint.

What it does:

1. Installs and imports required libraries.
2. Sets up the device, dtype, and core hyperparameters (model name, block size, batch size).
3. Mounts Google Drive and points to a chosen checkpoint file.
4. Loads:
   - The GPT-2 model and tokenizer.
   - The `second_head` multi-token (t+2) head with the correct shape.
5. Builds parameter groups for:
   - The GPT-2 transformer trunk.
   - The original `lm_head` (Head₁, t+1).
   - The `second_head` (Head₂, t+2).
6. Loads the checkpoint’s weights into the model and second head.
7. Reports at which training step the checkpoint was saved.

After running this cell, you will have:
- `model`, `second_head`, `tokenizer`, `DEVICE`, `BLOCK_SIZE`, `BATCH_SIZE`
- `trunk_params`, `head1_params`, `head2_params`
all ready for **further training** or **evaluation**, without rerunning any training code.


```python
# ============================================================
# From-scratch restore of dual-head GPT-2 from checkpoint
# ============================================================
# This cell:
#   - Installs deps (if needed)
#   - Sets device, dtype, and core configs
#   - Mounts Google Drive
#   - Loads GPT-2 + tokenizer
#   - Reconstructs second_head
#   - Builds parameter groups (trunk, head₁, head₂)
#   - Loads weights from a saved checkpoint
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import os
import torch
import torch.nn as nn
from transformers import AutoTokenizer, GPT2LMHeadModel

# -------------------------------
# 1. Device, dtype, base configs
# -------------------------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16  # matches training setup

MODEL_NAME = "gpt2"
BLOCK_SIZE = 512       # must match training
BATCH_SIZE = 32        # can be changed for eval/training as needed

# -------------------------------
# 2. Mount Google Drive & set ckpt
# -------------------------------
from google.colab import drive
drive.mount("/content/drive")

# Directory and checkpoint path (adjust if needed)
CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
CKPT_PATH = os.path.join(CKPT_DIR, "stage2_gpt2_dual_head_final.pt")

print(f"Restoring from checkpoint: {CKPT_PATH}")

# -------------------------------
# 3. Load tokenizer & GPT-2 model
# -------------------------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
# Avoid spurious warnings about long sequences
tokenizer.model_max_length = BLOCK_SIZE + 2

model = GPT2LMHeadModel.from_pretrained(MODEL_NAME)
model.to(DEVICE)
model.to(dtype)

hidden_size = model.lm_head.in_features
vocab_size = model.config.vocab_size

# -------------------------------
# 4. Recreate second_head module
# -------------------------------
# Linear head mapping final hidden states h_t to logits for x_{t+2}.
second_head = nn.Linear(hidden_size, vocab_size, device=DEVICE, dtype=dtype)

# -------------------------------
# 5. Build parameter groups
# -------------------------------
# We plan to train:
#   - model.transformer (trunk)
#   - model.lm_head     (Head₁: t+1)
#   - second_head       (Head₂: t+2)
#
# GPT-2 ties embeddings and lm_head weights internally, so we must
# ensure each Parameter object appears in exactly one param group.

for param in model.transformer.parameters():
    param.requires_grad = True

for param in model.lm_head.parameters():
    param.requires_grad = True

for param in second_head.parameters():
    param.requires_grad = True

trunk_params = []
head1_params = []
head2_params = list(second_head.parameters())

seen = set()

for name, param in model.named_parameters():
    if not param.requires_grad:
        continue
    if id(param) in seen:
        continue
    if name.startswith("transformer."):
        trunk_params.append(param)
        seen.add(id(param))
    elif name.startswith("lm_head."):
        head1_params.append(param)
        seen.add(id(param))
    # second_head params are separate and already in head2_params

# -------------------------------
# 6. Load checkpoint weights
# -------------------------------
ckpt = torch.load(CKPT_PATH, map_location=DEVICE)

model.load_state_dict(ckpt["model_state"])
second_head.load_state_dict(ckpt["second_head_state"])

start_step = ckpt.get("step", 0)
print(f"Checkpoint loaded successfully. Original training step: {start_step}")
print(f"Model, second_head, and parameter groups are ready for training/evaluation.")
```

## Extra Stage 2 Training with Checkpoints

This cell continues joint fine-tuning of the dual-head GPT-2 model for an
additional number of steps and periodically saves checkpoints to Google Drive.

It assumes the restore cell has already been run and provides:

- `model`, `second_head`, `tokenizer`, `DEVICE`, `BLOCK_SIZE`, `BATCH_SIZE`
- Parameter groups: `trunk_params`, `head1_params`, `head2_params`

What it does:

1. Creates an optimizer and cosine LR scheduler for this extra phase.
2. Streams FineWeb (`sample-10BT`) and builds a multi-token dataset with:
   - `input_ids`  (context)
   - `labels_t1`  (t+1)
   - `labels_t2`  (t+2)
3. Trains for `TRAINING_STEPS_EXTRA` steps with:
   - Loss = `L_t1 + λ * L_t2`
   - Gradient clipping
4. Logs losses/accuracies for both heads.
5. Saves:
   - Periodic checkpoints every `CHECKPOINT_EVERY` steps.
   - A final checkpoint at the end of training.

```python
# ============================================================
# Extra Stage 2 training: joint t+1/t+2 fine-tuning + checkpoints
# ============================================================
# Assumes:
#   - The restore cell has been run.
#   - We have: model, second_head, tokenizer, DEVICE, BLOCK_SIZE,
#     BATCH_SIZE, trunk_params, head1_params, head2_params.
#   - Google Drive is mounted at /content/drive.
# ============================================================

import os
import torch
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset
from transformers import get_cosine_schedule_with_warmup
from tqdm.auto import tqdm

# -------------------------------
# 1. Extra-training config
# -------------------------------
LR_TRUNK = 1e-5          # small LR for the trunk
LR_HEADS = 3e-4          # larger LR for heads
LAMBDA_T2 = 0.5          # weight on t+2 loss
TRAINING_STEPS_EXTRA = 5_000
LOG_EVERY = 100

# Checkpoint config
CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
os.makedirs(CKPT_DIR, exist_ok=True)
CKPT_BASENAME = "stage2_gpt2_dual_head_extra"
CHECKPOINT_EVERY = 2_000  # save every N steps

# -------------------------------
# 2. Optimizer & scheduler
# -------------------------------
optimizer = torch.optim.AdamW(
    [
        {"params": trunk_params, "lr": LR_TRUNK},
        {"params": head1_params, "lr": LR_HEADS},
        {"params": head2_params, "lr": LR_HEADS},
    ],
    weight_decay=0.01,
)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=500,
    num_training_steps=TRAINING_STEPS_EXTRA,
)

# -------------------------------
# 3. FineWeb dataset (multi-token)
# -------------------------------
print("Loading FineWeb (sample-10BT) in streaming mode for extra Stage 2 training...")
hf_stream_extra = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

class FineWebMultiTokenDataset(IterableDataset):
    """
    Iterable dataset yielding:
      - input_ids  : tokens at positions t .. t+T-1
      - labels_t1  : tokens at positions t+1 .. t+T   (t+1 head)
      - labels_t2  : tokens at positions t+2 .. t+T+1 (t+2 head)
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        buffer = []
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if not ids:
                continue

            buffer.extend(ids)

            while len(buffer) >= self.block_size + 2:
                chunk = buffer[: self.block_size + 2]
                buffer = buffer[self.block_size :]

                input_ids = chunk[: self.block_size]
                labels_t1 = chunk[1 : self.block_size + 1]
                labels_t2 = chunk[2 : self.block_size + 2]

                yield {
                    "input_ids": torch.tensor(input_ids, dtype=torch.long),
                    "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                    "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
                }

train_iterable_extra = FineWebMultiTokenDataset(hf_stream_extra, tokenizer, BLOCK_SIZE)

train_loader_extra = DataLoader(
    train_iterable_extra,
    batch_size=BATCH_SIZE,
    num_workers=4,
    pin_memory=True,
)

# -------------------------------
# 4. Extra training loop
# -------------------------------

torch.cuda.empty_cache()

extra_step = 0
running_loss_t1 = 0.0
running_loss_t2 = 0.0
running_correct_t1 = 0
running_correct_t2 = 0
running_tokens = 0

progress_bar = tqdm(total=TRAINING_STEPS_EXTRA, desc="Extra Stage 2 training", unit="step")

model.train()
second_head.train()

while extra_step < TRAINING_STEPS_EXTRA:
    for batch in train_loader_extra:
        extra_step += 1
        if extra_step > TRAINING_STEPS_EXTRA:
            break

        input_ids = batch["input_ids"].to(DEVICE)
        labels_t1 = batch["labels_t1"].to(DEVICE)
        labels_t2 = batch["labels_t2"].to(DEVICE)

        outputs = model.transformer(input_ids)
        hidden_states = outputs.last_hidden_state  # [B, T, H]

        logits_t1 = model.lm_head(hidden_states)   # [B, T, V]
        logits_t2 = second_head(hidden_states)     # [B, T, V]

        B, T, V = logits_t1.shape

        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )

        loss = loss_t1 + LAMBDA_T2 * loss_t2

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(trunk_params + head1_params + head2_params, max_norm=1.0)
        optimizer.step()
        scheduler.step()

        with torch.no_grad():
            preds_t1 = logits_t1.argmax(dim=-1)
            preds_t2 = logits_t2.argmax(dim=-1)
            correct_t1 = (preds_t1 == labels_t1).sum().item()
            correct_t2 = (preds_t2 == labels_t2).sum().item()
            total_tokens = labels_t1.numel()

        running_loss_t1 += loss_t1.item()
        running_loss_t2 += loss_t2.item()
        running_correct_t1 += correct_t1
        running_correct_t2 += correct_t2
        running_tokens += total_tokens

        if extra_step % LOG_EVERY == 0:
            avg_loss_t1 = running_loss_t1 / LOG_EVERY
            avg_loss_t2 = running_loss_t2 / LOG_EVERY
            avg_acc_t1 = running_correct_t1 / running_tokens
            avg_acc_t2 = running_correct_t2 / running_tokens

            lr = scheduler.get_last_lr()[0]
            progress_bar.set_postfix({
                "L_t1": f"{avg_loss_t1:.3f}",
                "L_t2": f"{avg_loss_t2:.3f}",
                "acc_t1": f"{avg_acc_t1:.3f}",
                "acc_t2": f"{avg_acc_t2:.3f}",
                "lr": f"{lr:.2e}",
            })
            progress_bar.update(LOG_EVERY)

            running_loss_t1 = 0.0
            running_loss_t2 = 0.0
            running_correct_t1 = 0
            running_correct_t2 = 0
            running_tokens = 0

            # ---- Periodic checkpointing for extra phase ----
            if extra_step % CHECKPOINT_EVERY == 0:
                ckpt_path = os.path.join(
                    CKPT_DIR,
                    f"{CKPT_BASENAME}_step_{extra_step}.pt"
                )
                torch.save(
                    {
                        "step": extra_step,
                        "model_state": model.state_dict(),
                        "second_head_state": second_head.state_dict(),
                        "optimizer_state": optimizer.state_dict(),
                        "scheduler_state": scheduler.state_dict(),
                        "config": {
                            "BLOCK_SIZE": BLOCK_SIZE,
                            "BATCH_SIZE": BATCH_SIZE,
                            "LR_TRUNK": LR_TRUNK,
                            "LR_HEADS": LR_HEADS,
                            "LAMBDA_T2": LAMBDA_T2,
                            "TRAINING_STEPS_EXTRA": TRAINING_STEPS_EXTRA,
                        },
                    },
                    ckpt_path,
                )
                print(f"\n[Checkpoint] Saved to {ckpt_path}\n")

progress_bar.close()

# Final checkpoint for this extra phase
final_ckpt_path = os.path.join(CKPT_DIR, f"{CKPT_BASENAME}_final.pt")
torch.save(
    {
        "step": extra_step,
        "model_state": model.state_dict(),
        "second_head_state": second_head.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
    },
    final_ckpt_path,
)
print(f"Extra Stage 2 training finished. Final checkpoint saved to {final_ckpt_path}")
```

## Evaluating the Dual-Head Model (t+1 and t+2)

With the model restored (and optionally further trained), we now want to
evaluate how well each head is performing.

This evaluation cell:

1. Runs a short **quantitative evaluation** on a streamed portion of
   FineWeb `sample-10BT`, computing:
   - t+1 loss and accuracy for the original GPT-2 head (Head₁)
   - t+2 loss and accuracy for the multi-token head (Head₂)
2. Prints a few **qualitative examples**:
   - Shows a context snippet from FineWeb
   - Shows the *true* next token (t+1) and *true* t+2 token
   - Shows the top-5 predictions from:
     - Head₁ for t+1
     - Head₂ for t+2

This gives both numeric and human-intuitive views of how well the model
“thinks ahead.”


```python
# ============================================================
# Evaluation: dual-head GPT-2 (t+1 and t+2)
# ============================================================
# Assumes:
#   - Restore cell has been run.
#   - We have: model, second_head, tokenizer, DEVICE, BLOCK_SIZE.
# ============================================================

import torch
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset

# -------------------------------
# 1. Evaluation config
# -------------------------------
EVAL_STEPS = 200       # number of batches for quantitative stats
EVAL_BATCH_SIZE = 16
NUM_EXAMPLES = 3       # number of qualitative examples to print
TOP_K = 5              # top-k predictions to display

# -------------------------------
# 2. Evaluation dataset (multi-token)
# -------------------------------

print("Loading FineWeb (sample-10BT) in streaming mode for evaluation...")
hf_stream_eval = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

class FineWebMultiTokenDataset(IterableDataset):
    """
    Same format as training:
      - input_ids  : positions t .. t+T-1
      - labels_t1  : positions t+1 .. t+T
      - labels_t2  : positions t+2 .. t+T+1
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        buffer = []
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if not ids:
                continue

            buffer.extend(ids)

            while len(buffer) >= self.block_size + 2:
                chunk = buffer[: self.block_size + 2]
                buffer = buffer[self.block_size :]

                input_ids = chunk[: self.block_size]
                labels_t1 = chunk[1 : self.block_size + 1]
                labels_t2 = chunk[2 : self.block_size + 2]

                yield {
                    "input_ids": torch.tensor(input_ids, dtype=torch.long),
                    "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                    "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
                }

eval_iterable = FineWebMultiTokenDataset(hf_stream_eval, tokenizer, BLOCK_SIZE)

eval_loader = DataLoader(
    eval_iterable,
    batch_size=EVAL_BATCH_SIZE,
    num_workers=2,
    pin_memory=True,
)

# -------------------------------
# 3. Quantitative evaluation
# -------------------------------

model.eval()
second_head.eval()

eval_loss_t1 = 0.0
eval_loss_t2 = 0.0
eval_correct_t1 = 0
eval_correct_t2 = 0
eval_tokens = 0

with torch.no_grad():
    for i, batch in enumerate(eval_loader):
        if i >= EVAL_STEPS:
            break

        input_ids = batch["input_ids"].to(DEVICE)
        labels_t1 = batch["labels_t1"].to(DEVICE)
        labels_t2 = batch["labels_t2"].to(DEVICE)

        outputs = model.transformer(input_ids)
        hidden_states = outputs.last_hidden_state  # [B, T, H]

        logits_t1 = model.lm_head(hidden_states)
        logits_t2 = second_head(hidden_states)

        B, T, V = logits_t1.shape

        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )

        preds_t1 = logits_t1.argmax(dim=-1)
        preds_t2 = logits_t2.argmax(dim=-1)

        correct_t1 = (preds_t1 == labels_t1).sum().item()
        correct_t2 = (preds_t2 == labels_t2).sum().item()
        total_tokens = labels_t1.numel()

        eval_loss_t1 += loss_t1.item()
        eval_loss_t2 += loss_t2.item()
        eval_correct_t1 += correct_t1
        eval_correct_t2 += correct_t2
        eval_tokens += total_tokens

avg_loss_t1 = eval_loss_t1 / max(EVAL_STEPS, 1)
avg_loss_t2 = eval_loss_t2 / max(EVAL_STEPS, 1)
avg_acc_t1 = eval_correct_t1 / max(eval_tokens, 1)
avg_acc_t2 = eval_correct_t2 / max(eval_tokens, 1)

print("=== Quantitative Evaluation (approximate) ===")
print(f"t+1 head  - loss: {avg_loss_t1:.3f} | acc: {avg_acc_t1:.3f}")
print(f"t+2 head  - loss: {avg_loss_t2:.3f} | acc: {avg_acc_t2:.3f}")
print()

# -------------------------------
# 4. Qualitative examples
# -------------------------------

print(f"=== Qualitative Examples (N={NUM_EXAMPLES}) ===")

hf_stream_examples = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

examples_printed = 0

with torch.no_grad():
    for example in hf_stream_examples:
        if examples_printed >= NUM_EXAMPLES:
            break

        text = example.get("text", "")
        if not text:
            continue

        token_ids = tokenizer.encode(text, add_special_tokens=False)
        if len(token_ids) < 4:
            continue

        context_ids = token_ids[: BLOCK_SIZE]
        context_len = len(context_ids)

        # ensure we have ground-truth t+1 and t+2 beyond context
        if context_len + 2 >= len(token_ids):
            continue

        input_ids = torch.tensor(context_ids, dtype=torch.long, device=DEVICE).unsqueeze(0)  # [1, T]

        outputs = model.transformer(input_ids)
        hidden_states = outputs.last_hidden_state  # [1, T, H]
        last_hidden = hidden_states[:, -1:, :]      # [1, 1, H]

        logits_next = model.lm_head(last_hidden)    # [1, 1, V]  t+1
        logits_future = second_head(last_hidden)    # [1, 1, V]  t+2

        probs_next = torch.softmax(logits_next[0, 0], dim=-1)
        probs_future = torch.softmax(logits_future[0, 0], dim=-1)

        topk_next = torch.topk(probs_next, k=TOP_K)
        topk_future = torch.topk(probs_future, k=TOP_K)

        true_t1_id = token_ids[context_len]
        true_t2_id = token_ids[context_len + 1]
        true_t1 = tokenizer.decode([true_t1_id])
        true_t2 = tokenizer.decode([true_t2_id])

        context_text = tokenizer.decode(context_ids[-80:])  # show last ~80 tokens

        print(f"\n--- Example {examples_printed + 1} ---")
        print("Context:")
        print(context_text)
        print()
        print(f"True next token (t+1):   {repr(true_t1)}")
        print(f"True token at t+2:       {repr(true_t2)}")
        print()

        print("Head₁ (t+1) top predictions:")
        for prob, idx in zip(topk_next.values, topk_next.indices):
            tok_str = tokenizer.decode([idx.item()])
            print(f"  {prob.item():.4f}  ->  {repr(tok_str)}")

        print("\nHead₂ (t+2) top predictions:")
        for prob, idx in zip(topk_future.values, topk_future.indices):
            tok_str = tokenizer.decode([idx.item()])
            print(f"  {prob.item():.4f}  ->  {repr(tok_str)}")

        examples_printed += 1

print("\nEvaluation and qualitative examples complete.")
```

## Meta-style Dual-Head GPT-2: Shared Trunk + Transformer Heads (t+1, t+2)

Architecture:

- **Trunk**: Pretrained GPT-2 (`gpt2`), fully unfrozen.
- **Heads**:
  - Head₁ (t+1): 2 GPT-2 transformer blocks on top of the trunk’s hidden states, then the **shared** unembedding matrix.
  - Head₂ (t+2): another independent stack of 2 GPT-2 transformer blocks, also using the **same** unembedding matrix.
- **Unembedding**: We reuse GPT-2’s `lm_head` (tied to `wte`) as a shared vocab projection for both heads.

Training objective:

- We stream the FineWeb `sample-10BT` subset and create sequences of length `BLOCK_SIZE`.
- For each example:
  - `input_ids`: tokens at positions `t .. t+T-1`
  - `labels_t1`: tokens at positions `t+1 .. t+T`   (for Head₁: t+1)
  - `labels_t2`: tokens at positions `t+2 .. t+T+1` (for Head₂: t+2)
- Loss:
  \[
  L = L_{t+1} + \lambda_{t+2} L_{t+2}
  \]
  where both losses are standard cross-entropy over the vocab.
- We **fully unfreeze** the GPT-2 trunk so both heads’ gradients backpropagate into it, encouraging it to learn representations that support both t+1 and t+2 prediction (Meta-style behavior).

We also:

- Track **top-1 accuracy** for each head.
- Save periodic and final checkpoints to Google Drive so training can be resumed or analyzed later.

```python
# ============================================================
# Meta-style dual-head GPT-2:
#   - Pretrained GPT-2 trunk (fully unfrozen)
#   - Two transformer heads (2 GPT-2 blocks each) for t+1 and t+2
#   - Shared unembedding (GPT-2's lm_head)
#   - Joint CE loss on FineWeb (sample-10BT)
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import os
import math
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset
from tqdm.auto import tqdm

from transformers import (
    AutoTokenizer,
    GPT2LMHeadModel,
    get_cosine_schedule_with_warmup,
)
from transformers.models.gpt2.modeling_gpt2 import GPT2Block

# -------------------------------
# 1. Device, dtype, base config
# -------------------------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16  # bf16 is numerically safer than fp16 on Ampere

MODEL_NAME = "gpt2"
BLOCK_SIZE = 512
BATCH_SIZE = 24   # adjust based on VRAM; A100 should handle 24–32 here
LR_TRUNK = 2e-5   # learning rate for the GPT-2 trunk + lm_head
LR_HEADS = 2e-4   # higher LR for the new transformer heads
LAMBDA_T2 = 0.7   # weight on t+2 loss (slightly downweighted vs t+1)
TRAINING_STEPS = 10_000
WARMUP_STEPS = 1_000
LOG_EVERY = 100

# -------------------------------
# 2. Mount Drive for checkpoints
# -------------------------------
from google.colab import drive
drive.mount("/content/drive")

CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
os.makedirs(CKPT_DIR, exist_ok=True)
CKPT_BASENAME = "meta_style_gpt2_dual_head"


# -------------------------------
# 3. Load tokenizer & GPT-2 base
# -------------------------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
# Avoid annoying long-sequence warnings (we always slice to BLOCK_SIZE anyway)
tokenizer.model_max_length = BLOCK_SIZE + 2

base_model = GPT2LMHeadModel.from_pretrained(MODEL_NAME)
base_model.to(DEVICE)
base_model.to(dtype)

config = base_model.config
hidden_size = config.hidden_size
vocab_size = config.vocab_size

# -------------------------------
# 4. Meta-style transformer heads
# -------------------------------
class TransformerHead(nn.Module):
    """
    A small stack of GPT-2 blocks that takes trunk hidden states
    and refines them for a particular prediction head (t+1, t+2, etc.).
    """

    def __init__(self, config, num_layers=2):
        super().__init__()
        self.layers = nn.ModuleList([GPT2Block(config) for _ in range(num_layers)])

    def forward(self, hidden_states):
        # hidden_states: [B, T, H]
        for block in self.layers:
            # GPT2Block returns (hidden_states, present, ...)
            hidden_states = block(hidden_states)[0]
        return hidden_states


class MetaStyleDualHeadGPT2(nn.Module):
    """
    Wraps a GPT-2 trunk with two transformer heads for t+1 and t+2,
    sharing a single unembedding (lm_head).
    """

    def __init__(self, base_model, num_head_layers=2):
        super().__init__()
        # Shared trunk (pretrained GPT-2)
        self.transformer = base_model.transformer
        # Shared unembedding (tied with embeddings internally)
        self.lm_head = base_model.lm_head
        self.config = base_model.config

        # Two independent transformer heads
        self.head_t1 = TransformerHead(self.config, num_layers=num_head_layers)
        self.head_t2 = TransformerHead(self.config, num_layers=num_head_layers)

    def forward(self, input_ids):
        """
        input_ids: [B, T]
        Returns:
          logits_t1: [B, T, V]  (for t+1 labels)
          logits_t2: [B, T, V]  (for t+2 labels)
        """
        outputs = self.transformer(input_ids)
        hidden_states = outputs.last_hidden_state  # [B, T, H]

        # Feed trunk hidden states through each head
        h1 = self.head_t1(hidden_states)          # [B, T, H]
        h2 = self.head_t2(hidden_states)          # [B, T, H]

        # Shared unembedding for both heads
        logits_t1 = self.lm_head(h1)              # [B, T, V]
        logits_t2 = self.lm_head(h2)              # [B, T, V]

        return logits_t1, logits_t2


model = MetaStyleDualHeadGPT2(base_model, num_head_layers=2).to(DEVICE)
model.to(dtype)

# Unfreeze everything: trunk + lm_head + both heads
for param in model.parameters():
    param.requires_grad = True

# -------------------------------
# 5. Build parameter groups
# -------------------------------
trunk_params = []
head_params = []
lm_params = []
seen = set()

for name, param in model.named_parameters():
    if not param.requires_grad:
        continue
    if id(param) in seen:
        continue

    if name.startswith("transformer."):
        trunk_params.append(param)
        seen.add(id(param))
    elif name.startswith("head_t1.") or name.startswith("head_t2."):
        head_params.append(param)
        seen.add(id(param))
    elif name.startswith("lm_head."):
        lm_params.append(param)
        seen.add(id(param))
    else:
        # Any remaining parameters (e.g., embeddings not under transformer.*)
        trunk_params.append(param)
        seen.add(id(param))

optimizer = torch.optim.AdamW(
    [
        {"params": trunk_params + lm_params, "lr": LR_TRUNK},
        {"params": head_params, "lr": LR_HEADS},
    ],
    weight_decay=0.01,
)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=WARMUP_STEPS,
    num_training_steps=TRAINING_STEPS,
)

# -------------------------------
# 6. FineWeb multi-token dataset
# -------------------------------
print("Loading FineWeb (sample-10BT) in streaming mode...")
hf_stream = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

class FineWebMultiTokenDataset(IterableDataset):
    """
    Iterable dataset yielding:
      - input_ids  : tokens at positions t .. t+T-1
      - labels_t1  : tokens at positions t+1 .. t+T
      - labels_t2  : tokens at positions t+2 .. t+T+1
    using a rolling buffer over the FineWeb token stream.
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        buffer = []
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if not ids:
                continue

            buffer.extend(ids)

            # We need BLOCK_SIZE + 2 tokens to define input + t+1 + t+2 labels
            while len(buffer) >= self.block_size + 2:
                # chunk length = BLOCK_SIZE + 2
                chunk = buffer[: self.block_size + 2]
                buffer = buffer[self.block_size :]

                input_ids = chunk[: self.block_size]            # [T]
                labels_t1 = chunk[1 : self.block_size + 1]      # [T]
                labels_t2 = chunk[2 : self.block_size + 2]      # [T]

                yield {
                    "input_ids": torch.tensor(input_ids, dtype=torch.long),
                    "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                    "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
                }

train_iterable = FineWebMultiTokenDataset(hf_stream, tokenizer, BLOCK_SIZE)

train_loader = DataLoader(
    train_iterable,
    batch_size=BATCH_SIZE,
    num_workers=4,
    pin_memory=True,
)

# -------------------------------
# 7. Training loop (joint t+1/t+2)
# -------------------------------
torch.cuda.empty_cache()

step = 0
running_loss_t1 = 0.0
running_loss_t2 = 0.0
running_correct_t1 = 0
running_correct_t2 = 0
running_tokens = 0

CHECKPOINT_EVERY = 2_000

progress_bar = tqdm(total=TRAINING_STEPS, desc="Meta-style joint training", unit="step")

model.train()

while step < TRAINING_STEPS:
    for batch in train_loader:
        step += 1
        if step > TRAINING_STEPS:
            break

        input_ids = batch["input_ids"].to(DEVICE)   # [B, T]
        labels_t1 = batch["labels_t1"].to(DEVICE)   # [B, T]
        labels_t2 = batch["labels_t2"].to(DEVICE)   # [B, T]

        logits_t1, logits_t2 = model(input_ids)     # each [B, T, V]
        B, T, V = logits_t1.shape

        # Cross-entropy for both heads
        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )

        loss = loss_t1 + LAMBDA_T2 * loss_t2

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        # Gradient clipping for stability
        torch.nn.utils.clip_grad_norm_(trunk_params + lm_params + head_params, max_norm=1.0)
        optimizer.step()
        scheduler.step()

        # Compute top-1 accuracies for monitoring
        with torch.no_grad():
            preds_t1 = logits_t1.argmax(dim=-1)  # [B, T]
            preds_t2 = logits_t2.argmax(dim=-1)  # [B, T]
            correct_t1 = (preds_t1 == labels_t1).sum().item()
            correct_t2 = (preds_t2 == labels_t2).sum().item()
            total_tokens = labels_t1.numel()

        running_loss_t1 += loss_t1.item()
        running_loss_t2 += loss_t2.item()
        running_correct_t1 += correct_t1
        running_correct_t2 += correct_t2
        running_tokens += total_tokens

        # Logging + checkpointing
        if step % LOG_EVERY == 0:
            avg_loss_t1 = running_loss_t1 / LOG_EVERY
            avg_loss_t2 = running_loss_t2 / LOG_EVERY
            avg_acc_t1 = running_correct_t1 / running_tokens
            avg_acc_t2 = running_correct_t2 / running_tokens
            lr = scheduler.get_last_lr()[0]

            progress_bar.set_postfix({
                "L_t1": f"{avg_loss_t1:.3f}",
                "L_t2": f"{avg_loss_t2:.3f}",
                "acc_t1": f"{avg_acc_t1:.3f}",
                "acc_t2": f"{avg_acc_t2:.3f}",
                "lr": f"{lr:.2e}",
            })
            progress_bar.update(LOG_EVERY)

            running_loss_t1 = 0.0
            running_loss_t2 = 0.0
            running_correct_t1 = 0
            running_correct_t2 = 0
            running_tokens = 0

            # Periodic checkpoint
            if step % CHECKPOINT_EVERY == 0:
                ckpt_path = os.path.join(
                    CKPT_DIR,
                    f"{CKPT_BASENAME}_step_{step}.pt"
                )
                torch.save(
                    {
                        "step": step,
                        "model_state": model.state_dict(),
                        "optimizer_state": optimizer.state_dict(),
                        "scheduler_state": scheduler.state_dict(),
                        "config": {
                            "BLOCK_SIZE": BLOCK_SIZE,
                            "BATCH_SIZE": BATCH_SIZE,
                            "LR_TRUNK": LR_TRUNK,
                            "LR_HEADS": LR_HEADS,
                            "LAMBDA_T2": LAMBDA_T2,
                            "TRAINING_STEPS": TRAINING_STEPS,
                        },
                    },
                    ckpt_path,
                )
                print(f"\n[Checkpoint] Saved to {ckpt_path}\n")

progress_bar.close()

# Final checkpoint
final_ckpt_path = os.path.join(CKPT_DIR, f"{CKPT_BASENAME}_final.pt")
torch.save(
    {
        "step": step,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
    },
    final_ckpt_path,
)
print(f"Meta-style joint training finished. Final checkpoint saved to {final_ckpt_path}")
```

## Extra Meta-Style Training Phase (Resume from Checkpoint)

This cell lets us **continue training** the Meta-style dual-head GPT-2 model
from a previously saved checkpoint.

What it does:

1. Mounts Google Drive and points to an existing checkpoint
   (e.g., `meta_style_gpt2_dual_head_final.pt` or any `_step_*.pt`).
2. Rebuilds the Meta-style architecture:
   - GPT-2 trunk (pretrained)
   - Two transformer heads (2 GPT-2 blocks each) for t+1 and t+2
   - Shared unembedding (`lm_head`)
3. Loads the checkpoint's `model_state` into this architecture.
4. Sets up a **new optimizer and scheduler** for an extra training phase:
   - Trunk + lm_head at a smaller LR
   - Heads at a higher LR
5. Streams FineWeb (`sample-10BT`) again and continues training for
   `TRAINING_STEPS_EXTRA` steps with the same loss:
   \[
     L = L_{t+1} + \lambda_{t+2} L_{t+2}
   \]
6. Logs t+1/t+2 losses and accuracies, and saves periodic + final checkpoints
   for this extra phase.

This does **not** require the original optimizer/scheduler state; it treats the
extra training as a new fine-tuning phase starting from the chosen checkpoint.


```python
# ============================================================
# Extra Meta-style training phase:
#   - Restore model from checkpoint
#   - Continue joint t+1/t+2 training for TRAINING_STEPS_EXTRA
#   - Use doc-local random windows over FineWeb with shuffling
#   - Seed RNGs based on checkpoint step so each phase sees new data
#   - Save new checkpoints
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import os
import random
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset
from tqdm.auto import tqdm

from transformers import (
    AutoTokenizer,
    GPT2LMHeadModel,
    get_cosine_schedule_with_warmup,
)
from transformers.models.gpt2.modeling_gpt2 import GPT2Block

# -------------------------------
# 1. Device, dtype, base config
# -------------------------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16

MODEL_NAME = "gpt2"
BLOCK_SIZE = 512
BATCH_SIZE = 24

LR_TRUNK = 1e-5     # slightly more conservative LR for extra phase
LR_HEADS = 1e-4     # heads can still move faster
LAMBDA_T2 = 1
TRAINING_STEPS_EXTRA = 5_000
WARMUP_STEPS_EXTRA = 500
LOG_EVERY = 100

# -------------------------------
# 2. Mount Drive & select ckpt
# -------------------------------
from google.colab import drive
drive.mount("/content/drive")

CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
os.makedirs(CKPT_DIR, exist_ok=True)

# Pick which checkpoint to resume from:
CKPT_NAME = "meta_style_gpt2_dual_head_final.pt"
CKPT_PATH = os.path.join(CKPT_DIR, CKPT_NAME)
print(f"Resuming from checkpoint: {CKPT_PATH}")

# -------------------------------
# 3. Define Meta-style classes
# -------------------------------
class TransformerHead(nn.Module):
    """
    A small stack of GPT-2 blocks that takes trunk hidden states
    and refines them for a particular prediction head (t+1, t+2, etc.).
    """

    def __init__(self, config, num_layers=2):
        super().__init__()
        self.layers = nn.ModuleList([GPT2Block(config) for _ in range(num_layers)])

    def forward(self, hidden_states):
        for block in self.layers:
            hidden_states = block(hidden_states)[0]
        return hidden_states


class MetaStyleDualHeadGPT2(nn.Module):
    """
    GPT-2 trunk + two transformer heads (t+1, t+2) sharing lm_head.
    """

    def __init__(self, base_model, num_head_layers=2):
        super().__init__()
        self.transformer = base_model.transformer
        self.lm_head = base_model.lm_head
        self.config = base_model.config

        self.head_t1 = TransformerHead(self.config, num_layers=num_head_layers)
        self.head_t2 = TransformerHead(self.config, num_layers=num_head_layers)

    def forward(self, input_ids):
        outputs = self.transformer(input_ids)
        hidden_states = outputs.last_hidden_state  # [B, T, H]

        h1 = self.head_t1(hidden_states)
        h2 = self.head_t2(hidden_states)

        logits_t1 = self.lm_head(h1)
        logits_t2 = self.lm_head(h2)

        return logits_t1, logits_t2

# -------------------------------
# 4. Rebuild model & load weights
# -------------------------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.model_max_length = BLOCK_SIZE + 2

base_model = GPT2LMHeadModel.from_pretrained(MODEL_NAME)
base_model.to(DEVICE)
base_model.to(dtype)

model = MetaStyleDualHeadGPT2(base_model, num_head_layers=2).to(DEVICE)
model.to(dtype)

# Load checkpoint weights
ckpt = torch.load(CKPT_PATH, map_location=DEVICE)
model.load_state_dict(ckpt["model_state"])
start_step = ckpt.get("step", 0)
print(f"Checkpoint loaded. Original step: {start_step}")

# -------------------------------
# 4a. Seed RNGs based on checkpoint step
#     This makes each extra-training phase see a new slice of data.
# -------------------------------
BASE_SEED = 1234
run_seed = BASE_SEED + start_step  # changes as training progresses

random.seed(run_seed)
torch.manual_seed(run_seed)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(run_seed)

print(f"Using run_seed={run_seed} for this extra training phase.")

# Unfreeze everything
for param in model.parameters():
    param.requires_grad = True

# -------------------------------
# 5. Build parameter groups
# -------------------------------
trunk_params = []
head_params = []
lm_params = []
seen = set()

for name, param in model.named_parameters():
    if not param.requires_grad:
        continue
    if id(param) in seen:
        continue

    if name.startswith("transformer."):
        trunk_params.append(param)
        seen.add(id(param))
    elif name.startswith("head_t1.") or name.startswith("head_t2."):
        head_params.append(param)
        seen.add(id(param))
    elif name.startswith("lm_head."):
        lm_params.append(param)
        seen.add(id(param))
    else:
        # Fallback (shouldn't really happen, but we keep it for safety)
        trunk_params.append(param)
        seen.add(id(param))

optimizer = torch.optim.AdamW(
    [
        {"params": trunk_params + lm_params, "lr": LR_TRUNK},
        {"params": head_params, "lr": LR_HEADS},
    ],
    weight_decay=0.01,
)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=WARMUP_STEPS_EXTRA,
    num_training_steps=TRAINING_STEPS_EXTRA,
)

# -------------------------------
# 6. FineWeb multi-token dataset
#     Now uses:
#       - streaming + shuffle (doc-level) with seed=run_seed
#       - doc-local random windows of size BLOCK_SIZE+2
# -------------------------------
print("Loading FineWeb (sample-10BT) in streaming mode for extra phase...")

hf_stream_extra = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
).shuffle(buffer_size=10_000, seed=run_seed)  # approximate doc-level shuffle


class FineWebMultiTokenDataset(IterableDataset):
    """
    Iterable dataset that:
      - Reads documents from a shuffled streaming HF dataset.
      - Tokenizes each document with GPT-2 tokenizer.
      - For each document, picks ONE random window of length BLOCK_SIZE+2:
            chunk = ids[start : start + BLOCK_SIZE + 2]
        where start is uniform over valid offsets.
      - Produces:
            input_ids : positions t .. t+BLOCK_SIZE-1
            labels_t1 : positions t+1 .. t+BLOCK_SIZE
            labels_t2 : positions t+2 .. t+BLOCK_SIZE+1

    This avoids the global rolling buffer and makes each training phase
    see different slices of the corpus thanks to run_seed.
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        super().__init__()
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if len(ids) < self.block_size + 2:
                # Skip very short docs
                continue

            # Choose a random starting offset inside this document
            max_start = len(ids) - (self.block_size + 2)
            start = random.randint(0, max_start)
            chunk = ids[start : start + self.block_size + 2]

            input_ids = chunk[: self.block_size]
            labels_t1 = chunk[1 : self.block_size + 1]
            labels_t2 = chunk[2 : self.block_size + 2]

            yield {
                "input_ids": torch.tensor(input_ids, dtype=torch.long),
                "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
            }

train_iterable_extra = FineWebMultiTokenDataset(hf_stream_extra, tokenizer, BLOCK_SIZE)

train_loader_extra = DataLoader(
    train_iterable_extra,
    batch_size=BATCH_SIZE,
    num_workers=4,
    pin_memory=True,
)

# -------------------------------
# 7. Extra training loop
# -------------------------------
torch.cuda.empty_cache()

extra_step = 0
running_loss_t1 = 0.0
running_loss_t2 = 0.0
running_correct_t1 = 0
running_correct_t2 = 0
running_tokens = 0

CHECKPOINT_EVERY = 2_000
CKPT_BASENAME_EXTRA = "meta_style_gpt2_dual_head"

progress_bar = tqdm(total=TRAINING_STEPS_EXTRA, desc="Extra Meta-style training", unit="step")

model.train()

while extra_step < TRAINING_STEPS_EXTRA:
    for batch in train_loader_extra:
        extra_step += 1
        if extra_step > TRAINING_STEPS_EXTRA:
            break

        input_ids = batch["input_ids"].to(DEVICE)
        labels_t1 = batch["labels_t1"].to(DEVICE)
        labels_t2 = batch["labels_t2"].to(DEVICE)

        logits_t1, logits_t2 = model(input_ids)
        B, T, V = logits_t1.shape

        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )

        loss = loss_t1 + LAMBDA_T2 * loss_t2

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(trunk_params + lm_params + head_params, max_norm=1.0)
        optimizer.step()
        scheduler.step()

        with torch.no_grad():
            preds_t1 = logits_t1.argmax(dim=-1)
            preds_t2 = logits_t2.argmax(dim=-1)
            correct_t1 = (preds_t1 == labels_t1).sum().item()
            correct_t2 = (preds_t2 == labels_t2).sum().item()
            total_tokens = labels_t1.numel()

        running_loss_t1 += loss_t1.item()
        running_loss_t2 += loss_t2.item()
        running_correct_t1 += correct_t1
        running_correct_t2 += correct_t2
        running_tokens += total_tokens

        if extra_step % LOG_EVERY == 0:
            avg_loss_t1 = running_loss_t1 / LOG_EVERY
            avg_loss_t2 = running_loss_t2 / LOG_EVERY
            avg_acc_t1 = running_correct_t1 / running_tokens
            avg_acc_t2 = running_correct_t2 / running_tokens
            lr = scheduler.get_last_lr()[0]

            progress_bar.set_postfix({
                "L_t1": f"{avg_loss_t1:.3f}",
                "L_t2": f"{avg_loss_t2:.3f}",
                "acc_t1": f"{avg_acc_t1:.3f}",
                "acc_t2": f"{avg_acc_t2:.3f}",
                "lr": f"{lr:.2e}",
            })
            progress_bar.update(LOG_EVERY)

            running_loss_t1 = 0.0
            running_loss_t2 = 0.0
            running_correct_t1 = 0
            running_correct_t2 = 0
            running_tokens = 0

            if extra_step % CHECKPOINT_EVERY == 0:
                ckpt_path = os.path.join(
                    CKPT_DIR,
                    f"{CKPT_BASENAME_EXTRA}_step_{start_step + extra_step}.pt"
                )
                torch.save(
                    {
                        "step": start_step + extra_step,
                        "model_state": model.state_dict(),
                        "optimizer_state": optimizer.state_dict(),
                        "scheduler_state": scheduler.state_dict(),
                        "config": {
                            "BLOCK_SIZE": BLOCK_SIZE,
                            "BATCH_SIZE": BATCH_SIZE,
                            "LR_TRUNK": LR_TRUNK,
                            "LR_HEADS": LR_HEADS,
                            "LAMBDA_T2": LAMBDA_T2,
                            "TRAINING_STEPS_EXTRA": TRAINING_STEPS_EXTRA,
                        },
                    },
                    ckpt_path,
                )
                print(f"\n[Checkpoint] Saved to {ckpt_path}\n")

progress_bar.close()

final_ckpt_path = os.path.join(
    CKPT_DIR,
    f"{CKPT_BASENAME_EXTRA}_final.pt"
)
torch.save(
    {
        "step": start_step + extra_step,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
    },
    final_ckpt_path,
)
print(f"Extra Meta-style training finished. Final checkpoint saved to {final_ckpt_path}")
```

## Evaluation: Loss, Perplexity, Top-k, and Adaptive Distribution Inspection

This cell evaluates the Meta-style dual-head GPT-2 model on FineWeb
(`sample-10BT`) using a small held-out stream.

It reports:

1. **Quantitative metrics (approximate)** over `EVAL_STEPS` minibatches:
   - For both heads (t+1 and t+2):
     - Average cross-entropy loss
     - **Perplexity**: `exp(avg_loss)`
     - **Top-1 accuracy**
     - **Top-k accuracy** (e.g., k=5)

2. **Qualitative examples**:
   - We sample `N_EXAMPLES` sequences from FineWeb and for each:
     - Print a truncated context (up to `BLOCK_SIZE` tokens).
     - Show the true next token (t+1) and the true token at t+2.
     - Run both heads at the last valid position and compute:
       - The full probability distribution over the vocab.
       - The **entropy** of the distribution at that position.
       - A **dynamic top-K** list:
         - If entropy is low (sharp distribution) → `K = 3`.
         - If entropy is moderate → `K = 5`.
         - If entropy is high (flat / confused) → `K = 10`.
     - For each head we print the adaptive top-K predictions (token + probability).

This lets us check:

- Whether t+2 predictions are sensible even when not exact.
- How “confident” or “confused” each head is (via entropy/perplexity).
- How often the true token appears in the top-K set vs just top-1.


```python
# ============================================================
# Meta-style evaluation:
#   - Loss + perplexity for t+1 and t+2
#   - Top-1 and top-k accuracy
#   - Qualitative examples with adaptive top-K distributions
#     * Context now shows the last N tokens up to t
#     * Qualitative examples are randomly sampled from the stream
# ============================================================

import math
import random
import torch
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset

from tqdm.auto import tqdm

# Make sure we have model, tokenizer, DEVICE, BLOCK_SIZE defined.
model.eval()

# -------------------------------
# 1. Config for eval
# -------------------------------
EVAL_STEPS = 500        # number of minibatches for quantitative eval
EVAL_BATCH_SIZE = 16    # can be smaller than training batch
TOP_K_METRIC = 5        # k used for quantitative top-k accuracy
N_EXAMPLES = 3          # number of qualitative examples
CONTEXT_TOKENS = 80     # how many tokens BEFORE t we show in the context

# -------------------------------
# 2. FineWeb dataset for eval
# -------------------------------
print("Loading FineWeb (sample-10BT) in streaming mode for evaluation...")

hf_stream_eval = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

class FineWebMultiTokenDataset(IterableDataset):
    """
    Same as training dataset:
      - input_ids  : tokens at positions t .. t+T-1
      - labels_t1  : tokens at positions t+1 .. t+T
      - labels_t2  : tokens at positions t+2 .. t+T+1
    """

    def __init__(self, hf_iterable, tokenizer, block_size):
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size

    def __iter__(self):
        buffer = []
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if not ids:
                continue

            buffer.extend(ids)

            while len(buffer) >= self.block_size + 2:
                chunk = buffer[: self.block_size + 2]
                buffer = buffer[self.block_size :]

                input_ids = chunk[: self.block_size]
                labels_t1 = chunk[1 : self.block_size + 1]
                labels_t2 = chunk[2 : self.block_size + 2]

                yield {
                    "input_ids": torch.tensor(input_ids, dtype=torch.long),
                    "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                    "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
                }

eval_iterable = FineWebMultiTokenDataset(hf_stream_eval, tokenizer, BLOCK_SIZE)

eval_loader = DataLoader(
    eval_iterable,
    batch_size=EVAL_BATCH_SIZE,
    num_workers=2,
    pin_memory=True,
)

# -------------------------------
# 3. Quantitative eval: loss, PPL, top-k
# -------------------------------
total_loss_t1 = 0.0
total_loss_t2 = 0.0
total_tokens = 0

correct_top1_t1 = 0
correct_top1_t2 = 0
correct_topk_t1 = 0
correct_topk_t2 = 0

steps_done = 0

with torch.no_grad():
    for batch in tqdm(eval_loader, total=EVAL_STEPS, desc="Quantitative eval", unit="step"):
        input_ids = batch["input_ids"].to(DEVICE)   # [B, T]
        labels_t1 = batch["labels_t1"].to(DEVICE)   # [B, T]
        labels_t2 = batch["labels_t2"].to(DEVICE)   # [B, T]

        logits_t1, logits_t2 = model(input_ids)     # [B, T, V] each

        B, T, V = logits_t1.shape

        # Cross-entropy losses
        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
            reduction="sum",
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
            reduction="sum",
        )

        total_loss_t1 += loss_t1.item()
        total_loss_t2 += loss_t2.item()
        total_tokens += labels_t1.numel()

        # Top-1 accuracy
        preds_t1 = logits_t1.argmax(dim=-1)  # [B, T]
        preds_t2 = logits_t2.argmax(dim=-1)  # [B, T]

        correct_top1_t1 += (preds_t1 == labels_t1).sum().item()
        correct_top1_t2 += (preds_t2 == labels_t2).sum().item()

        # Top-k accuracy (k = TOP_K_METRIC)
        topk_t1 = logits_t1.topk(TOP_K_METRIC, dim=-1).indices  # [B, T, K]
        topk_t2 = logits_t2.topk(TOP_K_METRIC, dim=-1).indices  # [B, T, K]

        labels_expanded_t1 = labels_t1.unsqueeze(-1).expand_as(topk_t1)
        labels_expanded_t2 = labels_t2.unsqueeze(-1).expand_as(topk_t2)

        correct_topk_t1 += (topk_t1 == labels_expanded_t1).any(dim=-1).sum().item()
        correct_topk_t2 += (topk_t2 == labels_expanded_t2).any(dim=-1).sum().item()

        steps_done += 1
        if steps_done >= EVAL_STEPS:
            break

# Aggregate metrics
avg_loss_t1 = total_loss_t1 / total_tokens
avg_loss_t2 = total_loss_t2 / total_tokens

ppl_t1 = math.exp(avg_loss_t1)
ppl_t2 = math.exp(avg_loss_t2)

top1_acc_t1 = correct_top1_t1 / total_tokens
top1_acc_t2 = correct_top1_t2 / total_tokens

topk_acc_t1 = correct_topk_t1 / total_tokens
topk_acc_t2 = correct_topk_t2 / total_tokens

print("\n=== Quantitative Evaluation (approximate) ===")
print(f"t+1 head  - loss: {avg_loss_t1:.3f} | ppl: {ppl_t1:.2f} | top-1 acc: {top1_acc_t1:.3f} | top-{TOP_K_METRIC} acc: {topk_acc_t1:.3f}")
print(f"t+2 head  - loss: {avg_loss_t2:.3f} | ppl: {ppl_t2:.2f} | top-1 acc: {top1_acc_t2:.3f} | top-{TOP_K_METRIC} acc: {topk_acc_t2:.3f}")

# -------------------------------
# 4. Qualitative examples with adaptive top-K
# -------------------------------

print(f"\n=== Qualitative Examples (N={N_EXAMPLES}) ===")

# New stream for examples (fresh iterator)
hf_stream_examples = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

examples_shown = 0

with torch.no_grad():
    for example in hf_stream_examples:
        text = example.get("text", "").strip()
        if not text:
            continue

        # Randomly sample from the stream so each run gets different examples.
        # Tune the keep probability as needed; 0.01 works fine in practice.
        if random.random() > 0.01:
            continue

        # Tokenize and ensure we have enough tokens for t+1 and t+2
        ids = tokenizer.encode(text, add_special_tokens=False)
        if len(ids) < BLOCK_SIZE + 2:
            continue

        # Take a slice of length BLOCK_SIZE + 2 to match our training format
        chunk = ids[: BLOCK_SIZE + 2]

        input_ids = torch.tensor(chunk[:BLOCK_SIZE], dtype=torch.long, device=DEVICE).unsqueeze(0)  # [1, T]
        labels_t1 = torch.tensor(chunk[1:BLOCK_SIZE+1], dtype=torch.long, device=DEVICE).unsqueeze(0)
        labels_t2 = torch.tensor(chunk[2:BLOCK_SIZE+2], dtype=torch.long, device=DEVICE).unsqueeze(0)

        logits_t1, logits_t2 = model(input_ids)  # [1, T, V] each

        # We'll inspect the last position in the sequence (index T-1)
        T = input_ids.size(1)
        pos = T - 1

        # Get the context tokens: the last CONTEXT_TOKENS tokens up to and including t.
        # This ensures we see what *immediately precedes* the predicted t+1 and t+2.
        start_idx = max(0, (pos + 1) - CONTEXT_TOKENS)
        context_ids = input_ids[0, start_idx:pos+1].tolist()
        context_text = tokenizer.decode(context_ids, clean_up_tokenization_spaces=True)

        # True tokens at t+1 and t+2 for this position
        true_t1_id = labels_t1[0, pos].item()
        true_t2_id = labels_t2[0, pos].item()
        true_t1_str = repr(tokenizer.decode([true_t1_id]))
        true_t2_str = repr(tokenizer.decode([true_t2_id]))

        # Distributions at this position
        logits_t1_pos = logits_t1[0, pos, :]    # [V]
        logits_t2_pos = logits_t2[0, pos, :]    # [V]

        probs_t1_pos = F.softmax(logits_t1_pos, dim=-1)
        probs_t2_pos = F.softmax(logits_t2_pos, dim=-1)

        # Entropy (in nats) and per-position "perplexity" from the predicted distribution
        def entropy_and_pp(probs):
            eps = 1e-12
            p = probs.clamp(min=eps)
            H = -(p * p.log()).sum()
            PP = torch.exp(H)
            return H.item(), PP.item()

        H1, PP1 = entropy_and_pp(probs_t1_pos)
        H2, PP2 = entropy_and_pp(probs_t2_pos)

        # Adaptive top-K based on entropy (sharper dist => smaller K)
        def adaptive_k(H):
            if H < 1.5:
                return 3
            elif H < 3.0:
                return 5
            else:
                return 10

        K1 = adaptive_k(H1)
        K2 = adaptive_k(H2)

        # Top-k for each head
        topk_t1_probs, topk_t1_ids = torch.topk(probs_t1_pos, K1)
        topk_t2_probs, topk_t2_ids = torch.topk(probs_t2_pos, K2)

        # Print the example
        examples_shown += 1
        print(f"\n--- Example {examples_shown} ---")
        print("Context:")
        print(context_text)  # already restricted by CONTEXT_TOKENS

        print(f"\nTrue next token (t+1):   {true_t1_str}")
        print(f"True token at t+2:       {true_t2_str}")

        print(f"\nHead₁ (t+1) distribution at this position:")
        print(f"  Entropy: {H1:.3f} nats | Token-level perplexity: {PP1:.2f}")
        print(f"  Adaptive top-K = {K1}")
        for p, idx in zip(topk_t1_probs.tolist(), topk_t1_ids.tolist()):
            tok = repr(tokenizer.decode([idx]))
            print(f"  {p:7.4f}  ->  {tok}")

        print(f"\nHead₂ (t+2) distribution at this position:")
        print(f"  Entropy: {H2:.3f} nats | Token-level perplexity: {PP2:.2f}")
        print(f"  Adaptive top-K = {K2}")
        for p, idx in zip(topk_t2_probs.tolist(), topk_t2_ids.tolist()):
            tok = repr(tokenizer.decode([idx]))
            print(f"  {p:7.4f}  ->  {tok}")

        if examples_shown >= N_EXAMPLES:
            break

print("\nEvaluation and qualitative examples complete.")
```

# FutureGPT: Meta-style multi-token Transformer (faithful small-scale port)

In this section we define a **Meta-style multi-token LLaMA decoder**, closely adapted from
the official implementation in `facebook/multi-token-prediction/llama/model.py`, but:

- **Scaled down** to fit comfortably on a single A100:
  - `dim = 1024`
  - `n_layers = 12`
  - `n_heads = 16`
  - `n_future_tokens = 2` (predicts t+1 and t+2)
  - `max_seq_len = 512`
- **Single-device implementation**:
  - Plain `nn.Embedding` and `nn.Linear` instead of `ParallelEmbedding`,
    `ColumnParallelLinear`, and `RowParallelLinear`.
  - No KV caching / `start_pos` handling — we focus on training & evaluation,
    not production inference.
- **Architectural details match Meta’s design**:
  - LLaMA-style internals:
    - RMSNorm
    - RoPE for Q/K
    - SwiGLU FFN: `w2( silu(w1(x)) * w3(x) )`
  - **Trunk + heads layout**:
    - `self.layers`: trunk + the **first prediction head** (t+1)
    - `self.extra_heads`: additional prediction heads (t+2, t+3, …)
    - Total transformer blocks = `n_layers`.
    - Trunk uses `self.layers[:-1]` (i.e., `n_layers - n_future_tokens` blocks).
    - Prediction heads = `[self.layers[-1]] + list(self.extra_heads)`.
  - Shared RMSNorm + shared output projection.

**Output shape:**

- For training / analysis, set `return_all_heads=True`:
  - logits shape: `(batch_size, seq_len, n_future_tokens, vocab_size)`
- For standard next-token decoding (`t+1` only), use `return_all_heads=False`:
  - logits shape: `(batch_size, seq_len, vocab_size)`

We will train this model with a **joint multi-token loss** like Meta:

\[
\mathcal{L} = \sum_{k=1}^{H} \lambda_k \,\mathcal{L}_{t+k}
\]

where `H = n_future_tokens` and \(\lambda_k\) are scalar weights (e.g., 1.0 for t+1,
0.7 for t+2), performing backpropagation through the shared trunk and all heads
simultaneously.


```python
# ============================================================
# FutureGPT: Meta-style multi-token LLaMA decoder (small)
#   - Adapted from facebook/multi-token-prediction/llama/model.py
#   - Single-device, non-parallel implementation
#   - Trunk + prediction heads layout matches Meta:
#       * self.layers: trunk + t+1 head
#       * self.extra_heads: additional heads (t+2, t+3, ...)
# ============================================================

import math
from dataclasses import dataclass
from typing import Optional, Tuple

import torch
import torch.nn as nn
import torch.nn.functional as F


# -------------------------------
# 1. Model hyperparameters
# -------------------------------

@dataclass
class ModelArgs:
    """
    Configuration for a small Meta-style multi-token model.

    This mirrors the structure of Meta's ModelArgs but uses smaller
    defaults suitable for a single A100.
    """
    dim: int = 1024
    n_layers: int = 12
    n_heads: int = 16
    n_kv_heads: Optional[int] = None  # for grouped-query; default: n_heads
    vocab_size: int = 50257           # will be set from tokenizer
    multiple_of: int = 256            # SwiGLU hidden dim multiple
    ffn_dim_multiplier: Optional[float] = None
    norm_eps: float = 1e-5
    n_future_tokens: int = 2          # number of prediction heads (t+1, t+2, ...)
    rope_theta: float = 10000.0

    max_batch_size: int = 32
    max_seq_len: int = 512


# -------------------------------
# 2. RMSNorm (LLaMA-style)
# -------------------------------

class RMSNorm(nn.Module):
    """
    Root Mean Square LayerNorm, as in LLaMA and Meta's MTP models.
    """

    def __init__(self, dim: int, eps: float = 1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def _norm(self, x: torch.Tensor) -> torch.Tensor:
        return x * torch.rsqrt(x.pow(2).mean(-1, keepdim=True) + self.eps)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        output = self._norm(x.float()).type_as(x)
        return output * self.weight


# -------------------------------
# 3. Rotary positional embeddings (RoPE)
# -------------------------------

def precompute_freqs_cis(dim: int, end: int, theta: float = 10000.0) -> torch.Tensor:
    """
    Precompute complex exponentials for RoPE.

    dim:   per-head dimension of Q/K (not the full hidden size).
    end:   maximum sequence length to support.
    theta: base frequency.
    """
    freqs = 1.0 / (theta ** (torch.arange(0, dim, 2)[: (dim // 2)].float() / dim))
    t = torch.arange(end)
    freqs = torch.outer(t, freqs).float()
    freqs_cis = torch.polar(torch.ones_like(freqs), freqs)  # complex64
    return freqs_cis  # [end, dim/2]


def reshape_for_broadcast(freqs_cis: torch.Tensor, x: torch.Tensor) -> torch.Tensor:
    """
    Adjust RoPE frequencies shape to broadcast with x.

    freqs_cis: [seq_len, dim/2]
    x:         [..., seq_len, dim/2] (complex view)
    """
    ndim = x.ndim
    assert 0 <= 1 < ndim
    assert freqs_cis.shape == (x.shape[1], x.shape[-1])
    shape = [d if i == 1 or i == ndim - 1 else 1 for i, d in enumerate(x.shape)]
    return freqs_cis.view(*shape)


def apply_rotary_emb(
    xq: torch.Tensor,
    xk: torch.Tensor,
    freqs_cis: torch.Tensor,
) -> Tuple[torch.Tensor, torch.Tensor]:
    """
    Apply RoPE to query and key tensors.

    xq, xk: [batch, seq, n_heads, head_dim]
    freqs_cis: precomputed RoPE frequencies [max_seq_len, head_dim/2]
    """
    xq_ = torch.view_as_complex(xq.float().reshape(*xq.shape[:-1], -1, 2))
    xk_ = torch.view_as_complex(xk.float().reshape(*xk.shape[:-1], -1, 2))

    freqs_cis = reshape_for_broadcast(freqs_cis, xq_)
    xq_out = torch.view_as_real(xq_ * freqs_cis).flatten(3)
    xk_out = torch.view_as_real(xk_ * freqs_cis).flatten(3)

    return xq_out.type_as(xq), xk_out.type_as(xk)


# -------------------------------
# 4. Multi-head attention (no model parallel)
# -------------------------------

class Attention(nn.Module):
    """
    Multi-head self-attention with RoPE.
    Single-device version (no model parallel, no KV cache).
    """

    def __init__(self, args: ModelArgs):
        super().__init__()
        self.n_heads = args.n_heads
        self.n_kv_heads = args.n_heads if args.n_kv_heads is None else args.n_kv_heads
        assert self.n_kv_heads == self.n_heads, "Grouped KV not implemented."
        self.head_dim = args.dim // args.n_heads

        self.wq = nn.Linear(args.dim, args.n_heads * self.head_dim, bias=False)
        self.wk = nn.Linear(args.dim, self.n_kv_heads * self.head_dim, bias=False)
        self.wv = nn.Linear(args.dim, self.n_kv_heads * self.head_dim, bias=False)
        self.wo = nn.Linear(args.n_heads * self.head_dim, args.dim, bias=False)

    def forward(
        self,
        x: torch.Tensor,
        freqs_cis: torch.Tensor,
        mask: Optional[torch.Tensor],
    ) -> torch.Tensor:
        """
        x: [B, T, dim]
        freqs_cis: [T, head_dim/2] RoPE frequencies
        mask: [1, 1, T, T] causal mask (0 or -inf)
        """
        B, T, _ = x.shape

        q = self.wq(x)
        k = self.wk(x)
        v = self.wv(x)

        q = q.view(B, T, self.n_heads, self.head_dim)
        k = k.view(B, T, self.n_kv_heads, self.head_dim)
        v = v.view(B, T, self.n_kv_heads, self.head_dim)

        # Apply RoPE to Q/K
        q, k = apply_rotary_emb(q, k, freqs_cis=freqs_cis)

        # [B, n_heads, T, head_dim]
        q = q.permute(0, 2, 1, 3)
        k = k.permute(0, 2, 1, 3)
        v = v.permute(0, 2, 1, 3)

        scores = torch.matmul(q, k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        if mask is not None:
            scores = scores + mask  # additive mask: 0 or -inf

        attn = F.softmax(scores.float(), dim=-1).type_as(q)
        out = torch.matmul(attn, v)  # [B, n_heads, T, head_dim]

        out = out.permute(0, 2, 1, 3).contiguous().view(B, T, -1)
        out = self.wo(out)
        return out


# -------------------------------
# 5. SwiGLU FeedForward
# -------------------------------

class FeedForward(nn.Module):
    """
    SwiGLU-style FFN as in LLaMA / Meta:
        FF(x) = W2( silu(W1(x)) * W3(x) )
    """

    def __init__(
        self,
        dim: int,
        hidden_dim: int,
        multiple_of: int,
        ffn_dim_multiplier: Optional[float],
    ):
        super().__init__()
        hidden_dim = int(2 * hidden_dim / 3)
        if ffn_dim_multiplier is not None:
            hidden_dim = int(ffn_dim_multiplier * hidden_dim)
        hidden_dim = multiple_of * ((hidden_dim + multiple_of - 1) // multiple_of)

        self.w1 = nn.Linear(dim, hidden_dim, bias=False)
        self.w2 = nn.Linear(hidden_dim, dim, bias=False)
        self.w3 = nn.Linear(dim, hidden_dim, bias=False)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.w2(F.silu(self.w1(x)) * self.w3(x))


# -------------------------------
# 6. Transformer block (Meta-style)
# -------------------------------

class TransformerBlock(nn.Module):
    """
    Single decoder block:
      - RMSNorm
      - Self-attention with RoPE
      - Residual connection
      - RMSNorm
      - SwiGLU FFN
      - Residual connection
    """

    def __init__(self, layer_id: int, args: ModelArgs):
        super().__init__()
        self.layer_id = layer_id
        self.args = args

        self.attention = Attention(args)
        self.feed_forward = FeedForward(
            dim=args.dim,
            hidden_dim=4 * args.dim,
            multiple_of=args.multiple_of,
            ffn_dim_multiplier=args.ffn_dim_multiplier,
        )
        self.attention_norm = RMSNorm(args.dim, eps=args.norm_eps)
        self.ffn_norm = RMSNorm(args.dim, eps=args.norm_eps)

    def forward(
        self,
        x: torch.Tensor,
        freqs_cis: torch.Tensor,
        mask: Optional[torch.Tensor],
    ) -> torch.Tensor:
        h = x + self.attention(self.attention_norm(x), freqs_cis, mask)
        out = h + self.feed_forward(self.ffn_norm(h))
        return out


# -------------------------------
# 7. Meta-style multi-token Transformer
# -------------------------------

class FutureGPTMultiToken(nn.Module):
    """
    Meta-style multi-token Transformer (small-scale):

    - Token embedding
    - Trunk of transformer blocks
    - Prediction heads implemented as extra transformer blocks
      (first head reuses the final trunk layer, as in Meta's code)
    - Shared RMSNorm + output projection
    - RoPE positional encoding

    Layout (matching Meta-style design):

      * self.layers:
          - length = n_layers - n_future_tokens + 1
          - first (n_layers - n_future_tokens) blocks: trunk-only
          - last block: serves as both final trunk layer and t+1 head

      * self.extra_heads:
          - length = n_future_tokens - 1
          - additional head blocks for t+2, t+3, ...

      Total TransformerBlock count = n_layers.
    """

    def __init__(self, params: ModelArgs):
        super().__init__()
        self.params = params
        self.vocab_size = params.vocab_size
        self.n_layers = params.n_layers
        self.n_future_tokens = params.n_future_tokens

        # Embedding
        self.tok_embeddings = nn.Embedding(params.vocab_size, params.dim)

        # Trunk + first head (as in Meta)
        self.layers = nn.ModuleList()
        for layer_id in range(params.n_layers - self.n_future_tokens + 1):
            self.layers.append(TransformerBlock(layer_id, params))

        # Additional prediction head blocks (t+2, t+3, ...)
        self.extra_heads = nn.ModuleList()
        for layer_id in range(
            params.n_layers - self.n_future_tokens + 1, params.n_layers
        ):
            self.extra_heads.append(TransformerBlock(layer_id, params))

        # Final norm + shared output projection
        self.norm = RMSNorm(params.dim, eps=params.norm_eps)
        self.output = nn.Linear(params.dim, params.vocab_size, bias=False)

        # Precompute RoPE frequencies
        self.register_buffer(
            "freqs_cis",
            precompute_freqs_cis(
                dim=params.dim // params.n_heads,
                end=params.max_seq_len * 2,
                theta=params.rope_theta,
            ),
            persistent=False,
        )

    def _build_causal_mask(self, T: int, device: torch.device) -> torch.Tensor:
        """
        Build a standard causal mask of shape [1, 1, T, T] with 0 for allowed,
        -inf for disallowed positions. Suitable for additive masking.
        """
        mask = torch.full((T, T), float("-inf"), device=device)
        mask = torch.triu(mask, diagonal=1)
        return mask.unsqueeze(0).unsqueeze(0)  # [1, 1, T, T]

    def forward(
        self,
        tokens: torch.Tensor,
        return_all_heads: bool = False,
    ) -> torch.Tensor:
        """
        tokens: [B, T] integer token IDs

        return_all_heads:
          - True:  logits shape [B, T, n_future_tokens, vocab_size]
          - False: logits shape [B, T, vocab_size] (uses t+1 head only)
        """
        B, T = tokens.shape
        device = tokens.device

        assert T <= self.params.max_seq_len, (
            f"Sequence length {T} > max_seq_len {self.params.max_seq_len}"
        )

        # Token embedding
        h = self.tok_embeddings(tokens)  # [B, T, dim]

        # RoPE frequencies for this sequence length
        # freqs_cis: [T, head_dim/2]
        freqs_cis = self.freqs_cis[:T].to(device=h.device)

        # Causal mask: [1, 1, T, T]
        mask = self._build_causal_mask(T, device=h.device)

        # -------------------------
        # Trunk (all but last layer)
        # -------------------------
        for layer in self.layers[:-1]:
            h = layer(h, freqs_cis=freqs_cis, mask=mask)
        h_trunk = h  # [B, T, dim]

        # -------------------------
        # Prediction heads
        #   First head reuses self.layers[-1]
        #   Additional heads come from self.extra_heads
        # -------------------------
        latents = []
        n_heads_to_use = self.n_future_tokens if return_all_heads else 1
        prediction_heads = [self.layers[-1]] + list(self.extra_heads)

        for head_layer in prediction_heads[:n_heads_to_use]:
            h_head = head_layer(h_trunk, freqs_cis=freqs_cis, mask=mask)
            latents.append(h_head)  # each [B, T, dim]

        # Stack head latents: [B, T, H, dim]
        h_all = torch.stack(latents, dim=-2)
        h_all = self.norm(h_all)
        logits_all = self.output(h_all).float()  # [B, T, H, vocab_size]

        if return_all_heads:
            return logits_all

        # Otherwise, return only next-token head (t+1) -> index 0
        logits_t1 = logits_all[:, :, 0, :]  # [B, T, V]
        return logits_t1


# -------------------------------
# 8. Quick sanity check
# -------------------------------

RUN_SANITY_CHECK = False
if RUN_SANITY_CHECK:
    args = ModelArgs(
        dim=1024,
        n_layers=12,
        n_heads=16,
        n_future_tokens=2,
        vocab_size=50257,
        max_seq_len=512,
    )
    model = FutureGPTMultiToken(args)
    print(model)

    x = torch.randint(0, args.vocab_size, (2, 8))
    with torch.no_grad():
        logits_t1 = model(x, return_all_heads=False)
        logits_all = model(x, return_all_heads=True)
    print("logits_t1 shape:", logits_t1.shape)     # [2, 8, V]
    print("logits_all shape:", logits_all.shape)   # [2, 8, 2, V]
```

## Training FutureGPTMultiToken from scratch (Meta-style, equal-head loss)

This cell trains `FutureGPTMultiToken` from scratch on FineWeb (sample-10BT),
using Meta-style multi-token training:

- Trunk computed once.
- Two heads (t+1, t+2) trained sequentially with separate backward passes.
- Equal loss weights per head (raw sum: L = L_t1 + L_t2).
- Streaming FineWeb with:
  - doc-level shuffle using `seed=run_seed`
  - one random window (BLOCK_SIZE+2 tokens) per document
- Checkpoints store:
  - `step` (global training step)
  - `examples_seen` (number of docs consumed)
  - `run_seed` (for data sampling reproducibility)


```python
# ============================================================
# Training FutureGPTMultiToken from scratch (Meta-style)
#   - FutureGPTMultiToken defined above
#   - Two heads (t+1, t+2) with equal loss weights
#   - Streaming FineWeb + doc-local random windows
#   - Checkpoints with step, examples_seen, run_seed
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import os
import math
import random
from dataclasses import dataclass
from typing import Optional

import torch
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset
from tqdm.auto import tqdm

from transformers import (
    AutoTokenizer,
    get_cosine_schedule_with_warmup,
)

# -------------------------------
# 1. Device, dtype, hyperparams
# -------------------------------

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16  # good default on A100

# Model + training configuration
TOKENIZER_NAME = "gpt2"
BLOCK_SIZE = 512
BATCH_SIZE = 24           # adjust if you hit OOM
TRAINING_STEPS = 10_000   # "decent length" to get started
WARMUP_STEPS = 1_000
LOG_EVERY = 100
CHECKPOINT_EVERY = 2_000

LR = 2e-4                 # single LR for all params (training from scratch)
WEIGHT_DECAY = 0.01

# Equal weights for both heads (Meta-style raw sum)
LAMBDA_T1 = 1.0
LAMBDA_T2 = 1.0

# Base seed for reproducible data sampling
BASE_SEED = 1234

# -------------------------------
# 2. Mount Drive for checkpoints
# -------------------------------

from google.colab import drive
drive.mount("/content/drive")

CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
os.makedirs(CKPT_DIR, exist_ok=True)
CKPT_BASENAME = "futuregpt_meta_style_small_fromscratch"

# -------------------------------
# 3. Tokenizer setup (GPT-2 vocab)
# -------------------------------

tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.model_max_length = BLOCK_SIZE + 2  # we manually slice

vocab_size = tokenizer.vocab_size

# -------------------------------
# 4. Instantiate FutureGPTMultiToken
#    (assumes ModelArgs & FutureGPTMultiToken are defined above)
# -------------------------------

args = ModelArgs(
    dim=1024,
    n_layers=12,
    n_heads=16,
    n_future_tokens=2,       # t+1 and t+2
    vocab_size=vocab_size,
    max_seq_len=BLOCK_SIZE,
)

model = FutureGPTMultiToken(args).to(DEVICE)
model.to(dtype)

print(model)

# -------------------------------
# 5. Optimizer & LR scheduler
# -------------------------------

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LR,
    weight_decay=WEIGHT_DECAY,
)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=WARMUP_STEPS,
    num_training_steps=TRAINING_STEPS,
)

# -------------------------------
# 6. Streaming FineWeb + utilities
# -------------------------------

def build_fineweb_stream(examples_seen_start: int, run_seed: int):
    """
    Build a streaming FineWeb dataset, skipping 'examples_seen_start'
    documents and then shuffling the remainder with 'run_seed'.
    """
    hf_stream = load_dataset(
        "HuggingFaceFW/fineweb",
        "sample-10BT",
        split="train",
        streaming=True,
    )

    if examples_seen_start > 0:
        hf_stream = hf_stream.skip(examples_seen_start)

    # Approximate doc-level shuffle with buffer
    hf_stream = hf_stream.shuffle(buffer_size=10_000, seed=run_seed)
    return hf_stream


class FineWebMultiTokenDataset(IterableDataset):
    """
    Iterable dataset that:
      - Reads documents from a streaming HF dataset.
      - Tokenizes each doc with GPT-2 tokenizer.
      - For each doc, picks ONE random window of length BLOCK_SIZE+2:
            chunk = ids[start : start + BLOCK_SIZE + 2]
        where 'start' is uniform over valid offsets.
      - Produces:
            input_ids : positions t .. t+BLOCK_SIZE-1
            labels_t1 : positions t+1 .. t+BLOCK_SIZE
            labels_t2 : positions t+2 .. t+BLOCK_SIZE+1
    """

    def __init__(self, hf_iterable, tokenizer, block_size: int, seed: int):
        super().__init__()
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size
        # Local RNG to avoid relying on global random state
        self.rng = random.Random(seed)

    def __iter__(self):
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if len(ids) < self.block_size + 2:
                # Skip very short docs
                continue

            max_start = len(ids) - (self.block_size + 2)
            start = self.rng.randint(0, max_start)
            chunk = ids[start : start + self.block_size + 2]

            input_ids = chunk[: self.block_size]
            labels_t1 = chunk[1 : self.block_size + 1]
            labels_t2 = chunk[2 : self.block_size + 2]

            yield {
                "input_ids": torch.tensor(input_ids, dtype=torch.long),
                "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
            }


# Initial data progress and seed
examples_seen_start = 0
run_seed = BASE_SEED  # first phase

# Seed RNGs for reproducible sampling
random.seed(run_seed)
torch.manual_seed(run_seed)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(run_seed)

print(f"Using run_seed={run_seed} (examples_seen_start={examples_seen_start}) for initial training phase.")

hf_stream = build_fineweb_stream(examples_seen_start, run_seed)
train_iterable = FineWebMultiTokenDataset(hf_stream, tokenizer, BLOCK_SIZE, seed=run_seed)

train_loader = DataLoader(
    train_iterable,
    batch_size=BATCH_SIZE,
    num_workers=0,     # keep 0 for reproducibility with custom IterableDataset
    pin_memory=True,
)

# -------------------------------
# 7. Helpers: trunk + head forward
# -------------------------------

def forward_trunk(model: FutureGPTMultiToken, input_ids: torch.Tensor):
    """
    Run only the trunk part of the model (all layers except the head blocks)
    matching FutureGPTMultiToken.forward up to h_trunk.
    """
    B, T = input_ids.shape
    device = input_ids.device

    assert T <= model.params.max_seq_len, (
        f"Sequence length {T} > max_seq_len {model.params.max_seq_len}"
    )

    # Token embedding
    h = model.tok_embeddings(input_ids)  # [B, T, dim]

    # RoPE frequencies
    freqs_cis = model.freqs_cis[:T].to(device=h.device)

    # Causal mask
    mask = model._build_causal_mask(T, device=h.device)

    # Trunk: all but last layer
    for layer in model.layers[:-1]:
        h = layer(h, freqs_cis=freqs_cis, mask=mask)

    h_trunk = h
    return h_trunk, freqs_cis, mask


def forward_head_from_trunk(
    model: FutureGPTMultiToken,
    h_trunk: torch.Tensor,
    freqs_cis: torch.Tensor,
    mask: torch.Tensor,
    head_index: int,
):
    """
    Run a single prediction head on shared trunk features.

    head_index:
      0 -> t+1 head (reuses model.layers[-1])
      1 -> t+2 head (model.extra_heads[0])
      2 -> t+3 head (model.extra_heads[1])
      etc.

    Returns:
      logits_k : [B, T, V]
    """
    if head_index == 0:
        head_layer = model.layers[-1]
    else:
        extra_idx = head_index - 1
        head_layer = model.extra_heads[extra_idx]

    h_head = head_layer(h_trunk, freqs_cis=freqs_cis, mask=mask)  # [B, T, dim]
    h_head = model.norm(h_head)                                   # [B, T, dim]
    logits_k = model.output(h_head).float()                       # [B, T, V]
    return logits_k


# -------------------------------
# 8. Training loop (sequential heads)
# -------------------------------

torch.cuda.empty_cache()
model.train()

global_step = 0
examples_seen = examples_seen_start

running_loss_t1 = 0.0
running_loss_t2 = 0.0
running_correct_t1 = 0
running_correct_t2 = 0
running_tokens = 0

progress_bar = tqdm(total=TRAINING_STEPS, desc="FutureGPT initial training", unit="step")

while global_step < TRAINING_STEPS:
    for batch in train_loader:
        global_step += 1
        if global_step > TRAINING_STEPS:
            break

        input_ids = batch["input_ids"].to(DEVICE)   # [B, T]
        labels_t1 = batch["labels_t1"].to(DEVICE)   # [B, T]
        labels_t2 = batch["labels_t2"].to(DEVICE)   # [B, T]

        B, T = input_ids.shape
        V = vocab_size

        # Track "examples seen" as number of docs used for training
        examples_seen += B

        optimizer.zero_grad(set_to_none=True)

        # ---- Trunk forward (shared for all heads) ----
        h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)

        # ---- Head 0: t+1 ----
        logits_t1 = forward_head_from_trunk(
            model, h_trunk, freqs_cis, mask, head_index=0
        )  # [B, T, V]

        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )

        # Backprop for head 0 (retain graph for next head)
        (LAMBDA_T1 * loss_t1).backward(retain_graph=True)

        # ---- Head 1: t+2 ----
        logits_t2 = forward_head_from_trunk(
            model, h_trunk, freqs_cis, mask, head_index=1
        )  # [B, T, V]

        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )

        # Backprop for head 1 (no retain_graph needed)
        (LAMBDA_T2 * loss_t2).backward()

        # Gradient clipping and optimizer step
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        scheduler.step()

        # ---- Metrics ----
        with torch.no_grad():
            preds_t1 = logits_t1.argmax(dim=-1)  # [B, T]
            preds_t2 = logits_t2.argmax(dim=-1)  # [B, T]
            correct_t1 = (preds_t1 == labels_t1).sum().item()
            correct_t2 = (preds_t2 == labels_t2).sum().item()
            total_tokens = labels_t1.numel()

        running_loss_t1 += loss_t1.item()
        running_loss_t2 += loss_t2.item()
        running_correct_t1 += correct_t1
        running_correct_t2 += correct_t2
        running_tokens += total_tokens

        # ---- Logging + checkpointing ----
        if global_step % LOG_EVERY == 0:
            avg_loss_t1 = running_loss_t1 / LOG_EVERY
            avg_loss_t2 = running_loss_t2 / LOG_EVERY
            avg_acc_t1 = running_correct_t1 / running_tokens
            avg_acc_t2 = running_correct_t2 / running_tokens
            lr = scheduler.get_last_lr()[0]

            progress_bar.set_postfix({
                "L_t1": f"{avg_loss_t1:.3f}",
                "L_t2": f"{avg_loss_t2:.3f}",
                "acc_t1": f"{avg_acc_t1:.3f}",
                "acc_t2": f"{avg_acc_t2:.3f}",
                "lr": f"{lr:.2e}",
                "docs_seen": examples_seen,
            })
            progress_bar.update(LOG_EVERY)

            running_loss_t1 = 0.0
            running_loss_t2 = 0.0
            running_correct_t1 = 0
            running_correct_t2 = 0
            running_tokens = 0

            # Periodic checkpoint
            if global_step % CHECKPOINT_EVERY == 0:
                ckpt_path = os.path.join(
                    CKPT_DIR,
                    f"{CKPT_BASENAME}_step_{global_step}.pt"
                )
                torch.save(
                    {
                        "step": global_step,
                        "examples_seen": examples_seen,
                        "run_seed": run_seed,
                        "model_state": model.state_dict(),
                        "optimizer_state": optimizer.state_dict(),
                        "scheduler_state": scheduler.state_dict(),
                        "config": {
                            "BLOCK_SIZE": BLOCK_SIZE,
                            "BATCH_SIZE": BATCH_SIZE,
                            "LR": LR,
                            "TRAINING_STEPS": TRAINING_STEPS,
                        },
                    },
                    ckpt_path,
                )
                print(f"\n[Checkpoint] Saved to {ckpt_path}\n")

progress_bar.close()

final_ckpt_path = os.path.join(CKPT_DIR, f"{CKPT_BASENAME}_final.pt")
torch.save(
    {
        "step": global_step,
        "examples_seen": examples_seen,
        "run_seed": run_seed,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
    },
    final_ckpt_path,
)
print(f"Training finished. Final checkpoint saved to {final_ckpt_path}")
```

## Extra training phase: continuing FutureGPTMultiToken from a checkpoint

This cell continues training **FutureGPTMultiToken** from a saved checkpoint in a
fresh Colab runtime.

Key behavior:

- **Cold-start friendly**:
  - You only need to re-run the model-definition cell, then this cell.
  - The model weights and training metadata are restored from a `.pt` checkpoint.

- **Meta-style training, same objective**:
  - Same two-head objective as the initial phase:
    - `t+1` head and `t+2` head trained via separate backward passes.
    - Equal loss weights:  
      \\( L = L_{t+1} + L_{t+2} \\).
  - Trunk is computed once; each head runs on shared trunk features.

- **Progress-aware data sampling**:
  - The checkpoint stores:
    - `step`: total training steps completed.
    - `examples_seen`: number of documents consumed so far.
  - This extra phase:
    - Uses `examples_seen` to `skip()` past already-seen docs in FineWeb.
    - Derives a new `run_seed = BASE_SEED + start_step` so each phase sees
      a *different* shuffled slice and different random windows.

- **Streaming FineWeb with doc-local windows**:
  - Uses `HuggingFaceFW/fineweb` (`sample-10BT`) in streaming mode.
  - For each document:
    - Tokenize with the GPT-2 tokenizer.
    - Sample one random window of size `BLOCK_SIZE + 2`.
    - Construct:
      - `input_ids`  : positions \\( t \dots t+T-1 \\)
      - `labels_t1`  : positions \\( t+1 \dots t+T \\)
      - `labels_t2`  : positions \\( t+2 \dots t+T+1 \\)

- **Checkpoints for chained phases**:
  - Periodically saves new checkpoints with updated:
    - `step`, `examples_seen`, `run_seed`,
    - `model_state`, `optimizer_state`, `scheduler_state`.
  - Final checkpoint for this extra phase can be used as the starting point
    for yet another continuation phase or for evaluation.

Usage:

1. Run the **model-definition cell** (with `ModelArgs` and `FutureGPTMultiToken`).
2. Set `CKPT_NAME` to the checkpoint you want to resume from.
3. Run this cell to perform another `TRAINING_STEPS_EXTRA` steps of Meta-style
   multi-token training on new data slices.


```python
# ============================================================
# Extra training phase for FutureGPTMultiToken (Meta-style)
#   - Resume from checkpoint (cold start)
#   - New phase uses:
#       * examples_seen from ckpt -> .skip()
#       * run_seed = BASE_SEED + start_step
#   - Equal head weights: L = L_t1 + L_t2
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import os
import math
import random

import torch
import torch.nn.functional as F
from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset
from tqdm.auto import tqdm

from transformers import (
    AutoTokenizer,
    get_cosine_schedule_with_warmup,
)

# -------------------------------
# 1. Device, dtype, base config
# -------------------------------

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16

TOKENIZER_NAME = "gpt2"
BLOCK_SIZE = 512
BATCH_SIZE = 24

# Extra-phase training schedule (can be shorter)
TRAINING_STEPS_EXTRA = 5_000
WARMUP_STEPS_EXTRA = 500
LOG_EVERY = 100
CHECKPOINT_EVERY = 2_000

# Learning rates for extra phase
LR_EXTRA = 1e-4      # could be same as initial or a bit lower
WEIGHT_DECAY = 0.01

# Equal head weights (raw sum)
LAMBDA_T1 = 1.0
LAMBDA_T2 = 1.0

BASE_SEED = 1234

# -------------------------------
# 2. Mount Drive & select ckpt
# -------------------------------

from google.colab import drive
drive.mount("/content/drive")

CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
os.makedirs(CKPT_DIR, exist_ok=True)

# Pick which checkpoint to resume from:
CKPT_NAME = "futuregpt_meta_style_small_fromscratch_final.pt"  # <-- edit if needed
CKPT_PATH = os.path.join(CKPT_DIR, CKPT_NAME)
print(f"Resuming from checkpoint: {CKPT_PATH}")

# -------------------------------
# 3. Tokenizer and model rebuild
#    (assumes ModelArgs & FutureGPTMultiToken are defined)
# -------------------------------

tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.model_max_length = BLOCK_SIZE + 2

vocab_size = tokenizer.vocab_size

args = ModelArgs(
    dim=1024,
    n_layers=12,
    n_heads=16,
    n_future_tokens=2,
    vocab_size=vocab_size,
    max_seq_len=BLOCK_SIZE,
)

model = FutureGPTMultiToken(args).to(DEVICE)
model.to(dtype)

# -------------------------------
# 4. Load checkpoint weights + metadata
# -------------------------------

ckpt = torch.load(CKPT_PATH, map_location=DEVICE)

model.load_state_dict(ckpt["model_state"])
start_step = ckpt.get("step", 0)
examples_seen_start = ckpt.get("examples_seen", 0)

# Derive a new run_seed for this phase based on total steps so far
run_seed = BASE_SEED + start_step
print(f"Checkpoint loaded. step={start_step}, examples_seen_start={examples_seen_start}, run_seed={run_seed}")

# Seed RNGs
random.seed(run_seed)
torch.manual_seed(run_seed)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(run_seed)

# -------------------------------
# 5. Extra-phase optimizer & scheduler
#    (new schedule, as in your earlier GPT-2 extra phase)
# -------------------------------

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LR_EXTRA,
    weight_decay=WEIGHT_DECAY,
)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=WARMUP_STEPS_EXTRA,
    num_training_steps=TRAINING_STEPS_EXTRA,
)

# -------------------------------
# 6. Streaming FineWeb with skip + shuffle
# -------------------------------

def build_fineweb_stream(examples_seen_start: int, run_seed: int):
    hf_stream = load_dataset(
        "HuggingFaceFW/fineweb",
        "sample-10BT",
        split="train",
        streaming=True,
    )
    if examples_seen_start > 0:
        hf_stream = hf_stream.skip(examples_seen_start)
    hf_stream = hf_stream.shuffle(buffer_size=10_000, seed=run_seed)
    return hf_stream


class FineWebMultiTokenDataset(IterableDataset):
    def __init__(self, hf_iterable, tokenizer, block_size: int, seed: int):
        super().__init__()
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size
        self.rng = random.Random(seed)

    def __iter__(self):
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
            )
            if len(ids) < self.block_size + 2:
                continue

            max_start = len(ids) - (self.block_size + 2)
            start = self.rng.randint(0, max_start)
            chunk = ids[start : start + self.block_size + 2]

            input_ids = chunk[: self.block_size]
            labels_t1 = chunk[1 : self.block_size + 1]
            labels_t2 = chunk[2 : self.block_size + 2]

            yield {
                "input_ids": torch.tensor(input_ids, dtype=torch.long),
                "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
            }


hf_stream_extra = build_fineweb_stream(examples_seen_start, run_seed)
train_iterable_extra = FineWebMultiTokenDataset(
    hf_stream_extra, tokenizer, BLOCK_SIZE, seed=run_seed
)

train_loader_extra = DataLoader(
    train_iterable_extra,
    batch_size=BATCH_SIZE,
    num_workers=0,
    pin_memory=True,
)

# -------------------------------
# 7. Helpers (reuse trunk/head logic)
# -------------------------------

def forward_trunk(model: FutureGPTMultiToken, input_ids: torch.Tensor):
    B, T = input_ids.shape
    device = input_ids.device

    assert T <= model.params.max_seq_len, (
        f"Sequence length {T} > max_seq_len {model.params.max_seq_len}"
    )

    h = model.tok_embeddings(input_ids)
    freqs_cis = model.freqs_cis[:T].to(device=h.device)
    mask = model._build_causal_mask(T, device=h.device)

    for layer in model.layers[:-1]:
        h = layer(h, freqs_cis=freqs_cis, mask=mask)

    return h, freqs_cis, mask


def forward_head_from_trunk(
    model: FutureGPTMultiToken,
    h_trunk: torch.Tensor,
    freqs_cis: torch.Tensor,
    mask: torch.Tensor,
    head_index: int,
):
    if head_index == 0:
        head_layer = model.layers[-1]
    else:
        extra_idx = head_index - 1
        head_layer = model.extra_heads[extra_idx]

    h_head = head_layer(h_trunk, freqs_cis=freqs_cis, mask=mask)
    h_head = model.norm(h_head)
    logits_k = model.output(h_head).float()
    return logits_k


# -------------------------------
# 8. Extra training loop
# -------------------------------

torch.cuda.empty_cache()
model.train()

extra_step = 0
global_step = start_step
examples_seen = examples_seen_start

running_loss_t1 = 0.0
running_loss_t2 = 0.0
running_correct_t1 = 0
running_correct_t2 = 0
running_tokens = 0

progress_bar = tqdm(total=TRAINING_STEPS_EXTRA, desc="FutureGPT extra training", unit="step")

while extra_step < TRAINING_STEPS_EXTRA:
    for batch in train_loader_extra:
        extra_step += 1
        global_step += 1
        if extra_step > TRAINING_STEPS_EXTRA:
            break

        input_ids = batch["input_ids"].to(DEVICE)
        labels_t1 = batch["labels_t1"].to(DEVICE)
        labels_t2 = batch["labels_t2"].to(DEVICE)

        B, T = input_ids.shape
        V = vocab_size

        examples_seen += B

        optimizer.zero_grad(set_to_none=True)

        # Trunk shared
        h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)

        # Head t+1
        logits_t1 = forward_head_from_trunk(
            model, h_trunk, freqs_cis, mask, head_index=0
        )
        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
        )
        (LAMBDA_T1 * loss_t1).backward(retain_graph=True)

        # Head t+2
        logits_t2 = forward_head_from_trunk(
            model, h_trunk, freqs_cis, mask, head_index=1
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
        )
        (LAMBDA_T2 * loss_t2).backward()

        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        scheduler.step()

        with torch.no_grad():
            preds_t1 = logits_t1.argmax(dim=-1)
            preds_t2 = logits_t2.argmax(dim=-1)
            correct_t1 = (preds_t1 == labels_t1).sum().item()
            correct_t2 = (preds_t2 == labels_t2).sum().item()
            total_tokens = labels_t1.numel()

        running_loss_t1 += loss_t1.item()
        running_loss_t2 += loss_t2.item()
        running_correct_t1 += correct_t1
        running_correct_t2 += correct_t2
        running_tokens += total_tokens

        if extra_step % LOG_EVERY == 0:
            avg_loss_t1 = running_loss_t1 / LOG_EVERY
            avg_loss_t2 = running_loss_t2 / LOG_EVERY
            avg_acc_t1 = running_correct_t1 / running_tokens
            avg_acc_t2 = running_correct_t2 / running_tokens
            lr = scheduler.get_last_lr()[0]

            progress_bar.set_postfix({
                "L_t1": f"{avg_loss_t1:.3f}",
                "L_t2": f"{avg_loss_t2:.3f}",
                "acc_t1": f"{avg_acc_t1:.3f}",
                "acc_t2": f"{avg_acc_t2:.3f}",
                "lr": f"{lr:.2e}",
                "docs_seen": examples_seen,
            })
            progress_bar.update(LOG_EVERY)

            running_loss_t1 = 0.0
            running_loss_t2 = 0.0
            running_correct_t1 = 0
            running_correct_t2 = 0
            running_tokens = 0

            if extra_step % CHECKPOINT_EVERY == 0:
                ckpt_path = os.path.join(
                    CKPT_DIR,
                    f"futuregpt_meta_style_small_extra_step_{global_step}.pt"
                )
                torch.save(
                    {
                        "step": global_step,
                        "examples_seen": examples_seen,
                        "run_seed": run_seed,
                        "model_state": model.state_dict(),
                        "optimizer_state": optimizer.state_dict(),
                        "scheduler_state": scheduler.state_dict(),
                        "config": {
                            "BLOCK_SIZE": BLOCK_SIZE,
                            "BATCH_SIZE": BATCH_SIZE,
                            "LR_EXTRA": LR_EXTRA,
                            "TRAINING_STEPS_EXTRA": TRAINING_STEPS_EXTRA,
                        },
                    },
                    ckpt_path,
                )
                print(f"\n[Checkpoint] Saved to {ckpt_path}\n")

progress_bar.close()

final_ckpt_path = os.path.join(
    CKPT_DIR,
    f"futuregpt_meta_style_small_final.pt"
)
torch.save(
    {
        "step": global_step,
        "examples_seen": examples_seen,
        "run_seed": run_seed,
        "model_state": model.state_dict(),
        "optimizer_state": optimizer.state_dict(),
        "scheduler_state": scheduler.state_dict(),
    },
    final_ckpt_path,
)
print(f"Extra training finished. Final checkpoint saved to {final_ckpt_path}")
```

## Evaluation: loss, perplexity, accuracies, and qualitative examples

This cell evaluates a saved **FutureGPTMultiToken** checkpoint on a small slice
of FineWeb.

What it does:

- **Loads latest checkpoint**:
  - Scans `CKPT_DIR` for files matching `CKPT_BASENAME`.
  - Prefers the `_final.pt` file if it exists, otherwise uses the highest-step
    checkpoint (largest `step` in the filename).
  - Rebuilds `FutureGPTMultiToken` with the same config and loads `model_state`.

- **Eval dataset (streaming FineWeb)**:
  - Uses `HuggingFaceFW/fineweb` (`sample-10BT`) in streaming mode.
  - For each document:
    - Tokenizes with the GPT-2 tokenizer.
    - Extracts **one random window** of length `BLOCK_SIZE + 2`.
    - Constructs:
      - `input_ids`: positions \\( t \dots t+T-1 \\)
      - `labels_t1`: positions \\( t+1 \dots t+T \\)
      - `labels_t2`: positions \\( t+2 \dots t+T+1 \\)
  - We iterate over a limited number of batches (e.g., `EVAL_STEPS`) to keep
    the eval lightweight.

- **Metrics** (per token, averaged over eval slice):
  - Cross-entropy loss for:
    - t+1 head (head 0)
    - t+2 head (head 1)
  - Perplexity for each head:
    - \\( \text{ppl}_{t+1} = \exp(L_{t+1}) \\)
    - \\( \text{ppl}_{t+2} = \exp(L_{t+2}) \\)
  - Top-1 accuracy: fraction of tokens where the argmax prediction is correct.
  - Top-k accuracy (k = 5): fraction where the true token is in the top-k.

- **Qualitative examples**:
  - Prints several random eval examples:
    - Decoded **context** (input window).
    - True **t+1** and **t+2** tokens at the last position.
    - Model’s top-k predictions (token + probability) for each head.

Usage:

1. Run the **model-definition cell** (with `ModelArgs` and `FutureGPTMultiToken`).
2. Set `CKPT_DIR` and `CKPT_BASENAME` to match your training setup.
3. Run this cell to evaluate the latest checkpoint.


```python
# ============================================================
# Evaluation for FutureGPTMultiToken:
#   - Load latest checkpoint
#   - Compute loss, perplexity, top-1 and top-k (k=5) accuracy
#   - Print qualitative examples with context and predictions
# ============================================================

!pip install -q "transformers>=4.44.0" "datasets>=2.20.0" accelerate

import os
import math
import random
import glob
import torch
import torch.nn.functional as F

from torch.utils.data import IterableDataset, DataLoader
from datasets import load_dataset
from tqdm.auto import tqdm

from transformers import AutoTokenizer

# -------------------------------
# 1. Device, dtype, eval config
# -------------------------------

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16  # for consistency with training

TOKENIZER_NAME = "gpt2"
BLOCK_SIZE = 512
BATCH_SIZE_EVAL = 16     # eval batch size (can adjust)
EVAL_STEPS = 100         # number of eval batches
TOP_K = 5                # for top-k accuracy and printouts

# NEW: limit how much context we print for each qualitative example
MAX_CONTEXT_TOKENS = 80   # tail length in tokens for display

# -------------------------------
# 2. Mount Drive & checkpoint dir
# -------------------------------

from google.colab import drive
drive.mount("/content/drive")

CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
CKPT_BASENAME = "futuregpt_meta_style_small_fromscratch"

os.makedirs(CKPT_DIR, exist_ok=True)

# -------------------------------
# 3. Helper: find latest checkpoint
# -------------------------------

def find_latest_checkpoint(ckpt_dir: str, basename: str):
    """
    Returns path to the 'latest' checkpoint:
      - Prefer *_final.pt if present.
      - Otherwise choose the *_step_XXXX.pt with the largest step.
    """
    pattern = os.path.join(ckpt_dir, f"{basename}*.pt")
    ckpts = glob.glob(pattern)
    if not ckpts:
        raise FileNotFoundError(f"No checkpoints found matching {pattern}")

    # Prefer final
    final_ckpts = [p for p in ckpts if p.endswith("_final.pt")]
    if final_ckpts:
        # If multiple, just pick the first (shouldn't happen)
        return final_ckpts[0]

    # Otherwise pick max step from *_step_{step}.pt
    best_path = None
    best_step = -1
    for p in ckpts:
        fname = os.path.basename(p)
        # Expect something like: {basename}_step_10000.pt
        parts = fname.split("_step_")
        if len(parts) != 2:
            continue
        step_part = parts[1].split(".pt")[0]
        try:
            step = int(step_part)
        except ValueError:
            continue
        if step > best_step:
            best_step = step
            best_path = p

    if best_path is None:
        raise RuntimeError(f"Could not parse steps from checkpoints: {ckpts}")

    return best_path

ckpt_path = find_latest_checkpoint(CKPT_DIR, CKPT_BASENAME)
print(f"Loading checkpoint from: {ckpt_path}")

# -------------------------------
# 4. Rebuild model & load weights
#    (requires ModelArgs and FutureGPTMultiToken defined above)
# -------------------------------

tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
# We slice manually, so set a large max_length to silence warnings.
tokenizer.model_max_length = 10_000_000

vocab_size = tokenizer.vocab_size

eval_args = ModelArgs(
    dim=1024,
    n_layers=12,
    n_heads=16,
    n_future_tokens=2,
    vocab_size=vocab_size,
    max_seq_len=BLOCK_SIZE,
)

model = FutureGPTMultiToken(eval_args).to(DEVICE)
model.to(dtype)

ckpt = torch.load(ckpt_path, map_location=DEVICE)
model.load_state_dict(ckpt["model_state"])
start_step = ckpt.get("step", None)
print(f"Checkpoint loaded. step={start_step}")

model.eval()

# -------------------------------
# 5. Eval dataset (streaming FineWeb)
# -------------------------------

print("Preparing streaming eval dataset from FineWeb (sample-10BT)...")

hf_stream_eval = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

class FineWebEvalDataset(IterableDataset):
    """
    Streaming eval dataset:

      - Reads documents from FineWeb stream.
      - Tokenizes each document with GPT-2 tokenizer.
      - For each document, samples ONE random window of length BLOCK_SIZE + 2:
            chunk = ids[start : start + BLOCK_SIZE + 2]
        where start is uniform over valid offsets.

      Produces:
            input_ids : positions t .. t+BLOCK_SIZE-1
            labels_t1 : positions t+1 .. t+BLOCK_SIZE
            labels_t2 : positions t+2 .. t+BLOCK_SIZE+1
    """

    def __init__(self, hf_iterable, tokenizer, block_size: int, seed: int = 1234):
        super().__init__()
        self.hf_iterable = hf_iterable
        self.tokenizer = tokenizer
        self.block_size = block_size
        self.seed = seed

    def __iter__(self):
        rng = random.Random(self.seed)
        for example in self.hf_iterable:
            text = example.get("text", "")
            if not text:
                continue

            ids = self.tokenizer.encode(
                text,
                add_special_tokens=False,
                truncation=False,
                max_length=None,
            )
            if len(ids) < self.block_size + 2:
                continue  # skip very short docs

            max_start = len(ids) - (self.block_size + 2)
            start = rng.randint(0, max_start)
            chunk = ids[start : start + self.block_size + 2]

            input_ids = chunk[: self.block_size]
            labels_t1 = chunk[1 : self.block_size + 1]
            labels_t2 = chunk[2 : self.block_size + 2]

            yield {
                "input_ids": torch.tensor(input_ids, dtype=torch.long),
                "labels_t1": torch.tensor(labels_t1, dtype=torch.long),
                "labels_t2": torch.tensor(labels_t2, dtype=torch.long),
            }

EVAL_SEED = 999  # fixed seed for deterministic eval sampling
eval_iterable = FineWebEvalDataset(hf_stream_eval, tokenizer, BLOCK_SIZE, seed=EVAL_SEED)

eval_loader = DataLoader(
    eval_iterable,
    batch_size=BATCH_SIZE_EVAL,
    num_workers=2,
    pin_memory=True,
)

# -------------------------------
# 6. Evaluation loop
# -------------------------------

total_loss_t1 = 0.0
total_loss_t2 = 0.0
total_tokens = 0

total_correct_t1 = 0
total_correct_t2 = 0

total_correct_topk_t1 = 0
total_correct_topk_t2 = 0

# We'll also stash a few batches for qualitative examples
saved_batches = []
MAX_SAVED_BATCHES = 3

progress_bar = tqdm(total=EVAL_STEPS, desc="FutureGPT eval", unit="batch")

with torch.no_grad():
    batches_seen = 0
    for batch in eval_loader:
        if batches_seen >= EVAL_STEPS:
            break
        batches_seen += 1

        input_ids = batch["input_ids"].to(DEVICE)   # [B, T]
        labels_t1 = batch["labels_t1"].to(DEVICE)   # [B, T]
        labels_t2 = batch["labels_t2"].to(DEVICE)   # [B, T]

        B, T = input_ids.shape
        V = vocab_size
        total_tokens += B * T

        # Full multi-head logits: [B, T, H, V]
        logits_all = model(input_ids, return_all_heads=True)

        logits_t1 = logits_all[:, :, 0, :]  # [B, T, V]
        logits_t2 = logits_all[:, :, 1, :]  # [B, T, V]

        # Losses
        loss_t1 = F.cross_entropy(
            logits_t1.view(B * T, V).float(),
            labels_t1.view(B * T),
            reduction="sum",
        )
        loss_t2 = F.cross_entropy(
            logits_t2.view(B * T, V).float(),
            labels_t2.view(B * T),
            reduction="sum",
        )

        total_loss_t1 += loss_t1.item()
        total_loss_t2 += loss_t2.item()

        # Top-1 accuracy
        preds_t1 = logits_t1.argmax(dim=-1)  # [B, T]
        preds_t2 = logits_t2.argmax(dim=-1)

        total_correct_t1 += (preds_t1 == labels_t1).sum().item()
        total_correct_t2 += (preds_t2 == labels_t2).sum().item()

        # Top-k accuracy
        topk_t1 = logits_t1.topk(TOP_K, dim=-1).indices  # [B, T, K]
        topk_t2 = logits_t2.topk(TOP_K, dim=-1).indices

        # Broadcast labels to [B, T, K] and compare
        labels_t1_exp = labels_t1.unsqueeze(-1).expand_as(topk_t1)
        labels_t2_exp = labels_t2.unsqueeze(-1).expand_as(topk_t2)

        total_correct_topk_t1 += (topk_t1 == labels_t1_exp).any(dim=-1).sum().item()
        total_correct_topk_t2 += (topk_t2 == labels_t2_exp).any(dim=-1).sum().item()

        # Save some batches for qualitative examples (CPU copies)
        if len(saved_batches) < MAX_SAVED_BATCHES:
            saved_batches.append({
                "input_ids": input_ids.cpu(),
                "labels_t1": labels_t1.cpu(),
                "labels_t2": labels_t2.cpu(),
                "logits_t1": logits_t1.cpu(),
                "logits_t2": logits_t2.cpu(),
            })

        progress_bar.update(1)

progress_bar.close()

# -------------------------------
# 7. Aggregate metrics
# -------------------------------

avg_loss_t1 = total_loss_t1 / total_tokens
avg_loss_t2 = total_loss_t2 / total_tokens

ppl_t1 = math.exp(avg_loss_t1)
ppl_t2 = math.exp(avg_loss_t2)

top1_acc_t1 = total_correct_t1 / total_tokens
top1_acc_t2 = total_correct_t2 / total_tokens

topk_acc_t1 = total_correct_topk_t1 / total_tokens
topk_acc_t2 = total_correct_topk_t2 / total_tokens

print("\n=== Quantitative evaluation ===")
print(f"Tokens evaluated        : {total_tokens}")
print(f"Avg loss (t+1)          : {avg_loss_t1:.4f}")
print(f"Avg loss (t+2)          : {avg_loss_t2:.4f}")
print(f"Perplexity (t+1)        : {ppl_t1:.2f}")
print(f"Perplexity (t+2)        : {ppl_t2:.2f}")
print(f"Top-1 accuracy (t+1)    : {top1_acc_t1:.4f}")
print(f"Top-1 accuracy (t+2)    : {top1_acc_t2:.4f}")
print(f"Top-{TOP_K} accuracy (t+1): {topk_acc_t1:.4f}")
print(f"Top-{TOP_K} accuracy (t+2): {topk_acc_t2:.4f}")

# -------------------------------
# 8. Qualitative examples
# -------------------------------

def print_example(idx_batch: int, idx_seq: int, k: int = TOP_K,
                  max_context_tokens: int = MAX_CONTEXT_TOKENS):
    """
    Print a single example:
      - decoded context (truncated to last `max_context_tokens` tokens)
      - true t+1 and t+2 at the last position
      - model's top-k predictions for t+1 and t+2
    """
    batch = saved_batches[idx_batch]
    input_ids = batch["input_ids"][idx_seq]   # [T]
    labels_t1 = batch["labels_t1"][idx_seq]
    labels_t2 = batch["labels_t2"][idx_seq]
    logits_t1 = batch["logits_t1"][idx_seq]   # [T, V]
    logits_t2 = batch["logits_t2"][idx_seq]

    # We'll look at the last position in the window
    pos = input_ids.shape[0] - 1

    all_tokens = input_ids.tolist()
    # Only show up to the last `max_context_tokens` tokens
    start_ctx = max(0, pos + 1 - max_context_tokens)
    context_tokens = all_tokens[start_ctx : pos + 1]

    context_text = tokenizer.decode(context_tokens)
    if start_ctx > 0:
        context_text = "..." + context_text

    true_t1 = labels_t1[pos].item()
    true_t2 = labels_t2[pos].item()

    # Top-k for this position
    probs_t1 = F.softmax(logits_t1[pos], dim=-1)
    probs_t2 = F.softmax(logits_t2[pos], dim=-1)

    topk_probs_t1, topk_ids_t1 = torch.topk(probs_t1, k)
    topk_probs_t2, topk_ids_t2 = torch.topk(probs_t2, k)

    print("\n----------------------------------------")
    print(f"[Example batch {idx_batch}, seq {idx_seq}, pos {pos}]")
    print("Context (truncated tail):")
    print(context_text.replace("\n", "\\n"))
    print("\nTrue next tokens:")
    print(f"  t+1: {tokenizer.decode([true_t1])!r} (id={true_t1})")
    print(f"  t+2: {tokenizer.decode([true_t2])!r} (id={true_t2})")

    print("\nModel top-k predictions at this position:")
    print("  t+1 head:")
    for rank in range(k):
        tok_id = topk_ids_t1[rank].item()
        prob = topk_probs_t1[rank].item()
        print(f"    #{rank+1}: {tokenizer.decode([tok_id])!r} (id={tok_id}, p={prob:.4f})")

    print("  t+2 head:")
    for rank in range(k):
        tok_id = topk_ids_t2[rank].item()
        prob = topk_probs_t2[rank].item()
        print(f"    #{rank+1}: {tokenizer.decode([tok_id])!r} (id={tok_id}, p={prob:.4f})")


print("\n=== Qualitative examples ===")
if not saved_batches:
    print("No saved batches to display (eval too short).")
else:
    # NEW: sample random (batch, seq) pairs for qualitative inspection
    N_QUAL_EXAMPLES = 5  # how many random examples to print

    all_pairs = []
    for b in range(len(saved_batches)):
        B = saved_batches[b]["input_ids"].shape[0]
        for s in range(B):
            all_pairs.append((b, s))

    if not all_pairs:
        print("Saved batches are empty; nothing to display.")
    else:
        num_to_show = min(N_QUAL_EXAMPLES, len(all_pairs))
        sampled_pairs = random.sample(all_pairs, num_to_show)

        for (b, s) in sampled_pairs:
            print_example(idx_batch=b, idx_seq=s, k=TOP_K,
                          max_context_tokens=MAX_CONTEXT_TOKENS)
```

# Analyzing the Final Hidden Layer Visually

### Setup: imports & helpers

```python
# ============================================
# Embedding analysis toolkit: imports & helpers
# ============================================
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def cosine_sim(a, b):
    """Cosine similarity between two vectors (last dim)."""
    a = F.normalize(a, dim=-1)
    b = F.normalize(b, dim=-1)
    return (a * b).sum(dim=-1)

def mean_pool(hidden_states, attention_mask=None):
    """
    Simple mean pooling over sequence dimension.
    hidden_states: [B, T, D]
    attention_mask: [B, T] (1 for real tokens, 0 for padding) or None.
    """
    if attention_mask is None:
        return hidden_states.mean(dim=1)
    mask = attention_mask.unsqueeze(-1).float()
    summed = (hidden_states * mask).sum(dim=1)
    count = mask.sum(dim=1).clamp(min=1.0)
    return summed / count
```

### Hook into FutureGPTMultiToken final hidden states

```python
# ============================================
# FutureGPTMultiToken: get final hidden states
# ============================================

def futuregpt_load_from_checkpoint(ckpt_path, tokenizer, dim=1024, n_layers=12,
                                   n_heads=16, n_future_tokens=2, block_size=512):
    """
    Rebuild FutureGPTMultiToken and load weights from a checkpoint.
    Assumes ModelArgs and FutureGPTMultiToken are already defined above.
    """
    args = ModelArgs(
        dim=dim,
        n_layers=n_layers,
        n_heads=n_heads,
        n_future_tokens=n_future_tokens,
        vocab_size=tokenizer.vocab_size,
        max_seq_len=block_size,
    )
    model = FutureGPTMultiToken(args).to(DEVICE)
    ckpt = torch.load(ckpt_path, map_location=DEVICE)
    model.load_state_dict(ckpt["model_state"])
    model.eval()
    return model


@torch.no_grad()
def futuregpt_hidden_states(model, input_ids, head_index=0):
    """
    Extract final hidden states for a given head from FutureGPTMultiToken.

    input_ids: [B, T]
    head_index:
      0 -> t+1 head (reuses model.layers[-1])
      1 -> t+2 head (model.extra_heads[0])
      ...
    Returns:
      h_final: [B, T, D]  (normalized, right before vocab projection)
    """
    model.eval()
    input_ids = input_ids.to(DEVICE)
    B, T = input_ids.shape

    # Embed tokens
    h = model.tok_embeddings(input_ids)  # [B, T, D]

    # RoPE freqs & mask
    freqs_cis = model.freqs_cis[:T].to(h.device)
    mask = model._build_causal_mask(T, device=h.device)

    # Trunk (all but last layer)
    for layer in model.layers[:-1]:
        h = layer(h, freqs_cis=freqs_cis, mask=mask)
    h_trunk = h  # [B, T, D]

    # Select head
    if head_index == 0:
        head_layer = model.layers[-1]
    else:
        extra_idx = head_index - 1
        head_layer = model.extra_heads[extra_idx]

    h_head = head_layer(h_trunk, freqs_cis=freqs_cis, mask=mask)  # [B, T, D]
    h_final = model.norm(h_head)  # [B, T, D]

    return h_final  # contextual token embeddings for that head
```

#### example

```python
# Example: load latest checkpoint and get token embeddings
from transformers import AutoTokenizer

TOKENIZER_NAME = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

CKPT_PATH = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points/futuregpt_meta_style_small_fromscratch_final.pt"

futuregpt = futuregpt_load_from_checkpoint(CKPT_PATH, tokenizer)
text = "The knight gazed at the stars and whispered, 'I love you.'"
enc = tokenizer(text, return_tensors="pt")
h_future = futuregpt_hidden_states(futuregpt, enc["input_ids"], head_index=0)  # [1, T, D]
```

### Hook into a stronger pretrained model

```python
# ============================================
# Pretrained GPT-2: get final hidden states
# ============================================
from transformers import GPT2LMHeadModel, AutoTokenizer

PRETRAINED_NAME = "gpt2-medium"  # or "gpt2-large" if VRAM allows

pt_tokenizer = AutoTokenizer.from_pretrained(PRETRAINED_NAME)
if pt_tokenizer.pad_token is None:
    pt_tokenizer.pad_token = pt_tokenizer.eos_token

pt_model = GPT2LMHeadModel.from_pretrained(
    PRETRAINED_NAME,
    output_hidden_states=True,
).to(DEVICE)
pt_model.eval()

@torch.no_grad()
def gpt2_hidden_states(texts):
    """
    Get final hidden states from a pretrained GPT-2 model.
    texts: list of strings
    Returns:
      hidden: [B, T, D]
      input_ids: [B, T]
      attn_mask: [B, T]
    """
    enc = pt_tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
    ).to(DEVICE)
    outputs = pt_model(**enc)
    # last_hidden_state: [B, T, D]
    hidden = outputs.hidden_states[-1]
    return hidden, enc["input_ids"], enc["attention_mask"]
```

### Word-level map for a single sentence

```python
# ============================================
# A) Token-level map for a single sentence
# ============================================
sentence = "The knight gazed at the stars and whispered, 'I love you.'"
hidden, input_ids, attn_mask = gpt2_hidden_states([sentence])  # [1, T, D]
hidden = hidden[0]                  # [T, D]
tokens = pt_tokenizer.convert_ids_to_tokens(input_ids[0])

# PCA → 2D
pca = PCA(n_components=2)
coords = pca.fit_transform(hidden.cpu().numpy())  # [T, 2]

plt.figure(figsize=(8, 6))
for (x, y), tok in zip(coords, tokens):
    plt.scatter(x, y)
    plt.text(x + 0.01, y + 0.01, tok, fontsize=9)
plt.title("GPT-2 token embeddings (final hidden layer)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

### Sentence-level semantics map

```python
# ============================================
# B) Sentence-level map (semantics / style)
# ============================================

sentences = [
    "I love you more than words can say.",
    "She kissed him softly under the moonlight.",
    "The nuclear reactor began to overheat rapidly.",
    "The defendant was found guilty on all charges.",
    "The weather today is sunny and pleasant.",
    "I hate you and everything you stand for.",
]

hidden, input_ids, attn_mask = gpt2_hidden_states(sentences)  # [B, T, D]
sentence_vecs = mean_pool(hidden, attn_mask)                  # [B, D]

pca = PCA(n_components=2)
coords = pca.fit_transform(sentence_vecs.cpu().numpy())       # [B, 2]

plt.figure(figsize=(8, 6))
for (x, y), s in zip(coords, sentences):
    plt.scatter(x, y)
    # Shorten label
    label = s if len(s) < 40 else s[:37] + "..."
    plt.text(x + 0.01, y + 0.01, label, fontsize=8)
plt.title("Sentence-level embedding map (GPT-2)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

### Concept directions (romance, toxicity, etc.)

```python
# ============================================
# C) Concept direction: romance vs neutral
# ============================================

romantic_sentences = [
    "I love you more than anything in this world.",
    "Her heart raced as he pulled her close.",
    "Their fingers intertwined as they watched the sunset.",
    "He whispered sweet words into her ear.",
]

neutral_sentences = [
    "The report is due on Monday morning.",
    "She placed the book on the table.",
    "The train arrived at the station on time.",
    "He logged into his computer and checked his email.",
]

h_rom, _, mask_rom = gpt2_hidden_states(romantic_sentences)
h_neu, _, mask_neu = gpt2_hidden_states(neutral_sentences)

vec_rom = mean_pool(h_rom, mask_rom).mean(dim=0, keepdim=True)  # [1, D]
vec_neu = mean_pool(h_neu, mask_neu).mean(dim=0, keepdim=True)  # [1, D]

romance_dir = (vec_rom - vec_neu)  # [1, D]
romance_dir = F.normalize(romance_dir, dim=-1)

def romance_score(text):
    h, _, m = gpt2_hidden_states([text])
    sent_vec = mean_pool(h, m)  # [1, D]
    return cosine_sim(sent_vec, romance_dir)[0].item()

test_texts = [
    "He gently kissed her forehead and smiled.",
    "He rebooted the server to apply the security patch.",
    "The judge announced the final verdict in a calm voice.",
]

for t in test_texts:
    print(f"romance_score('{t[:40]}...') = {romance_score(t):.3f}")
```

```python
# ============================================
# LLaMA-2-7B: setup and embedding helpers
# ============================================
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

from transformers import AutoModelForCausalLM, AutoTokenizer

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True

LLAMA_NAME = "meta-llama/Llama-2-7b-hf"  # requires HF token + license acceptance

llama_tokenizer = AutoTokenizer.from_pretrained(LLAMA_NAME)
if llama_tokenizer.pad_token is None:
    llama_tokenizer.pad_token = llama_tokenizer.eos_token

llama_model = AutoModelForCausalLM.from_pretrained(
    LLAMA_NAME,
    output_hidden_states=True,      # <-- we want all layer activations
    torch_dtype=torch.bfloat16,     # good for A100
    device_map={"": DEVICE},        # put everything on one GPU
)
llama_model.eval()


def mean_pool(hidden_states, attention_mask=None):
    """
    Mean-pool over sequence dimension.
    hidden_states: [B, T, D]
    attention_mask: [B, T] (1 = real, 0 = pad) or None.
    """
    if attention_mask is None:
        return hidden_states.mean(dim=1)
    mask = attention_mask.unsqueeze(-1).float()
    summed = (hidden_states * mask).sum(dim=1)
    count = mask.sum(dim=1).clamp(min=1.0)
    return summed / count


@torch.no_grad()
def llama_hidden_states(texts):
    """
    Get final hidden states from LLaMA-2-7B.
    texts: list[str]
    Returns:
      hidden: [B, T, D]  final layer token embeddings
      input_ids: [B, T]
      attention_mask: [B, T]
    """
    enc = llama_tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
    ).to(DEVICE)

    outputs = llama_model(**enc)
    # outputs.hidden_states is a tuple of [layer0, layer1, ..., layerN]
    hidden = outputs.hidden_states[-1]  # final layer: [B, T, D]
    return hidden, enc["input_ids"], enc["attention_mask"]
```

### Sentence-level embedding map (LLaMA 7B)

```python
# ============================================
# Sentence-level map with LLaMA-2-7B (PCA)
# ============================================
from sklearn.decomposition import PCA

sentences = [
    "I love you more than anything in this world.",
    "She kissed him softly under the moonlight.",
    "The CPU temperature started rising rapidly.",
    "The defendant was found guilty on all counts.",
    "The weather today is sunny and pleasant.",
    "I hate you and everything you stand for.",
    "He connected to the database and ran a query.",
    "The knight drew his sword and charged.",
]

hidden, input_ids, attn_mask = llama_hidden_states(sentences)  # [B, T, D]
sentence_vecs = mean_pool(hidden, attn_mask)                   # [B, D]
sentence_vecs_cpu = sentence_vecs.cpu().numpy()

pca = PCA(n_components=2)
coords = pca.fit_transform(sentence_vecs_cpu)                   # [B, 2]

plt.figure(figsize=(8, 6))
for (x, y), s in zip(coords, sentences):
    plt.scatter(x, y)
    label = s if len(s) < 40 else s[:37] + "..."
    plt.text(x + 0.01, y + 0.01, label, fontsize=8)
plt.title("LLaMA-2-7B sentence embeddings (PCA 2D)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

### Nearest-neighbor “poking around” in 7B space

```python
# ============================================
# Nearest neighbors in LLaMA-2-7B embedding space
# ============================================
import numpy as np

import torch.nn.functional as F

def nearest_neighbors(idx, k=5):
    """
    Given an index into `sentences`, print its k nearest neighbors
    in LLaMA-2-7B sentence embedding space (cosine similarity).
    """
    target = sentence_vecs[idx:idx+1]  # [1, D]
    sims = F.cosine_similarity(target, sentence_vecs, dim=-1).cpu().numpy()

    order = np.argsort(-sims)  # descending

    print(f"\nQuery [{idx}]: {sentences[idx]}")
    print("Nearest neighbors:")
    for j in order[:k]:
        print(f"  sim={sims[j]:.3f} | [{j}] {sentences[j]}")

# Example: poke around a romantic sentence
nearest_neighbors(0, k=5)
```

### Similarity heatmap (global view)

```python
# ============================================
# Pairwise similarity heatmap (LLaMA-2-7B)
# ============================================
sentence_vecs_t = torch.tensor(sentence_vecs_cpu)
sentence_vecs_norm = F.normalize(sentence_vecs_t, dim=-1)
sim_matrix = (sentence_vecs_norm @ sentence_vecs_norm.T).cpu().numpy()  # [N, N]

plt.figure(figsize=(6, 5))
plt.imshow(sim_matrix, cmap="viridis")
plt.colorbar(label="cosine similarity")
plt.xticks(range(len(sentences)), range(len(sentences)))
plt.yticks(range(len(sentences)), range(len(sentences)))
plt.title("LLaMA-2-7B sentence similarity heatmap")
plt.show()
```

## LLaMA-2-7B: probing innate concepts on random FineWeb excerpts

In this section we:

- Sample random excerpts from the FineWeb dataset.
- Embed each excerpt using the final hidden layer of LLaMA-2-7B.
- Project each excerpt into the same space as the model's token embeddings.
- For each excerpt, find the nearest tokens in embedding space.
  These nearest tokens act as "concept labels" the model itself
  thinks live near that excerpt, without us hand-defining categories.

This lets us compare our own interpretation of the excerpt's ideas/context
with the model's own internal notion of "nearby concepts" in its embedding space.


```python
# ============================================================
# Random FineWeb excerpts + "innate" *readable* concept words
# from LLaMA-2-7B embedding space
# ============================================================

!pip install -q "datasets>=2.20.0"

import random
import re
import torch
import torch.nn.functional as F
from datasets import load_dataset

# Assumes:
#   - DEVICE
#   - llama_tokenizer
#   - llama_model (AutoModelForCausalLM with output_hidden_states=True)
#   - llama_hidden_states(texts)
#   - mean_pool(hidden, mask)
# are already defined.


# -------------------------------
# 1. Load FineWeb (streaming)
# -------------------------------

print("Loading FineWeb (sample-10BT) in streaming mode for innate concept probing...")
fw_stream_for_probe = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

FW_INNATE_SEED = 123
random.seed(FW_INNATE_SEED)


# -------------------------------
# 2. Helper: sample random excerpts
# -------------------------------

def sample_fineweb_excerpts_raw(
    hf_stream,
    num_excerpts: int = 5,
    min_chars: int = 200,
    max_chars: int = 600,
):
    """
    Sample `num_excerpts` random-ish docs from FineWeb and take a random slice
    of each between [min_chars, max_chars] characters.
    """
    excerpts = []
    it = iter(hf_stream)

    while len(excerpts) < num_excerpts:
        try:
            ex = next(it)
        except StopIteration:
            break

        text = ex.get("text", "")
        if not text:
            continue

        text = text.strip()
        if len(text) < min_chars:
            continue

        window_size = random.randint(min_chars, max_chars)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if not snippet:
            continue

        excerpts.append(snippet)

    return excerpts


# -------------------------------
# 3. Build a "readable" vocab subset
# -------------------------------

def is_readable_token(text: str) -> bool:
    """
    Heuristic filter to keep only reasonably readable tokens:
    - After stripping, contains at least one ASCII letter.
    - Not mostly punctuation / digits.
    - Not obviously non-Latin scripts.
    """
    text = text.strip()
    if not text:
        return False

    # LLaMA-style: '▁word' -> 'word'
    text = text.replace("▁", " ").strip()

    # Require at least one Latin letter
    if not re.search(r"[A-Za-z]", text):
        return False

    # Filter out very short strings
    if len(text) < 3:
        return False

    # Exclude obvious non-Latin scripts (Cyrillic, CJK, Greek, etc.)
    if re.search(r"[\u0400-\u04FF\u4E00-\u9FFF\u0370-\u03FF]", text):
        return False

    # Avoid pure punctuation like ")))", ";;;", etc.
    if re.fullmatch(r"[\W_]+", text):
        return False

    return True


with torch.no_grad():
    # Input embeddings (tied with output for LLaMA-style models)
    full_emb = llama_model.get_input_embeddings().weight  # [V, D]
    full_emb = F.normalize(full_emb.float(), dim=-1)
    vocab_size, dim = full_emb.shape

readable_ids = []
readable_embs = []
readable_texts = []

for tid in range(vocab_size):
    # Decode as standalone token for a nicer human-readable form
    decoded = llama_tokenizer.decode([tid], skip_special_tokens=True)
    if not decoded:
        continue

    if is_readable_token(decoded):
        readable_ids.append(tid)
        readable_texts.append(decoded.strip())
        readable_embs.append(full_emb[tid])

if not readable_ids:
    raise RuntimeError("Readable vocab filter was too strict; got 0 tokens.")

readable_embs = torch.stack(readable_embs, dim=0)   # [N_readable, D]
readable_embs = F.normalize(readable_embs, dim=-1)  # keep them normalized
readable_ids = torch.tensor(readable_ids, dtype=torch.long)

print(f"Readable vocab size: {len(readable_ids)}")


# -------------------------------
# 4. Given an excerpt, get nearest *readable* tokens
# -------------------------------

@torch.no_grad()
def innate_concept_words_for_text_readable(
    text: str,
    top_k: int = 20,
):
    """
    - Embed the text with LLaMA-2-7B (final hidden, mean-pooled).
    - Compute cosine similarity against *readable* token embeddings only.
    - Return top_k nearest readable tokens as "innate concept words".
    """
    hidden, _, mask = llama_hidden_states([text])    # [1, T, D]
    vec = mean_pool(hidden, mask)                    # [1, D]
    vec = F.normalize(vec.float(), dim=-1)           # [1, D]

    # cosine sim [1, D] @ [D, N_readable] -> [1, N_readable]
    sims = torch.matmul(vec, readable_embs.T)[0].cpu()  # [N_readable]

    topk_vals, topk_idx = torch.topk(sims, k=min(top_k, readable_embs.shape[0]))

    results = []
    for score, idx in zip(topk_vals.tolist(), topk_idx.tolist()):
        token_text = readable_texts[idx]
        results.append((token_text, score))

    return results


def print_innate_concepts_for_random_excerpts_readable(
    num_excerpts: int = 5,
    min_chars: int = 200,
    max_chars: int = 600,
    top_k_tokens: int = 20,
):
    """
    - Sample random FineWeb excerpts.
    - For each:
        * print the raw text
        * print the nearest *readable* tokens in LLaMA-2-7B embedding space
    """
    excerpts = sample_fineweb_excerpts_raw(
        fw_stream_for_probe,
        num_excerpts=num_excerpts,
        min_chars=min_chars,
        max_chars=max_chars,
    )

    if not excerpts:
        print("No excerpts sampled (dataset exhausted or too strict filters).")
        return

    for i, text in enumerate(excerpts):
        print("=" * 80)
        print(f"Excerpt {i+1}/{len(excerpts)}")
        print("- Raw text (truncated to 800 chars) -")
        print(text[:800])
        if len(text) > 800:
            print("... [truncated] ...")
        print()

        concepts = innate_concept_words_for_text_readable(
            text,
            top_k=top_k_tokens,
        )

        print('Nearest readable tokens in LLaMA-2-7B space ("innate concepts"):')
        for tok, score in concepts:
            print(f"  {tok:25s}  cos_sim = {score: .3f}")
        print()


# -------------------------------
# 5. Run a probe
# -------------------------------

print_innate_concepts_for_random_excerpts_readable(
    num_excerpts=5,      # how many excerpts to inspect
    min_chars=200,       # min excerpt length
    max_chars=600,       # max excerpt length
    top_k_tokens=20,     # how many nearest tokens to display
)
```

```python
# ============================================================
# Nearest-neighbor passages in LLaMA-2-7B embedding space
#   - See what other FineWeb snippets cluster around a snippet
# ============================================================

import torch
import torch.nn.functional as F

# Assumes:
#   - fw_stream_for_probe
#   - sample_fineweb_excerpts_raw(...)
#   - llama_hidden_states(texts)
#   - mean_pool(hidden, mask)
# are already defined.


# -------------------------------
# 1. Embed snippets with LLaMA-2-7B
# -------------------------------

@torch.no_grad()
def embed_snippets_with_llama(
    snippets,
    batch_size: int = 4,
):
    """
    Take a list of raw text snippets, return:
      - embeddings: [N, D] L2-normalized pooled hidden states
    """
    all_vecs = []

    for i in range(0, len(snippets), batch_size):
        batch = snippets[i : i + batch_size]
        hidden, _, mask = llama_hidden_states(batch)   # [B, T, D]
        pooled = mean_pool(hidden, mask)               # [B, D]
        pooled = F.normalize(pooled.float(), dim=-1)   # [B, D]
        all_vecs.append(pooled.cpu())

    embs = torch.cat(all_vecs, dim=0)  # [N, D] on CPU
    return embs


# -------------------------------
# 2. Sample snippets + find neighbors
# -------------------------------

def nearest_neighbor_passages(
    num_excerpts: int = 40,
    min_chars: int = 200,
    max_chars: int = 600,
    k_neighbors: int = 5,
):
    """
    - Sample `num_excerpts` random excerpts from FineWeb.
    - Embed each with LLaMA-2-7B.
    - Compute cosine similarity matrix and show nearest neighbors
      for each probe excerpt.
    """
    snippets = sample_fineweb_excerpts_raw(
        fw_stream_for_probe,
        num_excerpts=num_excerpts,
        min_chars=min_chars,
        max_chars=max_chars,
    )

    if len(snippets) < 2:
        print("Not enough snippets sampled to compute neighbors.")
        return

    print(f"Sampled {len(snippets)} snippets. Embedding with LLaMA-2-7B...")
    embs = embed_snippets_with_llama(snippets)  # [N, D], normalized
    N = embs.shape[0]

    # Cosine similarity matrix: [N, N]
    sims = torch.matmul(embs, embs.T)  # still on CPU
    # Avoid self-matches by setting diagonal to -inf
    sims.fill_diagonal_(-1.0)

    for i in range(N):
        probe = snippets[i]
        row = sims[i]  # [N]

        # top k neighbors for this probe
        k = min(k_neighbors, N - 1)
        vals, idxs = torch.topk(row, k=k)

        print("=" * 80)
        print(f"PROBE {i+1}/{N}")
        print("- Raw text (truncated to 600 chars) -")
        print(probe[:600])
        if len(probe) > 600:
            print("... [truncated] ...")
        print()

        print(f"Nearest {k} neighbor passages in LLaMA-2-7B space:")
        for rank, (score, j) in enumerate(zip(vals.tolist(), idxs.tolist()), start=1):
            neighbor = snippets[j]
            print("-" * 80)
            print(f"Neighbor {rank} (index {j}, cos_sim = {score: .3f})")
            print(neighbor[:600])
            if len(neighbor) > 600:
                print("... [truncated] ...")
        print()


# -------------------------------
# 3. Run the passage-level probe
# -------------------------------

nearest_neighbor_passages(
    num_excerpts=30,   # number of snippets to embed + cluster
    min_chars=200,
    max_chars=600,
    k_neighbors=4,
)
```

```python
# ============================================================
# Concept axes in LLaMA-2-7B space + random FineWeb excerpts
#   - Build interpretable "concept vectors" from prototype
#     sentences (romantic, hostile, technical, casual, etc.)
#   - Project random FineWeb snippets onto these axes
# ============================================================

!pip install -q "datasets>=2.20.0"

import random
import torch
import torch.nn.functional as F
from datasets import load_dataset

# Assumes already defined:
#   DEVICE
#   llama_tokenizer
#   llama_model  (AutoModelForCausalLM with output_hidden_states=True)
#   llama_hidden_states(texts) -> (hidden, last_hidden, attention_mask)
#   mean_pool(hidden, mask)    -> [B, D] pooled vectors


# -------------------------------
# 1. Load FineWeb (streaming)
# -------------------------------

print("Loading FineWeb (sample-10BT) in streaming mode for concept probing...")
fw_stream_concepts = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

FW_CONCEPT_SEED = 777
random.seed(FW_CONCEPT_SEED)


def sample_fineweb_excerpts_raw(
    hf_stream,
    num_excerpts: int = 5,
    min_chars: int = 200,
    max_chars: int = 600,
):
    """
    Sample `num_excerpts` random-ish docs from FineWeb and take a random slice
    of each between [min_chars, max_chars] characters.
    """
    excerpts = []
    it = iter(hf_stream)

    while len(excerpts) < num_excerpts:
        try:
            ex = next(it)
        except StopIteration:
            break

        text = ex.get("text", "")
        if not text:
            continue

        text = text.strip()
        if len(text) < min_chars:
            continue

        window_size = random.randint(min_chars, max_chars)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if not snippet:
            continue

        excerpts.append(snippet)

    return excerpts


# -------------------------------
# 2. Define concept prototypes
# -------------------------------
# You can tweak these sentences to better match your intuition.
# The model will learn concept directions from these examples.

concept_prototypes = {
    "romantic": [
        "He looked into her eyes and felt a deep, gentle love.",
        "They walked together under the stars, holding hands.",
        "The letter was full of affection, tenderness, and devotion.",
    ],
    "hostile": [
        "His tone was sharp and aggressive, full of resentment.",
        "She snapped back with clear anger and irritation.",
        "The message sounded threatening and confrontational.",
    ],
    "technical": [
        "This paper describes the architecture of a neural network model.",
        "We analyze the algorithmic complexity of the proposed method.",
        "The system uses a distributed database and a replicated log.",
    ],
    "casual": [
        "Hey, what's up, just checking in on you.",
        "We grabbed some coffee and chatted about random stuff.",
        "It felt like a relaxed, everyday conversation between friends.",
    ],
    "medical": [
        "The patient presented with fever, fatigue, and joint pain.",
        "The treatment involves a combination of medication and therapy.",
        "Clinical trials showed a significant improvement in symptoms.",
    ],
    "spiritual": [
        "They gathered in silence to meditate and reflect.",
        "He spoke about faith, inner peace, and the meaning of life.",
        "The ceremony had a calm, sacred atmosphere.",
    ],
    "political": [
        "The government announced a new policy on climate change.",
        "The election campaign focused on economic reform and healthcare.",
        "Debates over public spending and taxation dominated the news.",
    ],
}


@torch.no_grad()
def build_concept_vectors(concept_prototypes: dict) -> dict:
    """
    Given a dict: concept -> list of prototype sentences,
    return a dict: concept -> normalized vector [D].
    """
    concept_vectors = {}

    for concept, examples in concept_prototypes.items():
        hidden, _, mask = llama_hidden_states(examples)   # [B, T, D]
        pooled = mean_pool(hidden, mask)                  # [B, D]
        pooled = F.normalize(pooled.float(), dim=-1)      # [B, D]
        # Average across examples for this concept
        v = pooled.mean(dim=0)                            # [D]
        v = F.normalize(v, dim=-1)                        # [D]
        concept_vectors[concept] = v.cpu()

    return concept_vectors


print("Building concept vectors from prototypes...")
concept_vectors = build_concept_vectors(concept_prototypes)
concept_names = list(concept_vectors.keys())
print("Concepts:", concept_names)


# -------------------------------
# 3. Score text along concept axes
# -------------------------------

@torch.no_grad()
def score_text_concepts(text: str, concept_vectors: dict, top_k: int = None):
    """
    - Embed `text` with LLaMA (final hidden, mean-pooled).
    - Compute cosine similarity with each concept vector.
    - Return sorted list of (concept, score).
    """
    hidden, _, mask = llama_hidden_states([text])   # [1, T, D]
    pooled = mean_pool(hidden, mask)               # [1, D]
    pooled = F.normalize(pooled.float(), dim=-1)[0].cpu()  # [D]

    scores = []
    for name, vec in concept_vectors.items():
        # vec: [D]; pooled: [D]
        score = torch.dot(pooled, vec).item()
        scores.append((name, score))

    # Sort descending by similarity
    scores.sort(key=lambda x: x[1], reverse=True)

    if top_k is not None:
        scores = scores[:top_k]

    return scores


def print_concept_scores_for_random_excerpts(
    num_excerpts: int = 5,
    min_chars: int = 200,
    max_chars: int = 600,
    top_k_concepts: int = 5,
):
    """
    - Sample random FineWeb snippets.
    - For each snippet:
        * print the raw text
        * print its concept scores (cosine similarity)
    """
    snippets = sample_fineweb_excerpts_raw(
        fw_stream_concepts,
        num_excerpts=num_excerpts,
        min_chars=min_chars,
        max_chars=max_chars,
    )

    if not snippets:
        print("No snippets sampled.")
        return

    for i, text in enumerate(snippets):
        print("=" * 80)
        print(f"EXCERPT {i+1}/{len(snippets)}")
        print("- Raw text (truncated to 800 chars) -")
        print(text[:800])
        if len(text) > 800:
            print("... [truncated] ...")
        print()

        scores = score_text_concepts(
            text,
            concept_vectors=concept_vectors,
            top_k=top_k_concepts,
        )

        print("Concept scores (higher = closer in hidden-state space):")
        for name, s in scores:
            print(f"  {name:10s}  cos_sim = {s: .3f}")
        print()


# -------------------------------
# 4. Run the probe
# -------------------------------

print_concept_scores_for_random_excerpts(
    num_excerpts=5,
    min_chars=200,
    max_chars=600,
    top_k_concepts=5,
)
```

# Visualizing how my multi heads intrepret the hidden state

```python
# ============================================================
# Visualizing how FutureGPTMultiToken heads "interpret" hidden states
#   - Use the same trunk + head structure as training
#   - For one random FineWeb snippet:
#       * get trunk states h_trunk
#       * get head1 states h_head1 (t+1 head)
#       * get head2 states h_head2 (t+2 head)
#   - Joint PCA -> 2D and scatter plot trunk/head1/head2
#   - Print tokens so you can inspect clusters by eye
# ============================================================

!pip install -q "datasets>=2.20.0" "transformers>=4.44.0" "scikit-learn>=1.4.0"

import os
import random
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

from datasets import load_dataset
from transformers import AutoTokenizer
from sklearn.decomposition import PCA

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
dtype = torch.bfloat16

# -------------------------------
# 1. Tokenizer + model loading
# -------------------------------
# If you already have `tokenizer` and `model` in memory from training,
# you can skip this block or comment it out and just reuse them.

TOKENIZER_NAME = "gpt2"
BLOCK_SIZE = 128  # shorter for analysis/visualization

tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.model_max_length = BLOCK_SIZE + 2

vocab_size = tokenizer.vocab_size

# ---- Reload model from checkpoint (optional but recommended) ----
# Adjust these paths/names to match your training setup.
from google.colab import drive
drive.mount("/content/drive")

CKPT_DIR = "/content/drive/My Drive/School/CS7391 - LLM Research Ind. Study/Check Points"
CKPT_NAME = "futuregpt_meta_style_small_fromscratch_final.pt"  # or a specific step
CKPT_PATH = os.path.join(CKPT_DIR, CKPT_NAME)
print(f"Loading checkpoint from: {CKPT_PATH}")

# Same args as training
args = ModelArgs(
    dim=1024,
    n_layers=12,
    n_heads=16,
    n_future_tokens=2,
    vocab_size=vocab_size,
    max_seq_len=BLOCK_SIZE,
)

model = FutureGPTMultiToken(args).to(DEVICE)
state = torch.load(CKPT_PATH, map_location=DEVICE)["model_state"]
model.load_state_dict(state)
model.to(dtype)
model.eval()

print("Model loaded for analysis:")
print(model)


# -------------------------------
# 2. Trunk + head helpers
#    (same logic as your training cell)
# -------------------------------

def forward_trunk(model: FutureGPTMultiToken, input_ids: torch.Tensor):
    """
    Run only the trunk part of the model, matching FutureGPTMultiToken.forward
    but stopping before prediction heads.

    Returns:
      h_trunk   : [B, T, dim] shared trunk features
      freqs_cis : [T, head_dim/2] RoPE frequencies used
      mask      : [1, 1, T, T] causal mask
    """
    B, T = input_ids.shape
    device = input_ids.device

    assert T <= model.params.max_seq_len, (
        f"Sequence length {T} > max_seq_len {model.params.max_seq_len}"
    )

    # Token embedding
    h = model.tok_embeddings(input_ids)  # [B, T, dim]

    # RoPE freqs
    freqs_cis = model.freqs_cis[:T].to(device=h.device)

    # Causal mask
    mask = model._build_causal_mask(T, device=h.device)

    # Trunk: all but last layer
    for layer in model.layers[:-1]:
        h = layer(h, freqs_cis=freqs_cis, mask=mask)

    h_trunk = h
    return h_trunk, freqs_cis, mask


def forward_head_from_trunk(
    model: FutureGPTMultiToken,
    h_trunk: torch.Tensor,
    freqs_cis: torch.Tensor,
    mask: torch.Tensor,
    head_index: int,
):
    """
    Run a single prediction head on shared trunk features.

    head_index:
      0 -> t+1 head (reuses model.layers[-1])
      1 -> t+2 head (model.extra_heads[0])

    Returns:
      h_head   : [B, T, dim]   (pre-logit head representation)
      logits_k : [B, T, V]     (optional, not needed for PCA)
    """
    if head_index == 0:
        head_layer = model.layers[-1]
    else:
        extra_idx = head_index - 1
        head_layer = model.extra_heads[extra_idx]

    h_head = head_layer(h_trunk, freqs_cis=freqs_cis, mask=mask)  # [B, T, dim]
    h_norm = model.norm(h_head)                                   # [B, T, dim]
    logits_k = model.output(h_norm).float()                       # [B, T, V]
    return h_head, logits_k


# -------------------------------
# 3. Sample a random FineWeb snippet
# -------------------------------

print("Loading FineWeb (sample-10BT) in streaming mode for visualization...")
fw_stream_vis = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

VIS_SEED = 2025
random.seed(VIS_SEED)

def random_fineweb_snippet(
    hf_stream,
    min_chars: int = 200,
    max_chars: int = 600,
):
    """
    Grab one random-ish document and take a random slice.
    """
    it = iter(hf_stream)
    while True:
        ex = next(it)
        text = ex.get("text", "")
        if not text:
            continue
        text = text.strip()
        if len(text) < min_chars:
            continue

        window_size = random.randint(min_chars, max_chars)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if snippet:
            return snippet

snippet = random_fineweb_snippet(fw_stream_vis, min_chars=200, max_chars=400)
print("=" * 80)
print("Random FineWeb snippet used for trunk/head visualization:")
print(snippet[:800])
print("=" * 80)


# -------------------------------
# 4. Run model to get trunk + head states
# -------------------------------

# Tokenize and truncate to BLOCK_SIZE
enc = tokenizer(
    snippet,
    add_special_tokens=False,
    return_tensors="pt",
)
input_ids = enc["input_ids"][:, :BLOCK_SIZE].to(DEVICE)  # [1, T]
T = input_ids.shape[1]

# Keep a copy of token strings for inspection
tokens = tokenizer.convert_ids_to_tokens(input_ids[0].cpu().tolist())

with torch.no_grad():
    h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)             # [1, T, D]
    h_head1, _ = forward_head_from_trunk(model, h_trunk, freqs_cis, mask, 0)  # [1, T, D]
    h_head2, _ = forward_head_from_trunk(model, h_trunk, freqs_cis, mask, 1)  # [1, T, D]

# Move to CPU + float32
h_trunk_np = h_trunk[0].float().cpu().numpy()   # [T, D]
h_head1_np = h_head1[0].float().cpu().numpy()   # [T, D]
h_head2_np = h_head2[0].float().cpu().numpy()   # [T, D]

D = h_trunk_np.shape[1]
print(f"Sequence length T = {T}, hidden dim D = {D}")


# -------------------------------
# 5. Joint PCA over trunk/head states
# -------------------------------

# Stack all states: [3T, D]
X_all = np.concatenate([h_trunk_np, h_head1_np, h_head2_np], axis=0)  # [3T, D]

# PCA -> 2D
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_all)  # [3T, 2]

# Split back out
X_trunk_2d = X_2d[:T]
X_head1_2d = X_2d[T:2*T]
X_head2_2d = X_2d[2*T:3*T]

print("Explained variance ratio (PC1, PC2):", pca.explained_variance_ratio_)


# -------------------------------
# 6. Plot trunk vs head1 vs head2
# -------------------------------

plt.figure(figsize=(10, 8))

plt.scatter(X_trunk_2d[:, 0], X_trunk_2d[:, 1], alpha=0.5, label="Trunk", s=20)
plt.scatter(X_head1_2d[:, 0], X_head1_2d[:, 1], alpha=0.5, label="Head 1 (t+1)", s=20)
plt.scatter(X_head2_2d[:, 0], X_head2_2d[:, 1], alpha=0.5, label="Head 2 (t+2)", s=20)

plt.title("FutureGPTMultiToken: PCA of token representations\nTrunk vs Head1 vs Head2")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True)
plt.show()


# -------------------------------
# 7. Print tokens + index for manual inspection
# -------------------------------

print("Token indices and strings for this snippet:")
for i, tok in enumerate(tokens):
    print(f"{i:3d}: {tok}")
```

## Delving deeper into the multi dimensional space of the multi token model

### Task 1 — Trunk Future-Token Category Centroid PCA

This cell samples a set of snippets from FineWeb, runs them through the **trunk** of `FutureGPTMultiToken`, and collects final trunk hidden states \(h_t\) for many token positions.

For each position \(t\), we:
- Take the trunk state \(h_t\).
- Look at the **actual tokens** at \(t+1\) and \(t+2\).
- Map those future tokens into coarse categories (`"word"`, `"punct"`, `"number"`, `"special"`).

We then:
- Compute **category centroids** for trunk states labeled by the type of token at \(t+1\) (e.g. `word_t+1`, `punct_t+1`).
- Do the same for \(t+2\) (e.g. `word_t+2`, `punct_t+2`).
- Stack all centroids and run **PCA (2D)** on them.
- Plot a scatter where each point is a centroid labeled by category and offset (`_t+1` vs `_t+2`).

If points like `word_t+1`, `word_t+2`, `punct_t+1`, `punct_t+2` are clearly separated in this 2D view, that’s visual evidence that the **same trunk space** linearly encodes information about *what kind of token* is coming at both t+1 and t+2.


```python
# ============================================================
# Task 1 — Trunk Future-Token Category Centroid PCA
#   - Sample snippets from FineWeb
#   - Run them through the trunk of FutureGPTMultiToken
#   - Label each trunk state h_t by the category of x_{t+1} and x_{t+2}
#   - Compute centroids per category/offset
#   - PCA on centroids and 2D scatter plot
# ============================================================

!pip install -q "datasets>=2.20.0" "transformers>=4.44.0" "scikit-learn>=1.4.0"

import random
import string
import numpy as np
import torch
import matplotlib.pyplot as plt

from collections import defaultdict
from datasets import load_dataset
from sklearn.decomposition import PCA

# ------------------------------------------------------------
# Assumptions:
#   - `model`, `tokenizer`, `forward_trunk` are already defined
#     as in your visualization cell.
#   - `model` is a FutureGPTMultiToken on DEVICE in eval mode.
#   - TOKENIZER_NAME, BLOCK_SIZE, DEVICE, dtype are set.
# If not, you can re-run your previous model-loading cell first.
# ------------------------------------------------------------

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# For this analysis, a slightly smaller block size is fine
BLOCK_SIZE_ANALYSIS = 128

# Make sure tokenizer max length isn't a problem
tokenizer.model_max_length = BLOCK_SIZE_ANALYSIS + 2

# -------------------------------
# 1. Helper: token → coarse category
# -------------------------------

def gpt2_token_category(tok_str: str) -> str:
    """
    Map a GPT-2 BPE token string to a coarse category:
      - 'number' : all digits
      - 'punct'  : all chars are punctuation
      - 'word'   : contains any alphabetic character
      - 'special': everything else (BPE artifacts, Unicode, etc.)
    """
    # Strip leading space marker 'Ġ' used in some BPEs (GPT-2 uses spaces in tokens)
    # but keep the rest.
    base = tok_str.lstrip("Ġ")

    # Empty after stripping -> treat as special/whitespace-ish
    if base == "":
        return "special"

    # Number: all digits
    if base.isdigit():
        return "number"

    # Punctuation: all chars are in string.punctuation
    if all(ch in string.punctuation for ch in base):
        return "punct"

    # Word: any alphabetic char
    if any(ch.isalpha() for ch in base):
        return "word"

    # Fallback
    return "special"


# -------------------------------
# 2. Sample trunk states + labels
# -------------------------------

print("Loading FineWeb (sample-10BT) in streaming mode for trunk centroid analysis...")
fw_stream = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

ANALYSIS_SEED = 2025
random.seed(ANALYSIS_SEED)

# How many examples / positions to collect
MAX_SNIPPETS = 80          # max number of documents to sample
MAX_TRUNK_POINTS = 20000   # max number of h_t vectors to keep

trunk_vectors = []   # list of np.array [D]
labels_next = []      # category of x_{t+1}
labels_next2 = []     # category of x_{t+2}

def sample_snippets_and_collect_trunk(
    hf_stream,
    max_snippets: int,
    max_points: int,
):
    """
    Iterate over the streaming FineWeb dataset, sample up to `max_snippets`
    snippets, run them through the trunk, and collect trunk states h_t
    + future-token categories up to `max_points` total positions.
    """
    global trunk_vectors, labels_next, labels_next2

    snippets_seen = 0
    total_points = 0

    it = iter(hf_stream)
    while snippets_seen < max_snippets and total_points < max_points:
        try:
            ex = next(it)
        except StopIteration:
            break

        text = ex.get("text", "")
        if not text:
            continue
        text = text.strip()
        if len(text) < 50:
            # Skip very short docs
            continue

        # Random window from this document
        window_size = random.randint(150, 400)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if not snippet:
            continue

        # Tokenize and truncate to BLOCK_SIZE_ANALYSIS
        enc = tokenizer(
            snippet,
            add_special_tokens=False,
            return_tensors="pt",
        )
        input_ids = enc["input_ids"][:, :BLOCK_SIZE_ANALYSIS].to(DEVICE)  # [1, T]
        B, T = input_ids.shape
        if T < 4:
            # Need at least a few tokens for t+2 lookahead
            continue

        # Run through the trunk
        with torch.no_grad():
            h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)
            # h_trunk: [1, T, D]
        h_trunk_np = h_trunk[0].float().cpu().numpy()  # [T, D]

        # Convert token ids to GPT-2 token strings
        ids = input_ids[0].cpu().tolist()
        toks = tokenizer.convert_ids_to_tokens(ids)

        # For each position t where t+2 is valid
        for t in range(0, T - 2):
            if total_points >= max_points:
                break

            h = h_trunk_np[t]  # [D]
            tok_t1 = toks[t + 1]
            tok_t2 = toks[t + 2]

            cat_t1 = gpt2_token_category(tok_t1)
            cat_t2 = gpt2_token_category(tok_t2)

            trunk_vectors.append(h)
            labels_next.append(cat_t1)
            labels_next2.append(cat_t2)

            total_points += 1

        snippets_seen += 1

    print(f"Collected {total_points} trunk states from {snippets_seen} snippets.")


sample_snippets_and_collect_trunk(
    fw_stream,
    max_snippets=MAX_SNIPPETS,
    max_points=MAX_TRUNK_POINTS,
)

trunk_vectors = np.stack(trunk_vectors)  # [N, D]
print("Trunk vectors shape:", trunk_vectors.shape)


# -------------------------------
# 3. Compute centroids per category/offset
# -------------------------------

def centroids_from_labels(vectors: np.ndarray, labels: list, min_count: int = 20):
    """
    Given a matrix of vectors [N, D] and a parallel list of labels (length N),
    compute per-label centroid = mean of vectors for that label.

    Only keep labels that appear at least `min_count` times.
    """
    buckets = defaultdict(list)
    for v, lab in zip(vectors, labels):
        buckets[lab].append(v)

    centroids = {}
    counts = {}
    for lab, vs in buckets.items():
        vs = np.stack(vs)
        if vs.shape[0] >= min_count:
            centroids[lab] = vs.mean(axis=0)
            counts[lab] = vs.shape[0]

    return centroids, counts


centroids_t1, counts_t1 = centroids_from_labels(trunk_vectors, labels_next, min_count=30)
centroids_t2, counts_t2 = centroids_from_labels(trunk_vectors, labels_next2, min_count=30)

print("\nCategory counts for t+1:")
for lab, c in counts_t1.items():
    print(f"  {lab:8s}: {c}")

print("\nCategory counts for t+2:")
for lab, c in counts_t2.items():
    print(f"  {lab:8s}: {c}")

# If some categories are missing for an offset (e.g., no 'number_t+2'),
# that's fine; they just won't appear in the plot.

# -------------------------------
# 4. Build centroid matrix & labels
# -------------------------------

centroid_labels = []
centroid_matrix = []

# First, t+1 centroids
for lab, vec in centroids_t1.items():
    centroid_labels.append(f"{lab}_t+1")
    centroid_matrix.append(vec)

# Then, t+2 centroids
for lab, vec in centroids_t2.items():
    centroid_labels.append(f"{lab}_t+2")
    centroid_matrix.append(vec)

centroid_matrix = np.stack(centroid_matrix)  # [C, D]
print("\nCentroid matrix shape:", centroid_matrix.shape)

# -------------------------------
# 5. PCA on centroids (2D)
# -------------------------------

pca = PCA(n_components=2)
centroids_2d = pca.fit_transform(centroid_matrix)  # [C, 2]

print("Explained variance ratio (PC1, PC2):", pca.explained_variance_ratio_)

# -------------------------------
# 6. Plot centroids with labels
# -------------------------------

plt.figure(figsize=(8, 6))

for (x, y), lab in zip(centroids_2d, centroid_labels):
    # Color by offset: t+1 vs t+2
    if lab.endswith("t+1"):
        color = "orange"
        marker = "o"
    else:
        color = "green"
        marker = "s"

    plt.scatter(x, y, c=color, marker=marker, s=80)

    # Slight offset for text to avoid covering the point
    plt.text(x + 0.02, y + 0.02, lab, fontsize=9)

plt.title("Trunk space centroids by future-token category (t+1 vs t+2)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.show()
```

ran thousands of samples through model. plotted average centroids for token embeddings in hidden state based on when gold was word/punct/or number  

hidden state = the final vector produced by the shared trunk before any prediction head is applied — the core internal representation the model uses to support both t+1 and t+2 predictions.

h_trunk : [B, T, D]  

where:
- B = batch size

- T = sequence length (token position of input string)

- D = model dimension (1024 for my model)

So for each token position t, we get a 1024-dimensional vector that represents:

- the token itself

- its context

- relationships to other tokens

- all information the heads will need to make predictions

### Interpreting the Trunk Centroid PCA (Simple Explanation)

This plot shows how the model’s **shared trunk** represents tokens based on what type of symbol comes next (t+1) or after that (t+2).  
We grouped thousands of hidden states by the category of their future token – **word**, **punctuation**, or **number** – and computed the average representation (a “centroid”) for each group.  

Plotting these centroids with PCA gives six points (three categories × two offsets).  
The fact that these points are **clearly separated** means:

- The trunk is already encoding **useful information about what kind of token is coming next**, even before the prediction heads operate.  
- The model represents **t+1 and t+2 differently**, showing that the trunk learns features relevant to both immediate and slightly more distant future predictions.

In short, the trunk hidden states contain structured information about **future token types**, and this structure is visible even in a simple 2-D projection.


### Trunk Hidden States: Full Cloud View by Future-Token Category

This plot shows **every trunk hidden state** we collected (one point per token position across all snippets), projected into 2D with PCA.

We make two panels:

- **Left:** each point is colored by the **category of the gold next token (t+1)**  
  (`word`, `punct`, `number`, `special`).

- **Right:** each point is colored by the **category of the gold token at t+2**.

Each dot is a single hidden state vector \(h_t\) (after the trunk).  
The color tells you **what kind of token the model will see in the next 1–2 steps**.

If the colored clouds for different categories occupy different regions, it means the trunk is organizing its internal representations in a way that **already reflects future token type**, even before the prediction heads are applied.


```python
# ============================================================
# Task 1B — Full Trunk State PCA Cloud by Future-Token Category
#   - Use the trunk_vectors, labels_next, labels_next2 from Task 1
#   - Fit PCA on all trunk vectors
#   - Plot each hidden state as a point in 2D
#   - Color by category of x_{t+1} (left) and x_{t+2} (right)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# ------------------------------------------------------------------
# Sanity check: trunk_vectors, labels_next, labels_next2 should
# already be populated from Task 1. If not, raise an error.
# ------------------------------------------------------------------
if "trunk_vectors" not in globals() or "labels_next" not in globals() or "labels_next2" not in globals():
    raise RuntimeError(
        "trunk_vectors / labels_next / labels_next2 not found. "
        "Please run the Task 1 cell first to collect trunk states."
    )

# Ensure numpy array
trunk_vectors_np = np.asarray(trunk_vectors)  # shape [N, D]
N, D = trunk_vectors_np.shape
print(f"Using {N} trunk hidden states of dimension {D} for PCA cloud plot.")

# ------------------------------------------------------------------
# 1. Fit PCA on all trunk vectors
# ------------------------------------------------------------------
pca_cloud = PCA(n_components=2)
trunk_2d = pca_cloud.fit_transform(trunk_vectors_np)  # [N, 2]

print("Explained variance ratio (PC1, PC2):", pca_cloud.explained_variance_ratio_)

# ------------------------------------------------------------------
# 2. Define color mapping for categories
# ------------------------------------------------------------------
categories = ["word", "punct", "number", "special"]
color_map = {
    "word":   "#1f77b4",  # blue
    "punct":  "#ff7f0e",  # orange
    "number": "#2ca02c",  # green
    "special":"#d62728",  # red
}

def reorder_labels(labels):
    """
    Reorder a flat list of labels into per-category index lists.
    Returns: dict {category: list_of_indices}
    """
    buckets = {cat: [] for cat in categories}
    for i, lab in enumerate(labels):
        if lab in buckets:
            buckets[lab].append(i)
        else:
            # Unknown label type (shouldn't really happen), treat as special
            buckets["special"].append(i)
    return buckets

# Group indices by category for t+1 and t+2 labelings
idx_by_cat_t1 = reorder_labels(labels_next)
idx_by_cat_t2 = reorder_labels(labels_next2)

# ------------------------------------------------------------------
# 3. Plot two panels: color by x_{t+1} category and x_{t+2} category
# ------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharex=True, sharey=True)

ax1, ax2 = axes

# ---- Left: color by t+1 category ----
for cat in categories:
    idxs = idx_by_cat_t1[cat]
    if not idxs:
        continue
    pts = trunk_2d[idxs]
    ax1.scatter(
        pts[:, 0], pts[:, 1],
        s=8,
        alpha=0.4,
        label=cat,
        c=color_map[cat]
    )

ax1.set_title("Trunk hidden states colored by gold next-token type (t+1)")
ax1.set_xlabel("PC1")
ax1.set_ylabel("PC2")
ax1.grid(True, linewidth=0.5, alpha=0.4)
ax1.axhline(0, color="gray", linewidth=0.5, alpha=0.5)
ax1.axvline(0, color="gray", linewidth=0.5, alpha=0.5)
ax1.legend(title="t+1 category", loc="best")

# ---- Right: color by t+2 category ----
for cat in categories:
    idxs = idx_by_cat_t2[cat]
    if not idxs:
        continue
    pts = trunk_2d[idxs]
    ax2.scatter(
        pts[:, 0], pts[:, 1],
        s=8,
        alpha=0.4,
        label=cat,
        c=color_map[cat]
    )

ax2.set_title("Trunk hidden states colored by gold next-token type (t+2)")
ax2.set_xlabel("PC1")
ax2.grid(True, linewidth=0.5, alpha=0.4)
ax2.axhline(0, color="gray", linewidth=0.5, alpha=0.5)
ax2.axvline(0, color="gray", linewidth=0.5, alpha=0.5)
ax2.legend(title="t+2 category", loc="best")

plt.suptitle("PCA of trunk hidden states: each point = one h_t, color = future token category")
plt.tight_layout()
plt.show()
```

Even though the individual trunk hidden states overlap heavily in 2D, they still show consistent differences based on the type of token coming next or after next.
These differences become clear when averaging into centroids, demonstrating that the trunk learns to encode future-token category information, even though that information is only a small part of what each hidden state contains.

### Task 2 — Head-Space Category Centroid PCA (Head 1 vs Head 2)

This analysis looks at **how the two prediction heads reshape the shared trunk space.**

We:
1. Sample snippets from FineWeb and run them through the model.
2. For each snippet, we compute:
   - `h_trunk` from the shared trunk.
   - `h_head1` from the **t+1 head** and `h_head2` from the **t+2 head** using `forward_head_from_trunk`.
3. For every position `t`:
   - We label `h_head1[t]` by the **gold category of the next token (t+1)**.
   - We label `h_head2[t]` by the **gold category of the token at t+2**.
4. We compute **centroids** in each head’s space:
   - For head1: average `h_head1` where the next token is a `word`, `punct`, `number`, etc.
   - For head2: average `h_head2` where the token at t+2 is in the same categories.
5. We stack all centroids and run **joint PCA** so we can plot them together:
   - Color by head (Head 1 vs Head 2).
   - Text labels include both category and offset (e.g., `word_h1_t+1`, `punct_h2_t+2`).

If head1 and head2 centroids land in different regions or show different separations between categories, that tells us the two heads are **interpreting the same trunk features differently** to serve the t+1 vs t+2 prediction tasks.


```python
# ============================================================
# Task 2 — Head-Space Category Centroid PCA (Head 1 vs Head 2)
#   - Sample snippets from FineWeb
#   - Get trunk states and head1/head2 states
#   - Label head1 states by gold x_{t+1} type
#   - Label head2 states by gold x_{t+2} type
#   - Compute centroids in each head space
#   - Joint PCA and scatter plot to compare how heads reshape trunk
# ============================================================

!pip install -q "datasets>=2.20.0" "transformers>=4.44.0" "scikit-learn>=1.4.0"

import random
import string
import numpy as np
import torch
import matplotlib.pyplot as plt

from collections import defaultdict
from datasets import load_dataset
from sklearn.decomposition import PCA

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ------------------------------------------------------------
# Assumptions:
#   - `model`, `tokenizer`, `forward_trunk`, and
#     `forward_head_from_trunk` are already defined (from your
#     previous visualization cell).
#   - model is on DEVICE and in eval mode.
# ------------------------------------------------------------

# Use a smaller block size for analysis
BLOCK_SIZE_ANALYSIS = 128
tokenizer.model_max_length = BLOCK_SIZE_ANALYSIS + 2

# -------------------------------
# 1. Helper: token → coarse category
#    (redefine here for safety; matches Task 1)
# -------------------------------

def gpt2_token_category(tok_str: str) -> str:
    """
    Map a GPT-2 BPE token string to a coarse category:
      - 'number' : all digits
      - 'punct'  : all chars are punctuation
      - 'word'   : contains any alphabetic character
      - 'special': everything else (BPE artifacts, Unicode, etc.)
    """
    base = tok_str.lstrip("Ġ")
    if base == "":
        return "special"
    if base.isdigit():
        return "number"
    if all(ch in string.punctuation for ch in base):
        return "punct"
    if any(ch.isalpha() for ch in base):
        return "word"
    return "special"


# -------------------------------
# 2. Sample head states + labels
# -------------------------------

print("Loading FineWeb (sample-10BT) in streaming mode for head-space analysis...")
fw_stream_heads = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

HEAD_ANALYSIS_SEED = 2026
random.seed(HEAD_ANALYSIS_SEED)

MAX_SNIPPETS_HEAD = 80        # number of docs to sample
MAX_POINTS_HEAD   = 20000     # max positions to collect

# Buckets for head1 and head2:
#   head1 uses categories of x_{t+1}
#   head2 uses categories of x_{t+2}
head1_buckets = defaultdict(list)  # cat -> [vectors]
head2_buckets = defaultdict(list)  # cat -> [vectors]

def sample_and_collect_head_states(
    hf_stream,
    max_snippets: int,
    max_points: int,
):
    """
    Iterate over FineWeb streaming data, sample up to `max_snippets`
    snippets, run them through trunk + heads, and fill head1/head2
    buckets keyed by future token category.

    head1 is labeled by category(x_{t+1});
    head2 is labeled by category(x_{t+2}).
    """
    snippets_seen = 0
    total_points = 0
    it = iter(hf_stream)

    while snippets_seen < max_snippets and total_points < max_points:
        try:
            ex = next(it)
        except StopIteration:
            break

        text = ex.get("text", "")
        if not text:
            continue
        text = text.strip()
        if len(text) < 50:
            continue

        # Random window from this document
        window_size = random.randint(150, 400)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if not snippet:
            continue

        # Tokenize and truncate
        enc = tokenizer(
            snippet,
            add_special_tokens=False,
            return_tensors="pt",
        )
        input_ids = enc["input_ids"][:, :BLOCK_SIZE_ANALYSIS].to(DEVICE)  # [1, T]
        B, T = input_ids.shape
        if T < 4:
            continue

        # Get trunk states
        with torch.no_grad():
            h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)
            # h_trunk: [1, T, D]

            # Run heads from trunk
            h_head1, _ = forward_head_from_trunk(
                model, h_trunk, freqs_cis, mask, head_index=0
            )  # [1, T, D]
            h_head2, _ = forward_head_from_trunk(
                model, h_trunk, freqs_cis, mask, head_index=1
            )  # [1, T, D]

        # Move to CPU numpy
        h_head1_np = h_head1[0].float().cpu().numpy()  # [T, D]
        h_head2_np = h_head2[0].float().cpu().numpy()  # [T, D]

        ids = input_ids[0].cpu().tolist()
        toks = tokenizer.convert_ids_to_tokens(ids)

        for t in range(0, T - 2):
            if total_points >= max_points:
                break

            tok_t1 = toks[t + 1]
            tok_t2 = toks[t + 2]

            cat_t1 = gpt2_token_category(tok_t1)
            cat_t2 = gpt2_token_category(tok_t2)

            # Collect head1 state labeled by t+1 category
            head1_buckets[cat_t1].append(h_head1_np[t])

            # Collect head2 state labeled by t+2 category
            head2_buckets[cat_t2].append(h_head2_np[t])

            total_points += 1

        snippets_seen += 1

    print(f"Collected {total_points} positions from {snippets_seen} snippets.")


sample_and_collect_head_states(
    fw_stream_heads,
    max_snippets=MAX_SNIPPETS_HEAD,
    max_points=MAX_POINTS_HEAD,
)

# -------------------------------
# 3. Compute centroids per head
# -------------------------------

def centroids_from_buckets(buckets: dict, min_count: int = 30):
    """
    Given a dict {label: [vectors]}, compute a centroid per label.
    Only keep labels that have at least `min_count` vectors.
    """
    centroids = {}
    counts = {}
    for lab, vs in buckets.items():
        if len(vs) >= min_count:
            arr = np.stack(vs)  # [N, D]
            centroids[lab] = arr.mean(axis=0)
            counts[lab] = arr.shape[0]
    return centroids, counts

centroids_h1, counts_h1 = centroids_from_buckets(head1_buckets, min_count=30)
centroids_h2, counts_h2 = centroids_from_buckets(head2_buckets, min_count=30)

print("\nHead 1 (t+1) category counts:")
for lab, c in counts_h1.items():
    print(f"  {lab:8s}: {c}")

print("\nHead 2 (t+2) category counts:")
for lab, c in counts_h2.items():
    print(f"  {lab:8s}: {c}")

# -------------------------------
# 4. Build joint centroid matrix + labels
# -------------------------------

joint_labels = []
joint_matrix = []

# Head 1 centroids, labeled with _h1_t+1
for lab, vec in centroids_h1.items():
    joint_labels.append(f"{lab}_h1_t+1")
    joint_matrix.append(vec)

# Head 2 centroids, labeled with _h2_t+2
for lab, vec in centroids_h2.items():
    joint_labels.append(f"{lab}_h2_t+2")
    joint_matrix.append(vec)

joint_matrix = np.stack(joint_matrix)  # [C, D]
print("\nJoint centroid matrix shape:", joint_matrix.shape)

# -------------------------------
# 5. Joint PCA on head1 + head2 centroids
# -------------------------------

pca_heads = PCA(n_components=2)
joint_2d = pca_heads.fit_transform(joint_matrix)

print("Explained variance ratio (PC1, PC2):", pca_heads.explained_variance_ratio_)

# -------------------------------
# 6. Joint scatter: compare head1 vs head2
# -------------------------------

plt.figure(figsize=(8, 6))

for (x, y), lab in zip(joint_2d, joint_labels):
    if "_h1_" in lab:
        color = "#1f77b4"  # blue
        marker = "o"
        head_name = "Head 1 (t+1)"
    else:
        color = "#2ca02c"  # green
        marker = "s"
        head_name = "Head 2 (t+2)"

    plt.scatter(x, y, c=color, marker=marker, s=90, edgecolors="k", linewidths=0.7)
    plt.text(x + 0.02, y + 0.02, lab, fontsize=9)

plt.title("Head-space centroids by future-token category\nHead 1 (t+1) vs Head 2 (t+2)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True, linewidth=0.5, alpha=0.5)
plt.axhline(0, color="gray", linewidth=0.5, alpha=0.5)
plt.axvline(0, color="gray", linewidth=0.5, alpha=0.5)

# Custom legend handles
import matplotlib.lines as mlines
head1_handle = mlines.Line2D([], [], color="#1f77b4", marker="o",
                             linewidth=0, markersize=8, label="Head 1 (t+1)")
head2_handle = mlines.Line2D([], [], color="#2ca02c", marker="s",
                             linewidth=0, markersize=8, label="Head 2 (t+2)")
plt.legend(handles=[head1_handle, head2_handle], loc="best")

plt.show()
```

based on the pc variance this represents roughly 4 different dimensions. one for each grouping

### Task 3 — Layer-Wise Gradient Importance for t+1 vs t+2

This analysis measures **how much each trunk layer contributes** to the losses of the two heads.

Steps:

1. Take one small batch of text from FineWeb and build:
   - `input_ids` (context tokens)
   - `labels_t1` (gold tokens at t+1)
   - `labels_t2` (gold tokens at t+2)

2. Run a custom forward pass that:
   - Applies the **shared trunk layers** (`model.layers[:-1]`).
   - Saves the output after each trunk layer and enables `retain_grad()` on them.
   - Applies Head 1 and Head 2 from the final trunk state to get `logits_t1` and `logits_t2`.

3. For **t+1**:
   - Compute cross-entropy loss between `logits_t1` and `labels_t1`.
   - Call `loss_t1.backward()`.
   - For each trunk layer ℓ, compute the mean gradient norm  
     \\(\|\partial \mathcal{L}_{t+1} / \partial h^{(\ell)}\|\\) averaged over batch and time.

4. Repeat for **t+2** with a fresh forward/backward pass, giving gradient norms for the second head.

5. Plot **layer index vs gradient norm** for both heads on the same chart.

If the t+2 curve peaks more in deeper layers or has a different shape from t+1, it suggests the t+2 head depends more on **deeper / more global** parts of the trunk, while t+1 may rely more on shallower or different layers.


```python
# ============================================================
# Task 3 — FIXED batch builder + rerun gradient analysis
#   (You can rerun this whole cell; it overwrites the old helpers.)
# ============================================================

import random
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
from datasets import load_dataset

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model.eval()

BLOCK_SIZE_ANALYSIS = 128   # trunk sequence length T
BATCH_SIZE_ANALYSIS = 4
tokenizer.model_max_length = BLOCK_SIZE_ANALYSIS + 2  # need T+2 tokens

print("Loading FineWeb (sample-10BT) for gradient analysis...")
fw_stream_grad = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

GRAD_SEED = 3030
random.seed(GRAD_SEED)

# -------------------------------
# 1. NEW: make_one_batch_from_stream with fixed length
# -------------------------------

def make_one_batch_from_stream(
    hf_stream,
    batch_size: int,
    block_size: int,
):
    """
    Build one batch where every example has the same length.

    We require block_size + 2 tokens so we can define:
      input_ids : positions 0 .. block_size-1        (length block_size)
      labels_t1 : positions 1 .. block_size          (length block_size)
      labels_t2 : positions 2 .. block_size+1        (length block_size)
    """

    input_batch = []
    labels1_batch = []
    labels2_batch = []

    it = iter(hf_stream)

    while len(input_batch) < batch_size:
        ex = next(it)
        text = ex.get("text", "")
        if not text:
            continue
        text = text.strip()
        if len(text) < 80:
            continue

        # Random window from this document
        window_size = random.randint(200, 500)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if not snippet:
            continue

        # Tokenize with explicit truncation to block_size+2 tokens
        enc = tokenizer(
            snippet,
            add_special_tokens=False,
            truncation=True,
            max_length=block_size + 2,
            return_tensors="pt",
        )
        ids_full = enc["input_ids"][0]  # [T_full]
        T_full = ids_full.shape[0]

        # Need at least block_size+2 tokens to form t, t+1, t+2 windows
        if T_full < block_size + 2:
            continue

        # Use exactly the first block_size+2 tokens
        ids_full = ids_full[: block_size + 2]  # length = block_size+2

        # Build aligned inputs/labels
        input_ids_trim = ids_full[:block_size]          # 0 .. block_size-1
        labels_t1 = ids_full[1 : block_size + 1]        # 1 .. block_size
        labels_t2 = ids_full[2 : block_size + 2]        # 2 .. block_size+1

        input_batch.append(input_ids_trim.unsqueeze(0))
        labels1_batch.append(labels_t1.unsqueeze(0))
        labels2_batch.append(labels_t2.unsqueeze(0))

    input_ids = torch.cat(input_batch, dim=0).to(DEVICE)   # [B, T]
    labels_t1 = torch.cat(labels1_batch, dim=0).to(DEVICE) # [B, T]
    labels_t2 = torch.cat(labels2_batch, dim=0).to(DEVICE) # [B, T]

    print(f"Constructed batch: input_ids shape = {input_ids.shape}")
    return input_ids, labels_t1, labels_t2


input_ids_batch, labels_t1_batch, labels_t2_batch = make_one_batch_from_stream(
    fw_stream_grad,
    batch_size=BATCH_SIZE_ANALYSIS,
    block_size=BLOCK_SIZE_ANALYSIS,
)

# -------------------------------
# 2. Forward helper that collects trunk layer outputs
# -------------------------------

def forward_with_trunk_layer_outputs(model, input_ids):
    """
    Run model up through the shared trunk and both heads,
    collecting trunk layer outputs with gradients enabled.
    """
    B, T = input_ids.shape
    device = input_ids.device

    h = model.tok_embeddings(input_ids)  # [B, T, D]

    freqs_cis = model.freqs_cis[:T].to(device=h.device)
    mask = model._build_causal_mask(T, device=h.device)

    layer_outputs = []

    # Shared trunk layers (all but last layer, which is head1)
    for layer in model.layers[:-1]:
        h = layer(h, freqs_cis=freqs_cis, mask=mask)
        h.retain_grad()
        layer_outputs.append(h)

    # Head 1 (t+1)
    h_head1 = model.layers[-1](h, freqs_cis=freqs_cis, mask=mask)
    h_head1 = model.norm(h_head1)
    logits_t1 = model.output(h_head1).float()

    # Head 2 (t+2)
    h_head2 = model.extra_heads[0](h, freqs_cis=freqs_cis, mask=mask)
    h_head2 = model.norm(h_head2)
    logits_t2 = model.output(h_head2).float()

    return logits_t1, logits_t2, layer_outputs

# -------------------------------
# 3. Gradient importance per layer
# -------------------------------

def gradient_importance_per_layer(
    model,
    input_ids,
    labels_t1,
    labels_t2,
):
    B, T = input_ids.shape
    num_trunk_layers = len(model.layers) - 1

    norms_t1 = [0.0 for _ in range(num_trunk_layers)]
    norms_t2 = [0.0 for _ in range(num_trunk_layers)]

    # ---- t+1 head ----
    model.zero_grad(set_to_none=True)
    logits_t1, _, layer_outputs = forward_with_trunk_layer_outputs(model, input_ids)

    loss_t1 = F.cross_entropy(
        logits_t1.view(B * T, -1),
        labels_t1.view(B * T),
        reduction="mean",
    )
    loss_t1.backward()

    for idx, h_out in enumerate(layer_outputs):
        grad = h_out.grad        # [B, T, D]
        norms_t1[idx] = grad.norm(dim=-1).mean().item()

    # ---- t+2 head ----
    model.zero_grad(set_to_none=True)
    _, logits_t2, layer_outputs = forward_with_trunk_layer_outputs(model, input_ids)

    loss_t2 = F.cross_entropy(
        logits_t2.view(B * T, -1),
        labels_t2.view(B * T),
        reduction="mean",
    )
    loss_t2.backward()

    for idx, h_out in enumerate(layer_outputs):
        grad = h_out.grad
        norms_t2[idx] = grad.norm(dim=-1).mean().item()

    return norms_t1, norms_t2


norms_t1, norms_t2 = gradient_importance_per_layer(
    model,
    input_ids_batch,
    labels_t1_batch,
    labels_t2_batch,
)

print("Gradient norms per trunk layer (t+1):", norms_t1)
print("Gradient norms per trunk layer (t+2):", norms_t2)

# -------------------------------
# 4. Plot layer index vs gradient norm
# -------------------------------

num_trunk_layers = len(model.layers) - 1
layers = list(range(num_trunk_layers))

plt.figure(figsize=(8, 5))
plt.plot(layers, norms_t1, marker="o", label="t+1 head (Head 1)")
plt.plot(layers, norms_t2, marker="s", label="t+2 head (Head 2)")

plt.title("Layer-wise gradient importance for t+1 vs t+2 losses\n(shared trunk layers only)")
plt.xlabel("Trunk layer index (0 = shallow, higher = deeper)")
plt.ylabel("Mean gradient norm")
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```

Slightly more importance across all layers for the head one prediction task throughout

### Task 4 (v2) — True Attention Patterns for a Prompt

The previous “similarity-based attention” collapsed onto the last token because it only compared raw hidden states.  
Here we instead reconstruct **real attention weights** from the model’s own attention modules:

1. Take a hand-written prompt and run it through the model.
2. For a chosen trunk layer ℓ:
   - Use that layer’s `wq` and `wk` projections to compute **queries** and **keys** from the current hidden states.
   - Apply the same **RoPE** frequencies and **causal mask** that the model uses.
   - Compute attention scores \(QK^\top / \sqrt{d}\) and softmax to get genuine attention weights \([H, T, T]\).
3. For the **last token** position:
   - Extract per-head attention vectors over all previous tokens.
   - Plot a **heatmap** (heads × positions) and compute an **attention-distance** metric:
     \[
     \mathbb{E}[\,\text{distance}\,] = \sum_i a_i \cdot (T-1-i)
     \]
     where \(a_i\) is the attention weight on position \(i\).

Heads that put mass on earlier positions (larger expected distance) are behaving more **long-range**, while heads that mostly attend to the last few tokens are more **local**.


```python
# ============================================================
# Task 4 — True attention visualization for a prompt
#   (uses the model's own Q/K projections + RoPE + mask)
# ============================================================

import math
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.eval()

# ------------------------------------------------------------
# 1. Helper: run up to a given trunk layer to get its input h
# ------------------------------------------------------------

def get_hidden_before_layer(model, input_ids, layer_index):
    """
    Run the model up through trunk layers, returning the hidden
    state *before* applying layer `layer_index`.

    layer_index is an index into model.layers (0-based).
    We assume trunk layers are model.layers[:-1].
    """
    B, T = input_ids.shape
    h = model.tok_embeddings(input_ids)  # [B, T, D]
    freqs_cis = model.freqs_cis[:T].to(h.device)
    mask = model._build_causal_mask(T, device=h.device)

    for idx, layer in enumerate(model.layers[:-1]):  # trunk only
        if idx == layer_index:
            # Return the state that this layer will see, along with freqs/mask
            return h, freqs_cis, mask
        h = layer(h, freqs_cis=freqs_cis, mask=mask)

    # If layer_index is the last trunk layer, we return after the loop
    # (i.e., its input is the output of the previous layer).
    return h, freqs_cis, mask

# ------------------------------------------------------------
# 2. Helper: compute real attention weights for a given layer
# ------------------------------------------------------------

def compute_attention_weights_for_layer(model, input_ids, layer_index):
    """
    For batch size 1:
      - Get the input hidden state h for trunk layer `layer_index`
      - Use that layer's Attention module (wq, wk) with RoPE + mask
        to compute real attention weights.

    Returns:
      attn_weights: [n_heads, T, T]  (B dimension removed)
      tokens      : list of token strings
    """
    B, T = input_ids.shape
    assert B == 1, "This helper assumes batch size 1 for visualization."

    # Hidden state *before* the chosen layer
    h, freqs_cis, mask = get_hidden_before_layer(model, input_ids, layer_index)
    h = h  # [1, T, D]

    layer = model.layers[layer_index]
    att_mod = layer.attention
    D = h.shape[-1]
    n_heads = model.params.n_heads
    n_kv_heads = getattr(model.params, "n_kv_heads", None) or n_heads
    head_dim = D // n_heads

    # --- Project to Q, K using the same weights as the model ---
    # Shape: [B, T, n_heads * head_dim] etc.
    q = att_mod.wq(h)
    k = att_mod.wk(h)

    # Reshape to [B, T, n_heads, head_dim]
    q = q.view(B, T, n_heads, head_dim)
    k = k.view(B, T, n_kv_heads, head_dim)

    # Apply RoPE (same helper used in the model)
    # apply_rotary_emb(q, k, freqs_cis)
    q, k = apply_rotary_emb(q, k, freqs_cis)  # [B, T, H, hd]

    # Move heads in front: [B, H, T, hd]
    q = q.transpose(1, 2)
    k = k.transpose(1, 2)

    # Scaled dot-product attention: [B, H, T, T]
    scores = torch.matmul(q, k.transpose(-2, -1)) / math.sqrt(head_dim)

    # Causal mask: mask is [1, 1, T, T], with -inf below diagonal
    scores = scores + mask

    attn_weights = F.softmax(scores, dim=-1)  # [B, H, T, T]

    # Remove batch dimension -> [H, T, T]
    attn_weights = attn_weights[0].detach().cpu().numpy()

    # Tokens for x-axis
    tokens = tokenizer.convert_ids_to_tokens(input_ids[0].tolist())
    return attn_weights, tokens

# ------------------------------------------------------------
# 3. Choose prompt and layer to visualize
# ------------------------------------------------------------

prompt = (
    "The knight walked slowly into the ancient castle, "
    "listening carefully for any hidden danger."
)

enc = tokenizer(prompt, add_special_tokens=False, return_tensors="pt")
input_ids_prompt = enc["input_ids"].to(DEVICE)  # [1, T]
B, T = input_ids_prompt.shape

tokens_prompt = tokenizer.convert_ids_to_tokens(input_ids_prompt[0].tolist())
print(f"Prompt length T = {T}")
print("Tokens:", tokens_prompt)

num_trunk_layers = len(model.layers) - 1  # trunk only
LAYER_TO_VISUALIZE = num_trunk_layers - 1  # deepest shared trunk layer

print(f"Visualizing attention at trunk layer index {LAYER_TO_VISUALIZE} (0-based)")

# Compute attention weights
attn_weights, tokens = compute_attention_weights_for_layer(
    model, input_ids_prompt, LAYER_TO_VISUALIZE
)  # [H, T, T]

n_heads = attn_weights.shape[0]
positions = np.arange(T)
last_pos = T - 1

# ------------------------------------------------------------
# 4. Extract per-head attention from the last token
# ------------------------------------------------------------

# attn_last[h, i] = attention of head h FROM last token TO position i
attn_last = attn_weights[:, last_pos, :]  # [H, T]

# Expected distance from last token for each head
distances = np.abs(positions - last_pos)  # [T]
attention_distance = (attn_last * distances[None, :]).sum(axis=1)  # [H]

print("Per-head expected attention distance from last token:")
for h_idx, dist in enumerate(attention_distance):
    print(f"  Head {h_idx:2d}: {dist:.2f} tokens back (expected)")

# ------------------------------------------------------------
# 5. Heatmap of attention for a subset of heads
# ------------------------------------------------------------

NUM_HEADS_TO_SHOW = n_heads
heads_to_show = list(range(NUM_HEADS_TO_SHOW))

fig, ax = plt.subplots(figsize=(10, 4))
attn_subset = attn_last[heads_to_show, :]  # [H_show, T]

im = ax.imshow(
    attn_subset,
    aspect="auto",
    interpolation="nearest",
    cmap="viridis",
)

ax.set_title(
    f"True attention from last token at layer {LAYER_TO_VISUALIZE}\n"
    "(rows = heads, columns = token positions)"
)
ax.set_xlabel("Token position")
ax.set_ylabel("Head index")

ax.set_xticks(range(T))
ax.set_xticklabels(tokens, rotation=90, fontsize=8)
ax.set_yticks(range(NUM_HEADS_TO_SHOW))
ax.set_yticklabels(heads_to_show)

fig.colorbar(im, ax=ax, label="Attention weight")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 6. Bar plot of per-head attention distance
# ------------------------------------------------------------

plt.figure(figsize=(8, 4))
plt.bar(range(n_heads), attention_distance)
plt.xlabel("Head index")
plt.ylabel("Expected distance from last token (in tokens)")
plt.title(
    f"Per-head attention distance at layer {LAYER_TO_VISUALIZE}\n"
    "larger = more long-range focus"
)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

This visualization shows how different attention heads in the model place weighted focus on different parts of the input sentence when producing the final token. Each head specializes slightly differently, with some focusing locally on nearby words and others attending to earlier or semantically important parts of the sentence.

### Task 5 — Logit-Lens Style Token Preferences for Head 1 vs Head 2

Here we ask: **“What kinds of tokens do the t+1 head and t+2 head most like to predict, across many contexts?”**

Procedure:

1. Sample many short sequences from FineWeb and run them through the **trunk**.
2. For each position, pass the trunk states through:
   - **Head 1 (t+1)** and  
   - **Head 2 (t+2)**  
   to get two sets of logits.
3. For each head, collect the **top-k predicted tokens** at every position and count how often each token appears.
4. Map these frequently predicted token IDs into **embedding space** using `model.tok_embeddings.weight`.
5. Run **PCA** on the union of these token embeddings and plot:
   - tokens mostly preferred by Head 1 in one color,
   - tokens mostly preferred by Head 2 in another.
   (Point size encodes how often the token appeared in that head’s top-k set.)
6. Also print the most common tokens for each head.

If Head 2’s cloud is dominated by function words (“the, and, of…”) and Head 1’s cloud is more lexically diverse/content-heavy, that’s evidence the two heads play different roles in prediction.


```python
# ============================================================
# Task 5 — Logit-lens style token preference per head
#   (FIXED: cast bfloat16 embeddings to float32 for NumPy/PCA)
# ============================================================

import random
from collections import Counter

import torch
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
from datasets import load_dataset
from sklearn.decomposition import PCA

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.eval()

# ------------------------------------------------------------
# Assumes these helpers already exist:
#   forward_trunk(model, input_ids)
#   forward_head_from_trunk(model, h_trunk, freqs_cis, mask, head_index)
# ------------------------------------------------------------

BLOCK_SIZE_LOGIT = 128
tokenizer.model_max_length = BLOCK_SIZE_LOGIT + 2
TOP_K = 5
MAX_POSITIONS = 5000
BATCH_SNIPPETS = 40

# ------------------------------------------------------------
# 1. FineWeb stream
# ------------------------------------------------------------

print("Loading FineWeb (sample-10BT) stream for logit-lens analysis...")
fw_stream_ll = load_dataset(
    "HuggingFaceFW/fineweb",
    "sample-10BT",
    split="train",
    streaming=True,
)

LOGIT_SEED = 4242
random.seed(LOGIT_SEED)

def random_snippet_from_stream(hf_stream, min_chars=200, max_chars=600):
    it = iter(hf_stream)
    while True:
        ex = next(it)
        text = ex.get("text", "")
        if not text:
            continue
        text = text.strip()
        if len(text) < min_chars:
            continue

        window_size = random.randint(min_chars, max_chars)
        if len(text) <= window_size:
            snippet = text
        else:
            max_start = len(text) - window_size
            start = random.randint(0, max_start)
            snippet = text[start : start + window_size]

        snippet = snippet.strip()
        if snippet:
            return snippet

# ------------------------------------------------------------
# 2. Collect top-k token predictions from head1/head2
# ------------------------------------------------------------

head1_counts = Counter()
head2_counts = Counter()
positions_seen = 0

print("Collecting top-k predictions from head1 (t+1) and head2 (t+2)...")

for _ in range(BATCH_SNIPPETS):
    if positions_seen >= MAX_POSITIONS:
        break

    text = random_snippet_from_stream(fw_stream_ll, min_chars=200, max_chars=600)

    enc = tokenizer(
        text,
        add_special_tokens=False,
        truncation=True,
        max_length=BLOCK_SIZE_LOGIT + 2,
        return_tensors="pt",
    )
    input_ids_full = enc["input_ids"].to(DEVICE)
    _, T_full = input_ids_full.shape
    if T_full < 3:
        continue

    T = min(T_full - 2, BLOCK_SIZE_LOGIT)
    input_ids = input_ids_full[:, :T]

    with torch.no_grad():
        h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)
        h_head1, logits_t1 = forward_head_from_trunk(
            model, h_trunk, freqs_cis, mask, head_index=0
        )
        h_head2, logits_t2 = forward_head_from_trunk(
            model, h_trunk, freqs_cis, mask, head_index=1
        )

    B, T, V = logits_t1.shape
    logits_t1_np = logits_t1[0]
    logits_t2_np = logits_t2[0]

    for t in range(T):
        if positions_seen >= MAX_POSITIONS:
            break

        topk1 = torch.topk(logits_t1_np[t], k=TOP_K).indices.cpu().tolist()
        head1_counts.update(topk1)

        topk2 = torch.topk(logits_t2_np[t], k=TOP_K).indices.cpu().tolist()
        head2_counts.update(topk2)

        positions_seen += 1

print(f"Total positions sampled: {positions_seen}")
print(f"Unique tokens in head1 top-{TOP_K}: {len(head1_counts)}")
print(f"Unique tokens in head2 top-{TOP_K}: {len(head2_counts)}")

# ------------------------------------------------------------
# 3. Text summary: most frequent tokens per head
# ------------------------------------------------------------

def print_top_tokens(counter, name, n=25):
    print(f"\nMost frequent tokens in {name} top-{TOP_K} predictions:")
    for tok_id, cnt in counter.most_common(n):
        tok_str = tokenizer.decode([tok_id]).replace("\n", "\\n")
        print(f"  {tok_str!r:12s} (id={tok_id:5d}) : {cnt}")

print_top_tokens(head1_counts, "Head 1 (t+1)")
print_top_tokens(head2_counts, "Head 2 (t+2)")

# ------------------------------------------------------------
# 4. Build embedding clouds (FIXED: cast bfloat16 -> float32)
# ------------------------------------------------------------

# NOTE: model.tok_embeddings.weight is bfloat16; cast to float32
tok_emb = (
    model.tok_embeddings.weight
    .detach()
    .to(torch.float32)      # <-- key fix
    .cpu()
    .numpy()
)  # [V, D_emb]
D_emb = tok_emb.shape[1]

all_token_ids = sorted(set(head1_counts.keys()) | set(head2_counts.keys()))
print(f"\nTotal distinct tokens in union: {len(all_token_ids)}")

labels = []
emb_matrix = []
sizes = []

for tok_id in all_token_ids:
    c1 = head1_counts[tok_id]
    c2 = head2_counts[tok_id]
    if c1 > c2:
        label = "h1"
    elif c2 > c1:
        label = "h2"
    else:
        label = "both"

    labels.append(label)
    emb_matrix.append(tok_emb[tok_id])
    sizes.append(20 + 5 * max(c1, c2))

emb_matrix = np.stack(emb_matrix, axis=0)
sizes = np.array(sizes)

# ------------------------------------------------------------
# 5. PCA over token embeddings for preferred tokens
# ------------------------------------------------------------

pca = PCA(n_components=2)
emb_2d = pca.fit_transform(emb_matrix)
print("Explained variance ratio (PC1, PC2):", pca.explained_variance_ratio_)

labels_arr = np.array(labels)

h1_x = emb_2d[labels_arr == "h1"]
h1_s = sizes[labels_arr == "h1"]

h2_x = emb_2d[labels_arr == "h2"]
h2_s = sizes[labels_arr == "h2"]

both_x = emb_2d[labels_arr == "both"]
both_s = sizes[labels_arr == "both"]

# ------------------------------------------------------------
# 6. Scatter plot: head1 vs head2 token preference clouds
# ------------------------------------------------------------

plt.figure(figsize=(9, 7))

if len(h1_x) > 0:
    plt.scatter(
        h1_x[:, 0],
        h1_x[:, 1],
        s=h1_s,
        alpha=0.6,
        label="Tokens favored by Head 1 (t+1)",
    )

if len(h2_x) > 0:
    plt.scatter(
        h2_x[:, 0],
        h2_x[:, 1],
        s=h2_s,
        alpha=0.6,
        label="Tokens favored by Head 2 (t+2)",
    )

if len(both_x) > 0:
    plt.scatter(
        both_x[:, 0],
        both_x[:, 1],
        s=both_s,
        alpha=0.3,
        label="Tokens shared between heads",
        edgecolors="none",
    )

plt.title(
    "Logit-lens token preferences in embedding space\n"
    "Head 1 (t+1) vs Head 2 (t+2)"
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. Show tokens strongly favored by one head over the other
# ------------------------------------------------------------

def show_head_specific_examples(counter_a, counter_b, name_a, name_b, n=15):
    print(f"\nTokens much more common in {name_a} than {name_b}:")
    diffs = []
    all_ids = set(counter_a.keys()) | set(counter_b.keys())
    for tok_id in all_ids:
        ca = counter_a[tok_id]
        cb = counter_b[tok_id]
        if ca + cb == 0:
            continue
        diffs.append((ca - cb, tok_id))
    diffs.sort(reverse=True)
    for diff, tok_id in diffs[:n]:
        tok_str = tokenizer.decode([tok_id]).replace("\n", "\\n")
        print(
            f"  {tok_str!r:12s} (id={tok_id:5d}) : Δcount={diff:+d}  "
            f"(a={counter_a[tok_id]}, b={counter_b[tok_id]})"
        )

show_head_specific_examples(head1_counts, head2_counts, "Head 1 (t+1)", "Head 2 (t+2)")
show_head_specific_examples(head2_counts, head1_counts, "Head 2 (t+2)", "Head 1 (t+1)")
```

### Interpretation of Logit-Lens Token Preference Analysis

This analysis examines which tokens each prediction head tends to favor
across many contexts. We collect the top-5 predicted tokens from the t+1 and
t+2 heads over ~5,000 positions and visualize their embeddings with PCA.

**Key findings:**

1. **Head 1 (t+1) favors a wide range of lexical content tokens.**
   It frequently predicts pronouns (“I”, “it”, “he”, “you”), verbs and
   morphological endings (“ED”, “ING”), and other sentence-specific continuations.
   This reflects a head focused on *immediate, contentful next-token prediction*.

2. **Head 2 (t+2) strongly favors high-frequency structural tokens.**
   Its most common predictions are punctuation (comma, period, colon), newline
   characters, dashes, and function words (“the”, “and”, “of”, “is”, “to”).
   This indicates that the t+2 head behaves more like a *language-structure
   prior*, predicting general scaffolding before fine lexical detail.

3. **PCA visualization**
   The PCA plot shows overlapping token clouds because only ~0.6% of variance
   is preserved when projecting 1024-dimensional embeddings into 2D. However,
   the *sizes and densities* of the clusters still reflect the underlying
   frequency difference: head2’s high-frequency function-word cluster appears
   as many large orange points near the center.

Overall, this analysis confirms that the two heads have distinct roles:
**Head 1 produces specific lexical predictions**, while **Head 2 favors
structural, high-frequency tokens**, reflecting broader uncertainty two steps
ahead.


### Task 6 — Trajectory of Trunk Hidden States for a Single Sentence

Here we want to see the **path a sentence walks through the trunk’s hidden space**.

Steps:

1. Fit (or reuse) a PCA model on a bunch of trunk hidden states so we have a
   2-D projection of the trunk representation space.
2. Pick a human-readable prompt, tokenize it, and run it through the trunk to
   obtain the hidden state at each position \(h_t\) (before any prediction head).
3. Project each \(h_t\) into the PCA space and plot them **in order**, connecting
   the points and labeling each with its token.

The resulting trajectory plot is a geometric “story” of how the model’s internal
representation evolves as it reads each token in the sentence.


```python
# ============================================================
# Task 6 — Trajectory of trunk hidden states for one sentence
# ============================================================

import random
import numpy as np
import matplotlib.pyplot as plt
from datasets import load_dataset
from sklearn.decomposition import PCA
import torch

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.eval()

# ------------------------------------------------------------
# 1. Ensure we have a PCA fitted on trunk states
#    - If `pca_trunk` already exists from previous cells, reuse it.
#    - Otherwise, we sample some trunk states now and fit a new PCA.
# ------------------------------------------------------------

NEED_PCA = "pca_trunk" not in globals()

if NEED_PCA:
    print("No existing `pca_trunk` found; fitting PCA on trunk states now...")

    BLOCK_SIZE_PCA = 128
    tokenizer.model_max_length = BLOCK_SIZE_PCA + 2

    # Streaming FineWeb for trunk samples
    fw_stream_traj = load_dataset(
        "HuggingFaceFW/fineweb",
        "sample-10BT",
        split="train",
        streaming=True,
    )

    def random_snippet_from_stream(hf_stream, min_chars=200, max_chars=600):
        it = iter(hf_stream)
        while True:
            ex = next(it)
            text = ex.get("text", "")
            if not text:
                continue
            text = text.strip()
            if len(text) < min_chars:
                continue

            window_size = random.randint(min_chars, max_chars)
            if len(text) <= window_size:
                snippet = text
            else:
                max_start = len(text) - window_size
                start = random.randint(0, max_start)
                snippet = text[start : start + window_size]

            snippet = snippet.strip()
            if snippet:
                return snippet

    MAX_TRUNK_VECS = 4000   # how many h_t vectors to collect for PCA
    all_trunk_vecs = []

    TRUNK_SEED = 777
    random.seed(TRUNK_SEED)

    while len(all_trunk_vecs) < MAX_TRUNK_VECS:
        text = random_snippet_from_stream(fw_stream_traj, min_chars=200, max_chars=600)

        enc = tokenizer(
            text,
            add_special_tokens=False,
            truncation=True,
            max_length=BLOCK_SIZE_PCA,
            return_tensors="pt",
        )
        input_ids = enc["input_ids"].to(DEVICE)  # [1, T]
        _, T = input_ids.shape
        if T < 3:
            continue

        with torch.no_grad():
            h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)  # [1, T, D]

        h_np = h_trunk[0].float().cpu().numpy()  # [T, D]

        # Optionally subsample positions if sequence is long
        if T > 64:
            idx = np.linspace(0, T - 1, 64).astype(int)
            h_np = h_np[idx]

        all_trunk_vecs.append(h_np)
        if len(all_trunk_vecs) * 64 >= MAX_TRUNK_VECS:
            break

    trunk_matrix = np.concatenate(all_trunk_vecs, axis=0)  # [N, D]
    print(f"Collected trunk matrix for PCA: {trunk_matrix.shape}")

    pca_trunk = PCA(n_components=2)
    pca_trunk.fit(trunk_matrix)
    print("Fitted pca_trunk on trunk hidden states.")
else:
    print("Reusing existing `pca_trunk` fitted earlier.")


# ------------------------------------------------------------
# 2. Choose a readable prompt and run it through the trunk
# ------------------------------------------------------------

PROMPT_TEXT = (
    "The knight walked slowly into the ancient castle, listening carefully "
    "for any hidden danger."
)

enc = tokenizer(
    PROMPT_TEXT,
    add_special_tokens=False,
    truncation=True,
    max_length=model.params.max_seq_len,
    return_tensors="pt",
)
input_ids = enc["input_ids"].to(DEVICE)  # [1, T]
T = input_ids.shape[1]
print(f"Prompt length T = {T}")
tokens = tokenizer.convert_ids_to_tokens(input_ids[0].cpu().tolist())

with torch.no_grad():
    h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)  # [1, T, D]

h_seq = h_trunk[0].float().cpu().numpy()  # [T, D]

# ------------------------------------------------------------
# 3. Project the per-token trunk states with the PCA from above
# ------------------------------------------------------------

h_2d = pca_trunk.transform(h_seq)  # [T, 2]

# ------------------------------------------------------------
# 4. Plot the trajectory: one point per token, connected in order
# ------------------------------------------------------------

plt.figure(figsize=(10, 7))

# Draw path
plt.plot(h_2d[:, 0], h_2d[:, 1], marker="o", linestyle="-", alpha=0.7)

# Label each point with the corresponding token (may get crowded if T is large)
for i, (x, y) in enumerate(h_2d):
    tok = tokens[i]
    # Use a small offset so text doesn't sit exactly on the marker
    plt.text(x + 0.05, y + 0.05, tok, fontsize=8)

plt.title(
    "Trajectory of trunk hidden states for one sentence\n"
    "(points in order of tokens; labels are tokenizer pieces)"
)
plt.xlabel("PC1 (trunk space)")
plt.ylabel("PC2 (trunk space)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Task 7 — Trajectory of Predicted Tokens per Head in Embedding Space

Here we want to see **the path of tokens the model “imagines” next**, separately for
the t+1 and t+2 heads.

For a single sentence:

1. We get the trunk state \(h_t\) at each position.
2. We run **Head 1 (t+1)** and **Head 2 (t+2)** to get logits at every position and
   take the **argmax predicted token** for each head.
3. We map those predicted token IDs into the **input embedding space**, then
   project them into 2-D using a PCA fitted on the embedding matrix.
4. We plot two trajectories:
   - blue path: tokens predicted by the t+1 head,
   - orange path: tokens predicted by the t+2 head.

Comparing these paths lets us see whether the t+2 head tends to move through a
more generic / function-word region of lexical space, while t+1 tracks richer,
more contentful predictions.


```python
# ============================================================
# Task 7 — Trajectory of predicted tokens per head in
#           embedding space (t+1 vs t+2 paths)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import torch
import torch.nn.functional as F

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.eval()

# ------------------------------------------------------------
# 1. Use the same sentence as in Task 6
# ------------------------------------------------------------

PROMPT_TEXT = (
    "The knight walked slowly into the ancient castle, listening carefully "
    "for any hidden danger."
)

enc = tokenizer(
    PROMPT_TEXT,
    add_special_tokens=False,
    truncation=True,
    max_length=model.params.max_seq_len,
    return_tensors="pt",
)
input_ids = enc["input_ids"].to(DEVICE)   # [1, T]
T = input_ids.shape[1]
tokens_input = tokenizer.convert_ids_to_tokens(input_ids[0].cpu().tolist())
print(f"Prompt length T = {T}")
print("Input tokens:", tokens_input)

with torch.no_grad():
    # Shared trunk features
    h_trunk, freqs_cis, mask = forward_trunk(model, input_ids)          # [1, T, D]
    # Head 1 (t+1)
    h_head1, logits_t1 = forward_head_from_trunk(
        model, h_trunk, freqs_cis, mask, head_index=0
    )                                                                   # logits: [1, T, V]
    # Head 2 (t+2)
    h_head2, logits_t2 = forward_head_from_trunk(
        model, h_trunk, freqs_cis, mask, head_index=1
    )                                                                   # logits: [1, T, V]

# Convert logits to CPU numpy for convenience
logits_t1_np = logits_t1[0].float().cpu().numpy()   # [T, V]
logits_t2_np = logits_t2[0].float().cpu().numpy()   # [T, V]

# ------------------------------------------------------------
# 2. For each position, take argmax predicted token for each head
# ------------------------------------------------------------

pred_ids_h1 = logits_t1[0].argmax(dim=-1).cpu().tolist()  # [T]
pred_ids_h2 = logits_t2[0].argmax(dim=-1).cpu().tolist()  # [T]

pred_tokens_h1 = tokenizer.convert_ids_to_tokens(pred_ids_h1)
pred_tokens_h2 = tokenizer.convert_ids_to_tokens(pred_ids_h2)

print("\nPredicted tokens per position:")
for t in range(T):
    print(
        f"t={t:2d}, input={tokens_input[t]!r:12s}  "
        f"t+1 head -> {pred_tokens_h1[t]!r:10s}  "
        f"t+2 head -> {pred_tokens_h2[t]!r:10s}"
    )

# ------------------------------------------------------------
# 3. Build an embedding PCA (or reuse existing one)
#    - We project token embeddings (not trunk states).
# ------------------------------------------------------------

# Get embedding matrix as float32
tok_emb = (
    model.tok_embeddings.weight
    .detach()
    .to(torch.float32)
    .cpu()
    .numpy()
)  # [V, D_emb]
D_emb = tok_emb.shape[1]
print(f"\nEmbedding matrix shape: {tok_emb.shape}")

if "pca_emb" in globals():
    print("Using existing `pca_emb` for embedding space.")
    pca_emb = pca_emb
else:
    print("Fitting new `pca_emb` on the whole embedding matrix...")
    pca_emb = PCA(n_components=2)
    pca_emb.fit(tok_emb)
    print("Explained variance ratio (PC1, PC2):",
          pca_emb.explained_variance_ratio_)

# ------------------------------------------------------------
# 4. Map predicted token IDs to embeddings and project to 2-D
# ------------------------------------------------------------

path_emb_h1 = tok_emb[pred_ids_h1]   # [T, D_emb]
path_emb_h2 = tok_emb[pred_ids_h2]   # [T, D_emb]

path_2d_h1 = pca_emb.transform(path_emb_h1)  # [T, 2]
path_2d_h2 = pca_emb.transform(path_emb_h2)  # [T, 2]

# ------------------------------------------------------------
# 5. Plot the two trajectories in embedding space
# ------------------------------------------------------------

plt.figure(figsize=(11, 8))

# Head 1 trajectory (t+1)
plt.plot(
    path_2d_h1[:, 0],
    path_2d_h1[:, 1],
    marker="o",
    linestyle="-",
    alpha=0.8,
    label="Head 1 (t+1 predictions)",
)

# Head 2 trajectory (t+2)
plt.plot(
    path_2d_h2[:, 0],
    path_2d_h2[:, 1],
    marker="x",
    linestyle="--",
    alpha=0.8,
    label="Head 2 (t+2 predictions)",
)

# Label each point with the predicted token (you can comment this out if cluttered)
for t in range(T):
    x1, y1 = path_2d_h1[t]
    x2, y2 = path_2d_h2[t]
    plt.text(x1 + 0.03, y1 + 0.03, pred_tokens_h1[t], fontsize=7, alpha=0.8)
    plt.text(x2 + 0.03, y2 + 0.03, pred_tokens_h2[t], fontsize=7, alpha=0.8, color="darkorange")

plt.title(
    "Predicted-token trajectories in embedding space\n"
    "Head 1 (t+1) vs Head 2 (t+2)"
)
plt.xlabel("PC1 (embedding space)")
plt.ylabel("PC2 (embedding space)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

This plot shows the predicted-token trajectories produced by the t+1 and t+2 heads when fed the same input sentence. Each predicted token is mapped into the model’s embedding space and then projected to 2-D using PCA.

The t+1 head (blue) produces more content-rich, varied guesses (e.g., “first”, “air”, “world”, “reason”). These tokens occupy more dispersed regions of embedding space, so the blue trajectory moves broadly.

In contrast, the t+2 head (orange) strongly favors generic function words such as “the”, “of”, “,”, and newline. These live in a tight region of embedding space, so the orange trajectory stays compact.

This confirms a functional difference between the heads:  
the immediate next-token head attempts meaningful continuation, while the two-step head defaults to high-frequency scaffolding words that are useful across many contexts.
